```python
import json
import matplotlib.pyplot as plt

file_path = "kbg_coverage_evaluated_dataset.jsonl"

scores = []

with open(file_path, "r", encoding="utf-8") as f:
    for line in f:
        if line.strip():
            data = json.loads(line)
            scores.append(data["coverage_evaluation"]["average_coverage_score"])

print("Number of entries:", len(scores))
print("Scores:", scores)
len(scores)
```

```python
plt.figure(figsize=(7,4))

plt.hist(scores, bins=10)

plt.xlabel("Average Coverage Score")
plt.ylabel("Number of Reports")
plt.title("Distribution of Average Coverage Scores")

plt.show()
```

```python
import json
import matplotlib.pyplot as plt
import numpy as np

file_path = "kbg_coverage_evaluated_dataset.jsonl"

q1_scores = []
q2_scores = []
q3_scores = []

with open(file_path, "r", encoding="utf-8") as f:
    for line in f:
        if line.strip():
            data = json.loads(line)
            evaluation = data["coverage_evaluation"]
            
            q1_scores.append(evaluation["question1"]["score"])
            q2_scores.append(evaluation["question2"]["score"])
            q3_scores.append(evaluation["question3"]["score"])

print("Q1 mean:", np.mean(q1_scores))
print("Q2 mean:", np.mean(q2_scores))
print("Q3 mean:", np.mean(q3_scores))
```
