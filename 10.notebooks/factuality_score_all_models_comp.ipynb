```python
import json
from pathlib import Path

# Folder containing all files
folder = Path(
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation\Factuality_temp\Organising_file"
)

# Process parts 01 to 06
for part in range(1, 7):

    part_str = f"{part:02d}"

    weighted_file = folder / (
        f"factuality_scores_weighted_sft_qwen25_1.5b_part2_50_{part_str}.jsonl"
    )

    normal_file = folder / (
        f"factuality_scores_[normal_sft_qwen25_1.5b]_part2_100_{part_str}.jsonl"
    )

    # Check that both files exist
    if not weighted_file.exists():
        print(f"❌ Weighted file not found: {weighted_file.name}")
        continue

    if not normal_file.exists():
        print(f"❌ Normal file not found: {normal_file.name}")
        continue

    # ---------------------------------------------------------
    # 1. Read indices from Weighted SFT file
    # ---------------------------------------------------------
    weighted_indices = set()

    with open(weighted_file, "r", encoding="utf-8") as f:
        for line in f:
            if line.strip():
                record = json.loads(line)
                weighted_indices.add(record["index"])

    # ---------------------------------------------------------
    # 2. Read Normal SFT file
    # ---------------------------------------------------------
    normal_records = []

    with open(normal_file, "r", encoding="utf-8") as f:
        for line in f:
            if line.strip():
                record = json.loads(line)

                # Keep only records whose index exists
                # in the corresponding Weighted file
                if record["index"] in weighted_indices:
                    normal_records.append(record)

    # ---------------------------------------------------------
    # 3. Backup original Normal file
    # ---------------------------------------------------------
    backup_file = normal_file.with_suffix(".original.jsonl")

    if not backup_file.exists():
        normal_file.rename(backup_file)

    # ---------------------------------------------------------
    # 4. Write filtered Normal file
    # ---------------------------------------------------------
    with open(normal_file, "w", encoding="utf-8") as f:
        for record in normal_records:
            f.write(json.dumps(record, ensure_ascii=False) + "\n")

    # ---------------------------------------------------------
    # 5. Report
    # ---------------------------------------------------------
    print(f"\nPart {part_str}")
    print(f"  Weighted indices : {len(weighted_indices)}")
    print(f"  Normal before    : {sum(1 for _ in open(backup_file, encoding='utf-8'))}")
    print(f"  Normal after     : {len(normal_records)}")

    if len(weighted_indices) == len(normal_records):
        print("  ✅ MATCH")
    else:
        print("  ⚠️ COUNT DOES NOT MATCH")

print("\nDone.")
```

```python
import json
from pathlib import Path

folder = Path(
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation\Factuality_temp\Organising_file"
)

for part in range(1, 7):

    p = f"{part:02d}"

    weighted_file = folder / (
        f"factuality_scores_weighted_sft_qwen25_1.5b_part2_50_{p}.jsonl"
    )

    normal_file = folder / (
        f"factuality_scores_[normal_sft_qwen25_1.5b]_part2_100_{p}.jsonl"
    )

    # -----------------------------
    # Read Weighted
    # -----------------------------
    with open(weighted_file, "r", encoding="utf-8") as f:
        weighted_records = [
            json.loads(line)
            for line in f
            if line.strip()
        ]

    # -----------------------------
    # Read Normal
    # -----------------------------
    with open(normal_file, "r", encoding="utf-8") as f:
        normal_records = [
            json.loads(line)
            for line in f
            if line.strip()
        ]

    # -----------------------------
    # Find intersection
    # -----------------------------
    weighted_indices = {
        record["index"] for record in weighted_records
    }

    normal_indices = {
        record["index"] for record in normal_records
    }

    common_indices = weighted_indices & normal_indices

    # -----------------------------
    # Keep only common indices
    # -----------------------------
    weighted_filtered = [
        record for record in weighted_records
        if record["index"] in common_indices
    ]

    normal_filtered = [
        record for record in normal_records
        if record["index"] in common_indices
    ]

    # -----------------------------
    # Backup originals
    # -----------------------------
    weighted_backup = weighted_file.with_suffix(".original.jsonl")
    normal_backup = normal_file.with_suffix(".original.jsonl")

    if not weighted_backup.exists():
        weighted_file.rename(weighted_backup)

    if not normal_backup.exists():
        normal_file.rename(normal_backup)

    # -----------------------------
    # Write filtered files
    # -----------------------------
    weighted_file = folder / (
        f"factuality_scores_weighted_sft_qwen25_1.5b_part2_50_{p}.jsonl"
    )

    normal_file = folder / (
        f"factuality_scores_[normal_sft_qwen25_1.5b]_part2_100_{p}.jsonl"
    )

    with open(weighted_file, "w", encoding="utf-8") as f:
        for record in weighted_filtered:
            f.write(json.dumps(record, ensure_ascii=False) + "\n")

    with open(normal_file, "w", encoding="utf-8") as f:
        for record in normal_filtered:
            f.write(json.dumps(record, ensure_ascii=False) + "\n")

    # -----------------------------
    # Report
    # -----------------------------
    print(f"\nPart {p}")
    print(f"Weighted before : {len(weighted_records)}")
    print(f"Normal before   : {len(normal_records)}")
    print(f"Intersection    : {len(common_indices)}")
    print(f"Weighted after  : {len(weighted_filtered)}")
    print(f"Normal after    : {len(normal_filtered)}")

    if len(weighted_filtered) == len(normal_filtered):
        print("✅ MATCH")
    else:
        print("❌ ERROR")


print("\nDone.")
```

```python
import glob
import json
import os
import re

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 1. FOLDER
# ============================================================

folder = r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation\Factuality_temp\Organising_file_1"


# ============================================================
# 2. FUNCTION TO READ JSONL
# ============================================================

def read_jsonl(file):

    records = []

    with open(file, "r", encoding="utf-8") as f:

        for line in f:

            if line.strip():

                records.append(
                    json.loads(line)
                )

    return records


# ============================================================
# 3. STORAGE
# ============================================================

records = []


# ============================================================
# 4. PROCESS SIX PARTS
# ============================================================

for part in range(1, 7):

    p = f"{part:02d}"

    # --------------------------------------------------------
    # Teacher
    # --------------------------------------------------------

    teacher_file = os.path.join(
        folder,
        f"factuality_scores_2_{p}.jsonl"
    )

    # --------------------------------------------------------
    # Baseline
    # --------------------------------------------------------

    baseline_file = os.path.join(
        folder,
        f"factuality_scores_[baseline_qwen25_15b]_part2_{p}.jsonl"
    )

    # --------------------------------------------------------
    # Normal SFT
    # --------------------------------------------------------

    normal_file = os.path.join(
        folder,
        f"factuality_scores_[normal_sft_qwen25_1.5b]_part2_100_{p}.jsonl"
    )

    # --------------------------------------------------------
    # Weighted SFT
    # --------------------------------------------------------

    weighted_file = os.path.join(
        folder,
        f"factuality_scores_weighted_sft_qwen25_1.5b_part2_50_{p}.jsonl"
    )


    files = {
        "Teacher": teacher_file,
        "Baseline": baseline_file,
        "Normal SFT": normal_file,
        "Weighted SFT": weighted_file
    }


    # --------------------------------------------------------
    # Check files
    # --------------------------------------------------------

    for model, file in files.items():

        if not os.path.exists(file):

            print(
                f"❌ Missing {model} file for Part {p}:"
            )

            print(file)

            raise FileNotFoundError(file)


    # --------------------------------------------------------
    # Read files
    # --------------------------------------------------------

    teacher = read_jsonl(teacher_file)
    baseline = read_jsonl(baseline_file)
    normal = read_jsonl(normal_file)
    weighted = read_jsonl(weighted_file)


    # --------------------------------------------------------
    # Create dictionaries indexed by report index
    # --------------------------------------------------------

    teacher_dict = {
        r["index"]: r
        for r in teacher
    }

    baseline_dict = {
        r["index"]: r
        for r in baseline
    }

    normal_dict = {
        r["index"]: r
        for r in normal
    }

    weighted_dict = {
        r["index"]: r
        for r in weighted
    }


    # --------------------------------------------------------
    # FOUR-WAY INTERSECTION
    # --------------------------------------------------------

    common_indices = (
        set(teacher_dict.keys())
        & set(baseline_dict.keys())
        & set(normal_dict.keys())
        & set(weighted_dict.keys())
    )


    print(
        f"Part {p}: "
        f"Teacher={len(teacher_dict)}, "
        f"Baseline={len(baseline_dict)}, "
        f"Normal={len(normal_dict)}, "
        f"Weighted={len(weighted_dict)}, "
        f"Common={len(common_indices)}"
    )


    # --------------------------------------------------------
    # Store only common cases
    # --------------------------------------------------------

    for index in common_indices:

        # Teacher
        teacher_scores = teacher_dict[index].get(
            "factuality_scores", {}
        )

        records.append({
            "part": p,
            "index": index,
            "case_id": f"{p}_{index}",
            "model": "Teacher",
            "average": teacher_scores.get(
                "average_factuality_score"
            )
        })


        # Baseline
        baseline_scores = baseline_dict[index].get(
            "factuality_scores", {}
        )

        records.append({
            "part": p,
            "index": index,
            "case_id": f"{p}_{index}",
            "model": "Baseline Student",
            "average": baseline_scores.get(
                "average_factuality_score"
            )
        })


        # Normal SFT
        normal_scores = normal_dict[index].get(
            "factuality_scores", {}
        )

        records.append({
            "part": p,
            "index": index,
            "case_id": f"{p}_{index}",
            "model": "Normal SFT Student",
            "average": normal_scores.get(
                "average_factuality_score"
            )
        })


        # Weighted SFT
        weighted_scores = weighted_dict[index].get(
            "factuality_scores", {}
        )

        records.append({
            "part": p,
            "index": index,
            "case_id": f"{p}_{index}",
            "model": "Weighted SFT Student",
            "average": weighted_scores.get(
                "average_factuality_score"
            )
        })


# ============================================================
# 5. DATAFRAME
# ============================================================

df = pd.DataFrame(records)


# ============================================================
# 6. CHECK NUMBER OF MATCHED CASES
# ============================================================

print("\n" + "=" * 70)
print("FINAL MATCHING")
print("=" * 70)

matched_cases = df["case_id"].nunique()

print(
    f"Matched cases across ALL FOUR models: {matched_cases}"
)

print("\nRecords per model:")

print(
    df.groupby("model")["case_id"]
      .nunique()
)


# ============================================================
# 7. CHECK FOR MISSING VALUES
# ============================================================

print("\nMissing factuality scores:")

print(
    df.groupby("model")["average"]
      .apply(lambda x: x.isna().sum())
)


# ============================================================
# 8. DESCRIPTIVE STATISTICS
# ============================================================

print("\n" + "=" * 70)
print("DESCRIPTIVE STATISTICS")
print("=" * 70)

stats = (
    df.groupby("model")["average"]
      .describe()
)

print(stats)


# ============================================================
# 9. KDE PLOT — NO HISTOGRAM
# ============================================================

plt.figure(figsize=(10, 6))


sns.kdeplot(
    data=df,
    x="average",
    hue="model",
    common_norm=False,
    fill=False,
    linewidth=2.5
)


plt.xlabel(
    "Average factuality score",
    fontsize=12
)

plt.ylabel(
    "Density",
    fontsize=12
)

plt.title(
    "Distribution of Average Factuality Scores",
    fontsize=14
)

plt.xlim(0, 1)

plt.tight_layout()

plt.show()
```

```python
import json
from pathlib import Path

# ============================================================
# BASE DIRECTORIES
# ============================================================

FACTUALITY_DIR = Path(
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation\Factuality_temp\Organising_file_1"
)

COVERAGE_DIR = Path(
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation\Factuality_temp\Organising_file_1_coverage"
)


# ============================================================
# FILE CONFIGURATION
# ============================================================

models = {
    "baseline": {
        "factuality_pattern": "factuality_scores_[baseline_qwen25_15b]_part2_{part:02d}.jsonl",
        "coverage_pattern": "coverage_scores_baseline_qwen25_15b_part2_{part:02d}.jsonl",
    },

    "normal_sft": {
        "factuality_pattern": "factuality_scores_[normal_sft_qwen25_1.5b]_part2_100_{part:02d}.jsonl",
        "coverage_pattern": "coverage_scores_normal_sft_qwen25_1.5b_part2_100_{part:02d}.jsonl",
    },

    "weighted_sft": {
        "factuality_pattern": "factuality_scores_[weighted_sft_qwen25_1.5b]_part2_50_{part:02d}.jsonl",
        "coverage_pattern": "coverage_scores_weighted_sft_qwen25_1.5b_part2_50_{part:02d}.jsonl",
    },

    "testing_2": {
        "factuality_pattern": "factuality_scores_[testing_2]_part2_{part:02d}.jsonl",
        "coverage_pattern": "coverage_testing_2_{part:02d}.jsonl",
    },
}


# ============================================================
# FUNCTION TO READ INDICES
# ============================================================

def get_indices(file_path):

    indices = []
    duplicates = set()
    seen = set()

    with open(file_path, "r", encoding="utf-8") as f:

        for line_number, line in enumerate(f, start=1):

            if not line.strip():
                continue

            try:
                data = json.loads(line)
            except json.JSONDecodeError:
                print(f"WARNING: Invalid JSON in {file_path.name}, line {line_number}")
                continue

            if "index" not in data:
                print(f"WARNING: No 'index' in {file_path.name}, line {line_number}")
                continue

            idx = data["index"]

            if idx in seen:
                duplicates.add(idx)

            seen.add(idx)
            indices.append(idx)

    return indices, duplicates


# ============================================================
# CHECK ONE PAIR
# ============================================================

def check_pair(factuality_file, coverage_file):

    fact_indices, fact_duplicates = get_indices(factuality_file)
    cov_indices, cov_duplicates = get_indices(coverage_file)

    fact_set = set(fact_indices)
    cov_set = set(cov_indices)

    matched = fact_set & cov_set
    only_factuality = fact_set - cov_set
    only_coverage = cov_set - fact_set

    return {
        "fact_count": len(fact_indices),
        "coverage_count": len(cov_indices),
        "matched": len(matched),
        "only_factuality": len(only_factuality),
        "only_coverage": len(only_coverage),
        "fact_duplicates": fact_duplicates,
        "coverage_duplicates": cov_duplicates,
        "only_factuality_indices": only_factuality,
        "only_coverage_indices": only_coverage,
        "matched_indices": matched,
    }


# ============================================================
# MAIN CHECK
# ============================================================

for model_name, patterns in models.items():

    print("\n" + "=" * 80)
    print(f"MODEL: {model_name.upper()}")
    print("=" * 80)

    total_fact = 0
    total_cov = 0
    total_matched = 0

    all_fact_indices = set()
    all_cov_indices = set()

    for part in range(1, 7):

        factuality_file = FACTUALITY_DIR / patterns["factuality_pattern"].format(part=part)
        coverage_file = COVERAGE_DIR / patterns["coverage_pattern"].format(part=part)

        print(f"\n--- PART {part:02d} ---")

        if not factuality_file.exists():
            print(f"❌ Missing factuality file:")
            print(f"   {factuality_file}")
            continue

        if not coverage_file.exists():
            print(f"❌ Missing coverage file:")
            print(f"   {coverage_file}")
            continue

        result = check_pair(factuality_file, coverage_file)

        print(f"Factuality indices : {result['fact_count']}")
        print(f"Coverage indices   : {result['coverage_count']}")
        print(f"Matched             : {result['matched']}")
        print(f"Only in factuality  : {result['only_factuality']}")
        print(f"Only in coverage    : {result['only_coverage']}")

        if result["fact_duplicates"]:
            print(
                f"⚠️ Duplicate factuality indices: "
                f"{len(result['fact_duplicates'])}"
            )

        if result["coverage_duplicates"]:
            print(
                f"⚠️ Duplicate coverage indices: "
                f"{len(result['coverage_duplicates'])}"
            )

        if result["only_factuality"]:
            print(
                "Indices only in factuality:",
                sorted(result["only_factuality_indices"])
            )

        if result["only_coverage"]:
            print(
                "Indices only in coverage:",
                sorted(result["only_coverage_indices"])
            )

        total_fact += result["fact_count"]
        total_cov += result["coverage_count"]
        total_matched += result["matched"]

        all_fact_indices.update(result["matched_indices"])
        all_cov_indices.update(result["matched_indices"])

    # ========================================================
    # OVERALL RESULT
    # ========================================================

    print("\n" + "-" * 80)
    print(f"OVERALL RESULT — {model_name.upper()}")
    print("-" * 80)

    print(f"Total factuality records : {total_fact}")
    print(f"Total coverage records   : {total_cov}")
    print(f"Total matched records    : {total_matched}")

    print(
        f"Unique matched indices across parts: "
        f"{len(all_fact_indices)}"
    )

    if total_matched == total_fact == total_cov:
        print("✅ PERFECT MATCH — all records correspond.")
    else:
        print("⚠️ MISMATCH EXISTS — inspect the results above.")
```

```python
import json
from pathlib import Path

# ============================================================
# DIRECTORIES
# ============================================================

FACTUALITY_DIR = Path(
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation\Factuality_temp\Organising_file_1"
)

COVERAGE_DIR = Path(
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation\Factuality_temp\Organising_file_1_coverage"
)


# ============================================================
# FILE CONFIGURATION
# ============================================================

file_pairs = {
    "baseline": {
        "reference": [
            f"factuality_scores_[baseline_qwen25_15b]_part2_{i:02d}.jsonl"
            for i in range(1, 7)
        ],
        "coverage": [
            f"coverage_scores_baseline_qwen25_15b_part2_{i:02d}.jsonl"
            for i in range(1, 7)
        ],
    },

    "normal_sft": {
        "reference": [
            f"factuality_scores_[normal_sft_qwen25_1.5b]_part2_100_{i:02d}.jsonl"
            for i in range(1, 7)
        ],
        "coverage": [
            f"coverage_scores_normal_sft_qwen25_1.5b_part2_100_{i:02d}.jsonl"
            for i in range(1, 7)
        ],
    },

    "weighted_sft": {
        "reference": [
            f"factuality_scores_weighted_sft_qwen25_1.5b_part2_50_{i:02d}.jsonl"
            for i in range(1, 7)
        ],
        "coverage": [
            f"coverage_scores_weighted_sft_qwen25_1.5b_part2_50_{i:02d}.jsonl"
            for i in range(1, 7)
        ],
    },

    "testing_2": {
        "reference": [
            f"factuality_scores_2_{i:02d}.jsonl"
            for i in range(1, 7)
        ],
        "coverage": [
            f"coverage_testing_2_{i:02d}.jsonl"
            for i in range(1, 7)
        ],
    },
}


# ============================================================
# READ JSONL
# ============================================================

def read_jsonl(file_path):

    records = []

    with open(file_path, "r", encoding="utf-8") as f:

        for line_number, line in enumerate(f, 1):

            if not line.strip():
                continue

            try:
                data = json.loads(line)
            except json.JSONDecodeError as e:
                print(
                    f"ERROR: Invalid JSON in {file_path.name}, "
                    f"line {line_number}: {e}"
                )
                raise

            records.append(data)

    return records


# ============================================================
# FILTER ONE COVERAGE FILE
# ============================================================

def filter_coverage(reference_file, coverage_file):

    # --------------------------------------------------------
    # Read reference factuality file
    # --------------------------------------------------------

    reference_records = read_jsonl(reference_file)

    reference_indices = {
        record["index"]
        for record in reference_records
        if "index" in record
    }

    # --------------------------------------------------------
    # Read coverage file
    # --------------------------------------------------------

    coverage_records = read_jsonl(coverage_file)

    original_count = len(coverage_records)

    # --------------------------------------------------------
    # Keep only records whose index is in reference
    # --------------------------------------------------------

    filtered_records = [
        record
        for record in coverage_records
        if record.get("index") in reference_indices
    ]

    filtered_indices = {
        record["index"]
        for record in filtered_records
        if "index" in record
    }

    # --------------------------------------------------------
    # Check for missing reference indices
    # --------------------------------------------------------

    missing_indices = reference_indices - filtered_indices

    extra_indices = filtered_indices - reference_indices

    # --------------------------------------------------------
    # Safety check
    # --------------------------------------------------------

    if missing_indices:
        raise RuntimeError(
            f"\nERROR: Some reference indices were not found "
            f"in coverage file:\n"
            f"{coverage_file.name}\n"
            f"Missing indices: {sorted(missing_indices)}"
        )

    if extra_indices:
        raise RuntimeError(
            f"\nERROR: Unexpected indices remain in:\n"
            f"{coverage_file.name}\n"
            f"Extra indices: {sorted(extra_indices)}"
        )

    # --------------------------------------------------------
    # Write filtered data back to ORIGINAL file
    # --------------------------------------------------------

    with open(coverage_file, "w", encoding="utf-8") as f:

        for record in filtered_records:
            f.write(
                json.dumps(
                    record,
                    ensure_ascii=False
                ) + "\n"
            )

    # --------------------------------------------------------
    # Return statistics
    # --------------------------------------------------------

    return {
        "reference_count": len(reference_records),
        "original_coverage_count": original_count,
        "filtered_coverage_count": len(filtered_records),
    }


# ============================================================
# PROCESS EVERYTHING
# ============================================================

print("\n")
print("=" * 90)
print("STARTING COVERAGE INTERSECTION")
print("=" * 90)


overall_results = {}


for model_name, files in file_pairs.items():

    print("\n")
    print("=" * 90)
    print(f"MODEL: {model_name.upper()}")
    print("=" * 90)

    model_results = []

    for part in range(6):

        reference_file = FACTUALITY_DIR / files["reference"][part]
        coverage_file = COVERAGE_DIR / files["coverage"][part]

        print(f"\nPART {part + 1:02d}")
        print("-" * 90)

        # ----------------------------------------------------
        # Check files exist
        # ----------------------------------------------------

        if not reference_file.exists():
            raise FileNotFoundError(
                f"Reference file not found:\n{reference_file}"
            )

        if not coverage_file.exists():
            raise FileNotFoundError(
                f"Coverage file not found:\n{coverage_file}"
            )

        print(f"Reference : {reference_file.name}")
        print(f"Coverage  : {coverage_file.name}")

        # ----------------------------------------------------
        # Filter
        # ----------------------------------------------------

        result = filter_coverage(
            reference_file,
            coverage_file
        )

        model_results.append(result)

        print(
            f"Reference records       : "
            f"{result['reference_count']}"
        )

        print(
            f"Coverage BEFORE         : "
            f"{result['original_coverage_count']}"
        )

        print(
            f"Coverage AFTER          : "
            f"{result['filtered_coverage_count']}"
        )

        print("✅ Intersection completed")

    overall_results[model_name] = model_results


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 90)
print("FINAL SUMMARY")
print("=" * 90)

for model_name, results in overall_results.items():

    total_reference = sum(
        r["reference_count"]
        for r in results
    )

    total_before = sum(
        r["original_coverage_count"]
        for r in results
    )

    total_after = sum(
        r["filtered_coverage_count"]
        for r in results
    )

    print(f"\n{model_name.upper()}")

    print(
        f"Reference total : {total_reference}"
    )

    print(
        f"Coverage before : {total_before}"
    )

    print(
        f"Coverage after  : {total_after}"
    )

    print(
        f"Removed         : "
        f"{total_before - total_after}"
    )

    if total_after == total_reference:
        print("✅ FINAL COUNTS MATCH")
    else:
        print("❌ COUNT MISMATCH")


print("\n")
print("=" * 90)
print("DONE")
print("=" * 90)
```

```python
import glob
import json
import pandas as pd
import numpy as np

from scipy.stats import friedmanchisquare, wilcoxon


# ============================================================
# PATH
# ============================================================

BASE_DIR = r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation\Factuality_temp\Organising_file_1_coverage"


# ============================================================
# FILES
# ============================================================

model_files = {
    "Baseline": "coverage_scores_baseline_qwen25_15b_part2_*.jsonl",

    "Normal SFT": "coverage_scores_normal_sft_qwen25_1.5b_part2_100_*.jsonl",

    "Weighted SFT": "coverage_scores_weighted_sft_qwen25_1.5b_part2_50_*.jsonl"
}


# ============================================================
# READ DATA
# ============================================================

records = []

for model_name, pattern in model_files.items():

    files = glob.glob(
        rf"{BASE_DIR}\{pattern}"
    )

    for filename in files:

        with open(filename, "r", encoding="utf-8") as f:

            for line in f:

                if not line.strip():
                    continue

                data = json.loads(line)

                scores = data.get("coverage_scores", {})

                records.append({
                    "model": model_name,
                    "index": data.get("index"),
                    "patient_coverage": scores.get(
                        "patient_coverage_score"
                    )
                })


df = pd.DataFrame(records)


# ============================================================
# PIVOT DATA
# ============================================================

pivot = df.pivot(
    index="index",
    columns="model",
    values="patient_coverage"
)


# Keep only complete cases
pivot = pivot.dropna(
    subset=[
        "Baseline",
        "Normal SFT",
        "Weighted SFT"
    ]
)


print("\nNumber of paired examples:")
print(len(pivot))

print("\nFirst few paired observations:")
print(pivot.head())


# ============================================================
# DESCRIPTIVE STATISTICS
# ============================================================

print("\n" + "=" * 70)
print("DESCRIPTIVE STATISTICS")
print("=" * 70)

for model in [
    "Baseline",
    "Normal SFT",
    "Weighted SFT"
]:

    scores = pivot[model]

    print(f"\n{model}")
    print(f"  N      = {len(scores)}")
    print(f"  Mean   = {scores.mean():.4f}")
    print(f"  Median = {scores.median():.4f}")
    print(f"  SD     = {scores.std():.4f}")
    print(f"  Min    = {scores.min():.4f}")
    print(f"  Max    = {scores.max():.4f}")


# ============================================================
# FRIEDMAN TEST
# ============================================================

baseline = pivot["Baseline"]
normal_sft = pivot["Normal SFT"]
weighted_sft = pivot["Weighted SFT"]


friedman_stat, friedman_p = friedmanchisquare(
    baseline,
    normal_sft,
    weighted_sft
)


print("\n" + "=" * 70)
print("FRIEDMAN TEST")
print("=" * 70)

print(f"Friedman statistic = {friedman_stat:.4f}")
print(f"p-value            = {friedman_p:.6f}")


if friedman_p < 0.05:
    print("Result: Significant overall difference (p < 0.05)")
else:
    print("Result: No significant overall difference (p >= 0.05)")


# ============================================================
# WILCOXON PAIRWISE TESTS
# ============================================================

comparisons = {
    "Baseline vs Normal SFT":
        (baseline, normal_sft),

    "Baseline vs Weighted SFT":
        (baseline, weighted_sft),

    "Normal SFT vs Weighted SFT":
        (normal_sft, weighted_sft)
}


raw_results = []


for comparison, (x, y) in comparisons.items():

    statistic, p_value = wilcoxon(
        x,
        y,
        alternative="two-sided"
    )

    raw_results.append({
        "comparison": comparison,
        "statistic": statistic,
        "raw_p": p_value
    })


# ============================================================
# HOLM CORRECTION
# ============================================================

# Sort by raw p-value
raw_results = sorted(
    raw_results,
    key=lambda x: x["raw_p"]
)

m = len(raw_results)

for i, result in enumerate(raw_results):

    corrected_p = (
        result["raw_p"] * (m - i)
    )

    result["holm_p"] = min(
        corrected_p,
        1.0
    )


# Ensure monotonicity of Holm-adjusted p-values
for i in range(1, m):

    raw_results[i]["holm_p"] = max(
        raw_results[i]["holm_p"],
        raw_results[i - 1]["holm_p"]
    )


# ============================================================
# EFFECT SIZE
# ============================================================

def rank_biserial_effect(x, y):

    differences = np.asarray(x) - np.asarray(y)

    differences = differences[
        differences != 0
    ]

    if len(differences) == 0:
        return 0.0

    ranks = pd.Series(
        np.abs(differences)
    ).rank()

    positive = ranks[
        differences > 0
    ].sum()

    negative = ranks[
        differences < 0
    ].sum()

    return (
        (positive - negative)
        / (positive + negative)
    )


# Calculate effect sizes
for result in raw_results:

    comparison = result["comparison"]

    x, y = comparisons[comparison]

    effect = rank_biserial_effect(
        x,
        y
    )

    result["rank_biserial"] = effect


# ============================================================
# PAIRWISE RESULTS
# ============================================================

results_df = pd.DataFrame(raw_results)


print("\n" + "=" * 70)
print("PAIRWISE WILCOXON TESTS")
print("=" * 70)

print(
    results_df[
        [
            "comparison",
            "statistic",
            "raw_p",
            "holm_p",
            "rank_biserial"
        ]
    ].to_string(index=False)
)


# ============================================================
# SIGNIFICANCE INTERPRETATION
# ============================================================

print("\n" + "=" * 70)
print("INTERPRETATION")
print("=" * 70)

for _, row in results_df.iterrows():

    p = row["holm_p"]
    effect = row["rank_biserial"]

    if p < 0.001:
        significance = "highly significant (p < 0.001)"
    elif p < 0.01:
        significance = "significant (p < 0.01)"
    elif p < 0.05:
        significance = "significant (p < 0.05)"
    else:
        significance = "not significant (p >= 0.05)"

    print(
        f"{row['comparison']}: "
        f"{significance}, "
        f"rank-biserial effect = {effect:.4f}"
    )
```

```python
import glob
import json
import pandas as pd
import numpy as np

from scipy.stats import friedmanchisquare, wilcoxon


# ============================================================
# PATH
# ============================================================

BASE_DIR = (
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation"
    r"\Factuality_temp\Organising_file_1_coverage"
)


# ============================================================
# FILE PATTERNS
# ============================================================

model_files = {
    "Baseline":
        "coverage_scores_baseline_qwen25_15b_part2_*.jsonl",

    "Normal SFT":
        "coverage_scores_normal_sft_qwen25_1.5b_part2_100_*.jsonl",

    "Weighted SFT":
        "coverage_scores_weighted_sft_qwen25_1.5b_part2_50_*.jsonl"
}


# ============================================================
# READ DATA
# ============================================================

records = []

for model_name, pattern in model_files.items():

    files = glob.glob(
        rf"{BASE_DIR}\{pattern}"
    )

    print(f"\n{model_name}")
    print(f"Files found: {len(files)}")

    for filename in files:

        # ----------------------------------------------------
        # Extract PART number from filename
        # ----------------------------------------------------

        filename_only = filename.split("\\")[-1]

        # Example:
        # coverage_scores_baseline_qwen25_15b_part2_03.jsonl

        part = filename_only.split("_")[-1].replace(
            ".jsonl", ""
        )

        with open(filename, "r", encoding="utf-8") as f:

            for line in f:

                if not line.strip():
                    continue

                data = json.loads(line)

                scores = data.get(
                    "coverage_scores",
                    {}
                )

                index = data.get("index")

                # ------------------------------------------------
                # IMPORTANT:
                # Use PART + INDEX as unique case identifier
                # ------------------------------------------------

                case_id = f"{part}_{index}"

                records.append({
                    "model": model_name,
                    "part": part,
                    "index": index,
                    "case_id": case_id,
                    "patient_coverage": scores.get(
                        "patient_coverage_score"
                    )
                })


# ============================================================
# DATAFRAME
# ============================================================

df = pd.DataFrame(records)

print("\n" + "=" * 70)
print("DATA SUMMARY")
print("=" * 70)

print(
    df.groupby("model")["case_id"]
      .nunique()
)


# ============================================================
# CHECK DUPLICATES
# ============================================================

duplicates = (
    df.groupby(["model", "case_id"])
      .size()
      .reset_index(name="count")
)

duplicates = duplicates[
    duplicates["count"] > 1
]

if len(duplicates) > 0:

    print("\n⚠️ DUPLICATES FOUND:")
    print(
        duplicates.to_string(
            index=False
        )
    )

    raise ValueError(
        "Duplicate case_id found. "
        "Please investigate before running the test."
    )

else:

    print(
        "\n✅ No duplicate model-case combinations found."
    )


# ============================================================
# PIVOT
# ============================================================

pivot = df.pivot(
    index="case_id",
    columns="model",
    values="patient_coverage"
)


# ============================================================
# KEEP ONLY COMPLETE CASES
# ============================================================

pivot = pivot.dropna(
    subset=[
        "Baseline",
        "Normal SFT",
        "Weighted SFT"
    ]
)


print("\n" + "=" * 70)
print("PAIRED DATA")
print("=" * 70)

print(
    f"Number of paired cases: {len(pivot)}"
)

print("\nFirst 10 cases:")
print(pivot.head(10))


# ============================================================
# EXTRACT SCORES
# ============================================================

baseline = pivot["Baseline"]
normal_sft = pivot["Normal SFT"]
weighted_sft = pivot["Weighted SFT"]


# ============================================================
# DESCRIPTIVE STATISTICS
# ============================================================

print("\n" + "=" * 70)
print("DESCRIPTIVE STATISTICS")
print("=" * 70)

for model in [
    "Baseline",
    "Normal SFT",
    "Weighted SFT"
]:

    scores = pivot[model]

    print(f"\n{model}")

    print(
        f"  N      = {len(scores)}"
    )

    print(
        f"  Mean   = {scores.mean():.4f}"
    )

    print(
        f"  Median = {scores.median():.4f}"
    )

    print(
        f"  SD     = {scores.std():.4f}"
    )

    print(
        f"  Min    = {scores.min():.4f}"
    )

    print(
        f"  Max    = {scores.max():.4f}"
    )


# ============================================================
# FRIEDMAN TEST
# ============================================================

friedman_stat, friedman_p = friedmanchisquare(
    baseline,
    normal_sft,
    weighted_sft
)


print("\n" + "=" * 70)
print("FRIEDMAN TEST")
print("=" * 70)

print(
    f"Friedman statistic = {friedman_stat:.4f}"
)

print(
    f"p-value            = {friedman_p:.6f}"
)


if friedman_p < 0.05:

    print(
        "✅ Significant overall difference "
        "(p < 0.05)"
    )

else:

    print(
        "❌ No significant overall difference "
        "(p >= 0.05)"
    )


# ============================================================
# WILCOXON PAIRWISE TESTS
# ============================================================

comparisons = {

    "Baseline vs Normal SFT":
        (baseline, normal_sft),

    "Baseline vs Weighted SFT":
        (baseline, weighted_sft),

    "Normal SFT vs Weighted SFT":
        (normal_sft, weighted_sft)
}


wilcoxon_results = []


for comparison, (x, y) in comparisons.items():

    statistic, p_value = wilcoxon(
        x,
        y,
        alternative="two-sided"
    )

    wilcoxon_results.append({

        "comparison": comparison,

        "statistic": statistic,

        "raw_p": p_value
    })


# ============================================================
# HOLM CORRECTION
# ============================================================

wilcoxon_results = sorted(
    wilcoxon_results,
    key=lambda x: x["raw_p"]
)

m = len(wilcoxon_results)


for i, result in enumerate(
    wilcoxon_results
):

    result["holm_p"] = min(
        result["raw_p"] * (m - i),
        1.0
    )


# Ensure Holm p-values are monotonic

for i in range(1, m):

    wilcoxon_results[i]["holm_p"] = max(
        wilcoxon_results[i]["holm_p"],
        wilcoxon_results[i - 1]["holm_p"]
    )


# ============================================================
# RANK-BISERIAL EFFECT SIZE
# ============================================================

def rank_biserial_effect(x, y):

    differences = (
        np.asarray(x)
        -
        np.asarray(y)
    )

    # Remove zero differences

    differences = differences[
        differences != 0
    ]

    if len(differences) == 0:

        return 0.0

    ranks = pd.Series(
        np.abs(differences)
    ).rank()

    positive = ranks[
        differences > 0
    ].sum()

    negative = ranks[
        differences < 0
    ].sum()

    effect = (
        (positive - negative)
        /
        (positive + negative)
    )

    return effect


# ============================================================
# CALCULATE EFFECT SIZES
# ============================================================

for result in wilcoxon_results:

    comparison = result[
        "comparison"
    ]

    x, y = comparisons[
        comparison
    ]

    result[
        "rank_biserial"
    ] = rank_biserial_effect(
        x,
        y
    )


# ============================================================
# RESULTS TABLE
# ============================================================

results_df = pd.DataFrame(
    wilcoxon_results
)


print("\n" + "=" * 70)
print("POST-HOC WILCOXON TESTS")
print("=" * 70)

print(
    results_df[
        [
            "comparison",
            "statistic",
            "raw_p",
            "holm_p",
            "rank_biserial"
        ]
    ].to_string(
        index=False
    )
)


# ============================================================
# INTERPRETATION
# ============================================================

print("\n" + "=" * 70)
print("INTERPRETATION")
print("=" * 70)


for _, row in results_df.iterrows():

    p = row["holm_p"]

    effect = row[
        "rank_biserial"
    ]

    if p < 0.001:

        significance = (
            "Highly significant "
            "(p < 0.001)"
        )

    elif p < 0.01:

        significance = (
            "Significant "
            "(p < 0.01)"
        )

    elif p < 0.05:

        significance = (
            "Significant "
            "(p < 0.05)"
        )

    else:

        significance = (
            "Not significant "
            "(p >= 0.05)"
        )

    print(
        f"\n{row['comparison']}"
    )

    print(
        f"  {significance}"
    )

    print(
        f"  Holm-adjusted p = "
        f"{p:.6f}"
    )

    print(
        f"  Rank-biserial effect = "
        f"{effect:.4f}"
    )
```

```python
import json
from pathlib import Path

# ============================================================
# PATHS
# ============================================================

WEIGHTED_SFT_DIR = Path(
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation\Factuality_temp\Organising_file_1"
)

DPO_DIR = Path(
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation\DPO\Factuality_Score\Factuality_Score_1"
)

# ============================================================
# MATCHING FIELD
# ============================================================

MATCH_FIELD = "index"

# ============================================================
# LOAD JSONL
# ============================================================

def load_jsonl(path):
    records = []

    with open(path, "r", encoding="utf-8") as f:
        for line_number, line in enumerate(f, start=1):
            line = line.strip()

            if not line:
                continue

            try:
                records.append(json.loads(line))
            except json.JSONDecodeError as e:
                print(f"ERROR in {path.name}, line {line_number}: {e}")

    return records


# ============================================================
# GET IDs
# ============================================================

def get_ids(records, field=MATCH_FIELD):
    ids = []

    for record in records:
        if field not in record:
            raise KeyError(
                f"Field '{field}' not found in record:\n{record}"
            )

        ids.append(record[field])

    return ids


# ============================================================
# MAIN COMPARISON
# ============================================================

all_results = []

print("\n")
print("=" * 110)
print("WEIGHTED SFT vs DPO — INTERSECTION ANALYSIS")
print("=" * 110)

for i in range(1, 7):

    weighted_path = (
        WEIGHTED_SFT_DIR /
        f"factuality_scores_weighted_sft_qwen25_1.5b_part2_50_{i:02d}.jsonl"
    )

    dpo_path = (
        DPO_DIR /
        f"factuality_scores_patient_report_wgt_dpo_{i:02d}.jsonl"
    )

    # --------------------------------------------------------
    # Check files
    # --------------------------------------------------------

    if not weighted_path.exists():
        print(f"\nWARNING: Missing Weighted SFT file:")
        print(weighted_path)
        continue

    if not dpo_path.exists():
        print(f"\nWARNING: Missing DPO file:")
        print(dpo_path)
        continue

    # --------------------------------------------------------
    # Load files
    # --------------------------------------------------------

    weighted_records = load_jsonl(weighted_path)
    dpo_records = load_jsonl(dpo_path)

    # --------------------------------------------------------
    # Extract IDs
    # --------------------------------------------------------

    weighted_ids = get_ids(weighted_records)
    dpo_ids = get_ids(dpo_records)

    # --------------------------------------------------------
    # Convert to sets
    # --------------------------------------------------------

    weighted_set = set(weighted_ids)
    dpo_set = set(dpo_ids)

    # --------------------------------------------------------
    # Intersection / matched
    # --------------------------------------------------------

    matched = weighted_set.intersection(dpo_set)

    # --------------------------------------------------------
    # Unmatched
    # --------------------------------------------------------

    weighted_only = weighted_set - dpo_set
    dpo_only = dpo_set - weighted_set

    # --------------------------------------------------------
    # Duplicate counts
    # --------------------------------------------------------

    weighted_duplicates = len(weighted_ids) - len(weighted_set)
    dpo_duplicates = len(dpo_ids) - len(dpo_set)

    # --------------------------------------------------------
    # Save results
    # --------------------------------------------------------

    result = {
        "part": f"{i:02d}",

        "weighted_sft_total_records": len(weighted_records),
        "dpo_total_records": len(dpo_records),

        "weighted_sft_unique_ids": len(weighted_set),
        "dpo_unique_ids": len(dpo_set),

        "matched": len(matched),

        "weighted_sft_unmatched": len(weighted_only),
        "dpo_unmatched": len(dpo_only),

        "weighted_sft_duplicates": weighted_duplicates,
        "dpo_duplicates": dpo_duplicates,
    }

    all_results.append(result)

    # --------------------------------------------------------
    # Print detailed result
    # --------------------------------------------------------

    print("\n" + "-" * 110)
    print(f"PART {i:02d}")
    print("-" * 110)

    print(f"Weighted SFT total records : {len(weighted_records)}")
    print(f"DPO total records         : {len(dpo_records)}")

    print(f"Weighted SFT unique IDs   : {len(weighted_set)}")
    print(f"DPO unique IDs            : {len(dpo_set)}")

    print(f"Matched                   : {len(matched)}")
    print(f"Weighted SFT unmatched    : {len(weighted_only)}")
    print(f"DPO unmatched             : {len(dpo_only)}")

    print(f"Weighted SFT duplicates   : {weighted_duplicates}")
    print(f"DPO duplicates            : {dpo_duplicates}")

    # --------------------------------------------------------
    # Print actual unmatched IDs
    # --------------------------------------------------------

    print("\nWeighted SFT ONLY IDs:")
    print(sorted(weighted_only))

    print("\nDPO ONLY IDs:")
    print(sorted(dpo_only))


# ============================================================
# SUMMARY TABLE
# ============================================================

print("\n")
print("=" * 110)
print("SUMMARY")
print("=" * 110)

print(
    f"{'Part':<8}"
    f"{'Weighted SFT':<18}"
    f"{'DPO':<15}"
    f"{'Matched':<15}"
    f"{'SFT Unmatched':<18}"
    f"{'DPO Unmatched':<18}"
)

print("-" * 110)

for r in all_results:

    print(
        f"{r['part']:<8}"
        f"{r['weighted_sft_total_records']:<18}"
        f"{r['dpo_total_records']:<15}"
        f"{r['matched']:<15}"
        f"{r['weighted_sft_unmatched']:<18}"
        f"{r['dpo_unmatched']:<18}"
    )


# ============================================================
# OVERALL TOTALS
# ============================================================

total_weighted = sum(
    r["weighted_sft_total_records"]
    for r in all_results
)

total_dpo = sum(
    r["dpo_total_records"]
    for r in all_results
)

total_matched = sum(
    r["matched"]
    for r in all_results
)

total_weighted_unmatched = sum(
    r["weighted_sft_unmatched"]
    for r in all_results
)

total_dpo_unmatched = sum(
    r["dpo_unmatched"]
    for r in all_results
)

print("\n")
print("=" * 110)
print("OVERALL TOTAL")
print("=" * 110)

print(f"Weighted SFT total       : {total_weighted}")
print(f"DPO total                : {total_dpo}")
print(f"Matched total            : {total_matched}")
print(f"Weighted SFT unmatched  : {total_weighted_unmatched}")
print(f"DPO unmatched            : {total_dpo_unmatched}")

print("\nAnalysis complete.")
print("NO FILES WERE MODIFIED.")
```

```python
import json
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ============================================================
# FOLDER
# ============================================================

FOLDER = Path(
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation\DPO\Factuality_Score\Factuality_Score_1"
)

# ============================================================
# FILE NAMES
# ============================================================

weighted_files = [
    FOLDER / f"factuality_scores_weighted_sft_qwen25_1.5b_part2_50_{i:02d}.jsonl"
    for i in range(1, 7)
]

dpo_files = [
    FOLDER / f"factuality_scores_patient_report_wgt_dpo_{i:02d}.jsonl"
    for i in range(1, 7)
]


# ============================================================
# LOAD JSONL
# ============================================================

def load_jsonl(filepath):
    records = []

    with open(filepath, "r", encoding="utf-8") as f:
        for line_number, line in enumerate(f, start=1):

            if not line.strip():
                continue

            try:
                records.append(json.loads(line))
            except json.JSONDecodeError as e:
                print(f"JSON error in {filepath.name}, line {line_number}: {e}")

    return records


# ============================================================
# SAVE JSONL
# ============================================================

def save_jsonl(filepath, records):

    with open(filepath, "w", encoding="utf-8") as f:

        for record in records:
            f.write(
                json.dumps(
                    record,
                    ensure_ascii=False
                ) + "\n"
            )


# ============================================================
# PAIRWISE INTERSECTION
# ============================================================

print("\n")
print("=" * 100)
print("PAIRWISE INTERSECTION: WEIGHTED SFT vs DPO")
print("=" * 100)

all_results = []

for part in range(6):

    weighted_path = weighted_files[part]
    dpo_path = dpo_files[part]

    print("\n" + "-" * 100)
    print(f"PART {part + 1:02d}")
    print("-" * 100)

    # --------------------------------------------------------
    # Check files
    # --------------------------------------------------------

    if not weighted_path.exists():
        raise FileNotFoundError(
            f"Weighted SFT file not found:\n{weighted_path}"
        )

    if not dpo_path.exists():
        raise FileNotFoundError(
            f"DPO file not found:\n{dpo_path}"
        )

    # --------------------------------------------------------
    # Load original files
    # --------------------------------------------------------

    weighted_records = load_jsonl(weighted_path)
    dpo_records = load_jsonl(dpo_path)

    print(f"Original Weighted SFT records: {len(weighted_records)}")
    print(f"Original DPO records         : {len(dpo_records)}")

    # --------------------------------------------------------
    # Get indices
    # --------------------------------------------------------

    weighted_indices = {
        record["index"]
        for record in weighted_records
        if "index" in record
    }

    dpo_indices = {
        record["index"]
        for record in dpo_records
        if "index" in record
    }

    # --------------------------------------------------------
    # INTERSECTION
    # --------------------------------------------------------

    common_indices = weighted_indices.intersection(dpo_indices)

    # --------------------------------------------------------
    # Filter BOTH datasets
    # --------------------------------------------------------

    weighted_matched = [
        record
        for record in weighted_records
        if record.get("index") in common_indices
    ]

    dpo_matched = [
        record
        for record in dpo_records
        if record.get("index") in common_indices
    ]

    # --------------------------------------------------------
    # Sort both by index
    # --------------------------------------------------------

    weighted_matched.sort(key=lambda x: x["index"])
    dpo_matched.sort(key=lambda x: x["index"])

    # --------------------------------------------------------
    # Verify matching
    # --------------------------------------------------------

    weighted_final_indices = [
        record["index"]
        for record in weighted_matched
    ]

    dpo_final_indices = [
        record["index"]
        for record in dpo_matched
    ]

    indices_match = weighted_final_indices == dpo_final_indices

    # --------------------------------------------------------
    # Print results
    # --------------------------------------------------------

    print(f"Common indices              : {len(common_indices)}")
    print(f"Matched Weighted SFT records: {len(weighted_matched)}")
    print(f"Matched DPO records         : {len(dpo_matched)}")

    print(
        f"Weighted SFT removed        : "
        f"{len(weighted_records) - len(weighted_matched)}"
    )

    print(
        f"DPO removed                 : "
        f"{len(dpo_records) - len(dpo_matched)}"
    )

    print(f"Indices identical?          : {indices_match}")

    if not indices_match:
        raise RuntimeError(
            f"ERROR: Final indices do not match for part {part + 1:02d}"
        )

    # --------------------------------------------------------
    # REPLACE ORIGINAL FILES
    # --------------------------------------------------------

    save_jsonl(weighted_path, weighted_matched)
    save_jsonl(dpo_path, dpo_matched)

    print("\nFiles replaced successfully:")
    print(f"  {weighted_path.name}")
    print(f"  {dpo_path.name}")

    all_results.append({
        "part": part + 1,
        "weighted_original": len(weighted_records),
        "dpo_original": len(dpo_records),
        "intersection": len(common_indices),
        "weighted_final": len(weighted_matched),
        "dpo_final": len(dpo_matched),
        "weighted_removed": len(weighted_records) - len(weighted_matched),
        "dpo_removed": len(dpo_records) - len(dpo_matched),
        "indices_match": indices_match
    })


# ============================================================
# SUMMARY
# ============================================================

summary_df = pd.DataFrame(all_results)

print("\n")
print("=" * 100)
print("FINAL SUMMARY")
print("=" * 100)

print(summary_df.to_string(index=False))


# ============================================================
# FINAL GLOBAL CHECK
# ============================================================

print("\n")
print("=" * 100)
print("FINAL VERIFICATION")
print("=" * 100)

total_weighted = summary_df["weighted_final"].sum()
total_dpo = summary_df["dpo_final"].sum()
total_intersection = summary_df["intersection"].sum()

print(f"Total Weighted SFT after intersection: {total_weighted}")
print(f"Total DPO after intersection         : {total_dpo}")
print(f"Total matched records                : {total_intersection}")

if total_weighted == total_dpo:
    print("\nSUCCESS: Weighted SFT and DPO have the same total number of matched records.")
else:
    print("\nWARNING: Total numbers do not match.")


# ============================================================
# LOAD THE NOW-MATCHED FILES FOR DISTRIBUTION ANALYSIS
# ============================================================

records = []

for part in range(6):

    # --------------------------------------------------------
    # Weighted SFT
    # --------------------------------------------------------

    weighted_path = weighted_files[part]

    with open(weighted_path, "r", encoding="utf-8") as f:

        for line in f:

            if not line.strip():
                continue

            data = json.loads(line)

            scores = data.get("factuality_scores", {})

            records.append({
                "model": "Weighted SFT",
                "part": part + 1,
                "index": data.get("index"),
                "patient": scores.get(
                    "patient_factuality_score"
                ),
                "syndrome": scores.get(
                    "syndrome_factuality_score"
                ),
                "average": scores.get(
                    "average_factuality_score"
                )
            })

    # --------------------------------------------------------
    # DPO
    # --------------------------------------------------------

    dpo_path = dpo_files[part]

    with open(dpo_path, "r", encoding="utf-8") as f:

        for line in f:

            if not line.strip():
                continue

            data = json.loads(line)

            scores = data.get("factuality_scores", {})

            records.append({
                "model": "DPO",
                "part": part + 1,
                "index": data.get("index"),
                "patient": scores.get(
                    "patient_factuality_score"
                ),
                "syndrome": scores.get(
                    "syndrome_factuality_score"
                ),
                "average": scores.get(
                    "average_factuality_score"
                )
            })


df = pd.DataFrame(records)


# ============================================================
# DATA SUMMARY
# ============================================================

print("\n")
print("=" * 100)
print("DISTRIBUTION DATA")
print("=" * 100)

print("\nNumber of records:")
print(df.groupby("model").size())

print("\nDescriptive statistics:")
print(
    df.groupby("model")[["patient", "syndrome", "average"]]
    .describe()
)


# ============================================================
# DISTRIBUTIONS
# ============================================================

metrics = {
    "patient": "Patient Factuality Score",
    "syndrome": "Syndrome Factuality Score",
    "average": "Average Factuality Score"
}


# ============================================================
# SEPARATE DISTRIBUTION — WEIGHTED SFT
# ============================================================

for metric, label in metrics.items():

    plt.figure(figsize=(8, 5))

    sns.histplot(
        data=df[df["model"] == "Weighted SFT"],
        x=metric,
        bins=20,
        kde=True
    )

    plt.xlabel(label)
    plt.ylabel("Number of reports")
    plt.title(f"Weighted SFT — Distribution of {label}")

    plt.tight_layout()
    plt.show()


# ============================================================
# SEPARATE DISTRIBUTION — DPO
# ============================================================

for metric, label in metrics.items():

    plt.figure(figsize=(8, 5))

    sns.histplot(
        data=df[df["model"] == "DPO"],
        x=metric,
        bins=20,
        kde=True
    )

    plt.xlabel(label)
    plt.ylabel("Number of reports")
    plt.title(f"DPO — Distribution of {label}")

    plt.tight_layout()
    plt.show()


print("\n")
print("=" * 100)
print("COMPLETE")
print("=" * 100)

print("The original files have been replaced with their pairwise intersections.")
print("Weighted SFT and DPO now contain exactly the same indices within each part.")
print("Distributions have been generated separately for both models.")
```

```python
import json
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import (
    wilcoxon,
    ttest_rel,
    shapiro
)


# ============================================================
# 1. FOLDER
# ============================================================

FOLDER = Path(
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation\DPO\Factuality_Score\Factuality_Score_1"
)


# ============================================================
# 2. FILES
# ============================================================

weighted_files = [
    FOLDER / f"factuality_scores_weighted_sft_qwen25_1.5b_part2_50_{i:02d}.jsonl"
    for i in range(1, 7)
]

dpo_files = [
    FOLDER / f"factuality_scores_patient_report_wgt_dpo_{i:02d}.jsonl"
    for i in range(1, 7)
]


# ============================================================
# 3. LOAD JSONL
# ============================================================

def load_jsonl(filepath):

    records = []

    with open(filepath, "r", encoding="utf-8") as f:

        for line_number, line in enumerate(f, start=1):

            if not line.strip():
                continue

            try:
                records.append(json.loads(line))

            except json.JSONDecodeError as e:
                print(
                    f"JSON error in {filepath.name}, "
                    f"line {line_number}: {e}"
                )

    return records


# ============================================================
# 4. SAVE JSONL
# ============================================================

def save_jsonl(filepath, records):

    with open(filepath, "w", encoding="utf-8") as f:

        for record in records:

            f.write(
                json.dumps(
                    record,
                    ensure_ascii=False
                ) + "\n"
            )


# ============================================================
# 5. PAIRWISE INTERSECTION
# ============================================================

print("\n")
print("=" * 100)
print("PAIRWISE INTERSECTION")
print("=" * 100)

intersection_results = []

for part in range(6):

    weighted_path = weighted_files[part]
    dpo_path = dpo_files[part]

    print("\n" + "-" * 100)
    print(f"PART {part + 1:02d}")
    print("-" * 100)

    # --------------------------------------------------------
    # Load
    # --------------------------------------------------------

    weighted_records = load_jsonl(weighted_path)
    dpo_records = load_jsonl(dpo_path)

    print(
        f"Original Weighted SFT records: "
        f"{len(weighted_records)}"
    )

    print(
        f"Original DPO records: "
        f"{len(dpo_records)}"
    )

    # --------------------------------------------------------
    # Get indices
    # --------------------------------------------------------

    weighted_indices = {
        record["index"]
        for record in weighted_records
        if "index" in record
    }

    dpo_indices = {
        record["index"]
        for record in dpo_records
        if "index" in record
    }

    # --------------------------------------------------------
    # Intersection
    # --------------------------------------------------------

    common_indices = (
        weighted_indices.intersection(dpo_indices)
    )

    # --------------------------------------------------------
    # Filter both files
    # --------------------------------------------------------

    weighted_matched = [
        record
        for record in weighted_records
        if record.get("index") in common_indices
    ]

    dpo_matched = [
        record
        for record in dpo_records
        if record.get("index") in common_indices
    ]

    # --------------------------------------------------------
    # Sort by index
    # --------------------------------------------------------

    weighted_matched.sort(
        key=lambda x: x["index"]
    )

    dpo_matched.sort(
        key=lambda x: x["index"]
    )

    # --------------------------------------------------------
    # Verify
    # --------------------------------------------------------

    weighted_final_indices = [
        record["index"]
        for record in weighted_matched
    ]

    dpo_final_indices = [
        record["index"]
        for record in dpo_matched
    ]

    indices_match = (
        weighted_final_indices == dpo_final_indices
    )

    if not indices_match:

        raise RuntimeError(
            f"Indices do not match for part {part + 1:02d}"
        )

    # --------------------------------------------------------
    # Replace original files
    # --------------------------------------------------------

    save_jsonl(
        weighted_path,
        weighted_matched
    )

    save_jsonl(
        dpo_path,
        dpo_matched
    )

    # --------------------------------------------------------
    # Results
    # --------------------------------------------------------

    weighted_removed = (
        len(weighted_records)
        - len(weighted_matched)
    )

    dpo_removed = (
        len(dpo_records)
        - len(dpo_matched)
    )

    print(
        f"Intersection: {len(common_indices)}"
    )

    print(
        f"Weighted SFT removed: {weighted_removed}"
    )

    print(
        f"DPO removed: {dpo_removed}"
    )

    print(
        f"Indices identical: {indices_match}"
    )

    intersection_results.append({
        "Part": part + 1,
        "Weighted_SFT_Original": len(weighted_records),
        "DPO_Original": len(dpo_records),
        "Intersection": len(common_indices),
        "Weighted_SFT_Removed": weighted_removed,
        "DPO_Removed": dpo_removed
    })


# ============================================================
# 6. INTERSECTION SUMMARY
# ============================================================

intersection_df = pd.DataFrame(
    intersection_results
)

print("\n")
print("=" * 100)
print("INTERSECTION SUMMARY")
print("=" * 100)

print(
    intersection_df.to_string(
        index=False
    )
)

print("\nTotal Weighted SFT after matching:",
      intersection_df["Intersection"].sum())

print("Total DPO after matching:",
      intersection_df["Intersection"].sum())


# ============================================================
# 7. EXTRACT ONLY AVERAGE FACTUALITY SCORE
# ============================================================

analysis_records = []

for part in range(6):

    # --------------------------------------------------------
    # Weighted SFT
    # --------------------------------------------------------

    weighted_records = load_jsonl(
        weighted_files[part]
    )

    for record in weighted_records:

        scores = record.get(
            "factuality_scores",
            {}
        )

        average_score = scores.get(
            "average_factuality_score"
        )

        if average_score is not None:

            analysis_records.append({
                "model": "Weighted SFT",
                "part": part + 1,
                "index": record.get("index"),
                "average_factuality_score": float(
                    average_score
                )
            })

    # --------------------------------------------------------
    # DPO
    # --------------------------------------------------------

    dpo_records = load_jsonl(
        dpo_files[part]
    )

    for record in dpo_records:

        scores = record.get(
            "factuality_scores",
            {}
        )

        average_score = scores.get(
            "average_factuality_score"
        )

        if average_score is not None:

            analysis_records.append({
                "model": "DPO",
                "part": part + 1,
                "index": record.get("index"),
                "average_factuality_score": float(
                    average_score
                )
            })


df = pd.DataFrame(
    analysis_records
)


# ============================================================
# 8. BASIC CHECK
# ============================================================

print("\n")
print("=" * 100)
print("DATA USED FOR STATISTICAL ANALYSIS")
print("=" * 100)

print(
    df.groupby("model").size()
)

print(
    "\nMissing values:"
)

print(
    df["average_factuality_score"]
    .isna()
    .sum()
)


# ============================================================
# 9. DESCRIPTIVE STATISTICS
# ============================================================

descriptive = (
    df
    .groupby("model")["average_factuality_score"]
    .agg(
        N="count",
        Mean="mean",
        Median="median",
        SD="std",
        Min="min",
        Q1=lambda x: x.quantile(0.25),
        Q3=lambda x: x.quantile(0.75),
        Max="max"
    )
)

descriptive["IQR"] = (
    descriptive["Q3"]
    - descriptive["Q1"]
)

print("\n")
print("=" * 100)
print("DESCRIPTIVE STATISTICS")
print("=" * 100)

print(
    descriptive.round(4).to_string()
)


# ============================================================
# 10. EXTRACT PAIRED DATA
# ============================================================

weighted = (
    df[df["model"] == "Weighted SFT"]
    [["index", "average_factuality_score"]]
    .rename(
        columns={
            "average_factuality_score":
                "weighted_sft"
        }
    )
)

dpo = (
    df[df["model"] == "DPO"]
    [["index", "average_factuality_score"]]
    .rename(
        columns={
            "average_factuality_score":
                "dpo"
        }
    )
)

paired = pd.merge(
    weighted,
    dpo,
    on="index",
    how="inner"
)

print("\n")
print("=" * 100)
print("PAIRED DATA")
print("=" * 100)

print(
    f"Number of paired observations: "
    f"{len(paired)}"
)


# ============================================================
# 11. CALCULATE PAIRED DIFFERENCE
# ============================================================

paired["difference"] = (
    paired["dpo"]
    - paired["weighted_sft"]
)

differences = (
    paired["difference"]
    .dropna()
    .values
)


# ============================================================
# 12. DESCRIPTIVE PAIRED DIFFERENCE
# ============================================================

mean_difference = np.mean(
    differences
)

median_difference = np.median(
    differences
)

sd_difference = np.std(
    differences,
    ddof=1
)

print("\n")
print("=" * 100)
print("PAIRED DIFFERENCE: DPO - WEIGHTED SFT")
print("=" * 100)

print(
    f"Mean difference: "
    f"{mean_difference:.4f}"
)

print(
    f"Median difference: "
    f"{median_difference:.4f}"
)

print(
    f"SD of difference: "
    f"{sd_difference:.4f}"
)


# ============================================================
# 13. NORMALITY TEST OF PAIRED DIFFERENCES
# ============================================================

if len(differences) >= 3:

    shapiro_stat, shapiro_p = shapiro(
        differences
    )

    print("\n")
    print("=" * 100)
    print("SHAPIRO-WILK TEST OF PAIRED DIFFERENCES")
    print("=" * 100)

    print(
        f"Statistic = {shapiro_stat:.4f}"
    )

    print(
        f"p-value = {shapiro_p:.6f}"
    )

    if shapiro_p < 0.05:

        print(
            "Conclusion: Differences significantly "
            "deviate from normality."
        )

    else:

        print(
            "Conclusion: No significant evidence "
            "of non-normality."
        )


# ============================================================
# 14. PAIRED T-TEST
# ============================================================

t_stat, t_p = ttest_rel(
    paired["dpo"],
    paired["weighted_sft"]
)

print("\n")
print("=" * 100)
print("PAIRED T-TEST")
print("=" * 100)

print(
    f"t-statistic = {t_stat:.4f}"
)

print(
    f"p-value = {t_p:.6f}"
)


# ============================================================
# 15. WILCOXON SIGNED-RANK TEST
# ============================================================

wilcoxon_stat, wilcoxon_p = wilcoxon(
    paired["dpo"],
    paired["weighted_sft"],
    alternative="two-sided"
)

print("\n")
print("=" * 100)
print("WILCOXON SIGNED-RANK TEST")
print("=" * 100)

print(
    f"W-statistic = {wilcoxon_stat:.4f}"
)

print(
    f"p-value = {wilcoxon_p:.6f}"
)


# ============================================================
# 16. 95% CI FOR MEAN PAIRED DIFFERENCE
# ============================================================

n = len(differences)

standard_error = (
    sd_difference
    / np.sqrt(n)
)

t_critical = 2.0

# More accurate critical value using scipy
from scipy.stats import t

t_critical = t.ppf(
    0.975,
    df=n - 1
)

ci_lower = (
    mean_difference
    - t_critical * standard_error
)

ci_upper = (
    mean_difference
    + t_critical * standard_error
)

print("\n")
print("=" * 100)
print("95% CONFIDENCE INTERVAL FOR MEAN DIFFERENCE")
print("=" * 100)

print(
    f"Mean difference (DPO - Weighted SFT): "
    f"{mean_difference:.4f}"
)

print(
    f"95% CI: "
    f"[{ci_lower:.4f}, {ci_upper:.4f}]"
)


# ============================================================
# 17. COHEN'S dz EFFECT SIZE
# ============================================================

cohens_dz = (
    mean_difference
    / sd_difference
)

print("\n")
print("=" * 100)
print("EFFECT SIZE")
print("=" * 100)

print(
    f"Cohen's dz = {cohens_dz:.4f}"
)


# ============================================================
# 18. INTERPRETATION
# ============================================================

print("\n")
print("=" * 100)
print("STATISTICAL INTERPRETATION")
print("=" * 100)

if wilcoxon_p < 0.05:

    print(
        "Wilcoxon result: SIGNIFICANT difference "
        "between Weighted SFT and DPO "
        "(p < 0.05)."
    )

else:

    print(
        "Wilcoxon result: NO statistically significant "
        "difference between Weighted SFT and DPO "
        "(p >= 0.05)."
    )


if mean_difference > 0:

    print(
        f"DPO has a higher mean average factuality score "
        f"by {mean_difference:.4f}."
    )

elif mean_difference < 0:

    print(
        f"DPO has a lower mean average factuality score "
        f"by {abs(mean_difference):.4f}."
    )

else:

    print(
        "The mean scores are identical."
    )


# ============================================================
# 19. DISTRIBUTION — WEIGHTED SFT
# ============================================================

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df[df["model"] == "Weighted SFT"],
    x="average_factuality_score",
    bins=20,
    kde=True
)

plt.xlabel(
    "Average Factuality Score"
)

plt.ylabel(
    "Number of Reports"
)

plt.title(
    "Weighted SFT — Distribution of Average Factuality Scores"
)

plt.tight_layout()

plt.show()


# ============================================================
# 20. DISTRIBUTION — DPO
# ============================================================

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df[df["model"] == "DPO"],
    x="average_factuality_score",
    bins=20,
    kde=True
)

plt.xlabel(
    "Average Factuality Score"
)

plt.ylabel(
    "Number of Reports"
)

plt.title(
    "DPO — Distribution of Average Factuality Scores"
)

plt.tight_layout()

plt.show()


# ============================================================
# 21. OPTIONAL: DIRECT VISUAL COMPARISON
# ============================================================

plt.figure(figsize=(9, 5))

sns.kdeplot(
    data=df,
    x="average_factuality_score",
    hue="model",
    fill=True,
    common_norm=False,
    alpha=0.3
)

plt.xlabel(
    "Average Factuality Score"
)

plt.ylabel(
    "Density"
)

plt.title(
    "Average Factuality Score — Weighted SFT vs DPO"
)

plt.tight_layout()

plt.show()


# ============================================================
# 22. FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 100)
print("FINAL SUMMARY")
print("=" * 100)

print(
    "\nDescriptive statistics:"
)

print(
    descriptive.round(4)
)

print(
    "\nPaired comparison:"
)

print(
    f"N = {len(paired)}"
)

print(
    f"Mean difference (DPO - Weighted SFT) = "
    f"{mean_difference:.4f}"
)

print(
    f"95% CI = "
    f"[{ci_lower:.4f}, {ci_upper:.4f}]"
)

print(
    f"Paired t-test p = "
    f"{t_p:.6f}"
)

print(
    f"Wilcoxon p = "
    f"{wilcoxon_p:.6f}"
)

print(
    f"Cohen's dz = "
    f"{cohens_dz:.4f}"
)

print("\nAnalysis complete.")
```

```python
import json
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import (
    wilcoxon,
    ttest_rel,
    t
)


# ============================================================
# 1. FOLDER
# ============================================================

FOLDER = Path(
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation\DPO\Factuality_Score\Factuality_Score_1"
)


# ============================================================
# 2. FILES
# ============================================================

weighted_files = [
    FOLDER / f"factuality_scores_weighted_sft_qwen25_1.5b_part2_50_{i:02d}.jsonl"
    for i in range(1, 7)
]

dpo_files = [
    FOLDER / f"factuality_scores_patient_report_wgt_dpo_{i:02d}.jsonl"
    for i in range(1, 7)
]


# ============================================================
# 3. LOAD JSONL
# ============================================================

def load_jsonl(filepath):

    records = []

    with open(filepath, "r", encoding="utf-8") as f:

        for line in f:

            if not line.strip():
                continue

            records.append(json.loads(line))

    return records


# ============================================================
# 4. EXTRACT AVERAGE FACTUALITY SCORES
# ============================================================

weighted_records = []
dpo_records = []

for part in range(1, 7):

    # --------------------------------------------------------
    # Weighted SFT
    # --------------------------------------------------------

    records = load_jsonl(
        weighted_files[part - 1]
    )

    for record in records:

        score = record.get(
            "factuality_scores", {}
        ).get(
            "average_factuality_score"
        )

        if score is not None:

            weighted_records.append({
                "part": part,
                "index": record["index"],
                "average": float(score)
            })


    # --------------------------------------------------------
    # DPO
    # --------------------------------------------------------

    records = load_jsonl(
        dpo_files[part - 1]
    )

    for record in records:

        score = record.get(
            "factuality_scores", {}
        ).get(
            "average_factuality_score"
        )

        if score is not None:

            dpo_records.append({
                "part": part,
                "index": record["index"],
                "average": float(score)
            })


# ============================================================
# 5. DATAFRAMES
# ============================================================

weighted_df = pd.DataFrame(
    weighted_records
)

dpo_df = pd.DataFrame(
    dpo_records
)

print("\n" + "=" * 80)
print("DATA CHECK")
print("=" * 80)

print(
    "Weighted SFT records:",
    len(weighted_df)
)

print(
    "DPO records:",
    len(dpo_df)
)


# ============================================================
# 6. VERIFY PAIRING WITHIN EACH PART
# ============================================================

weighted_keys = set(
    zip(
        weighted_df["part"],
        weighted_df["index"]
    )
)

dpo_keys = set(
    zip(
        dpo_df["part"],
        dpo_df["index"]
    )
)

print(
    "\nWeighted SFT unique (part,index):",
    len(weighted_keys)
)

print(
    "DPO unique (part,index):",
    len(dpo_keys)
)

print(
    "Same paired reports:",
    weighted_keys == dpo_keys
)


# ============================================================
# 7. CREATE PAIRED DATA — PART + INDEX
# ============================================================

paired = pd.merge(
    weighted_df,
    dpo_df,
    on=["part", "index"],
    how="inner",
    suffixes=(
        "_weighted_sft",
        "_dpo"
    )
)

print(
    "\nNumber of paired observations:",
    len(paired)
)


# ============================================================
# 8. CHECK DUPLICATES
# ============================================================

duplicates = paired.duplicated(
    subset=["part", "index"]
).sum()

print(
    "Duplicate pairs:",
    duplicates
)


if len(paired) != 68:

    raise ValueError(
        f"Expected 68 pairs but found {len(paired)}."
    )

if duplicates != 0:

    raise ValueError(
        "Duplicate (part, index) pairs detected."
    )


# ============================================================
# 9. DESCRIPTIVE STATISTICS
# ============================================================

summary = pd.DataFrame({

    "N": [
        len(weighted_df),
        len(dpo_df)
    ],

    "Mean": [
        weighted_df["average"].mean(),
        dpo_df["average"].mean()
    ],

    "Median": [
        weighted_df["average"].median(),
        dpo_df["average"].median()
    ],

    "SD": [
        weighted_df["average"].std(),
        dpo_df["average"].std()
    ],

    "Min": [
        weighted_df["average"].min(),
        dpo_df["average"].min()
    ],

    "Q1": [
        weighted_df["average"].quantile(0.25),
        dpo_df["average"].quantile(0.25)
    ],

    "Q3": [
        weighted_df["average"].quantile(0.75),
        dpo_df["average"].quantile(0.75)
    ],

    "Max": [
        weighted_df["average"].max(),
        dpo_df["average"].max()
    ]
}, index=[
    "Weighted SFT",
    "DPO"
])

summary["IQR"] = (
    summary["Q3"]
    - summary["Q1"]
)


print("\n" + "=" * 80)
print("DESCRIPTIVE STATISTICS")
print("=" * 80)

print(
    summary.round(4).to_string()
)


# ============================================================
# 10. PAIRED DIFFERENCES
# ============================================================

paired["difference"] = (
    paired["average_dpo"]
    - paired["average_weighted_sft"]
)

differences = paired["difference"].values


# ============================================================
# 11. DIFFERENCE STATISTICS
# ============================================================

mean_difference = differences.mean()

median_difference = np.median(
    differences
)

sd_difference = differences.std(
    ddof=1
)

print("\n" + "=" * 80)
print("PAIRED DIFFERENCE: DPO - WEIGHTED SFT")
print("=" * 80)

print(
    f"Mean difference   : {mean_difference:.4f}"
)

print(
    f"Median difference : {median_difference:.4f}"
)

print(
    f"SD difference     : {sd_difference:.4f}"
)


# ============================================================
# 12. PAIRED T-TEST
# ============================================================

t_stat, t_p = ttest_rel(
    paired["average_dpo"],
    paired["average_weighted_sft"]
)

print("\n" + "=" * 80)
print("PAIRED T-TEST")
print("=" * 80)

print(
    f"t-statistic : {t_stat:.4f}"
)

print(
    f"p-value     : {t_p:.6f}"
)


# ============================================================
# 13. WILCOXON SIGNED-RANK TEST
# ============================================================

w_stat, w_p = wilcoxon(
    paired["average_dpo"],
    paired["average_weighted_sft"],
    alternative="two-sided"
)

print("\n" + "=" * 80)
print("WILCOXON SIGNED-RANK TEST")
print("=" * 80)

print(
    f"W-statistic : {w_stat:.4f}"
)

print(
    f"p-value     : {w_p:.6f}"
)


# ============================================================
# 14. 95% CI FOR MEAN PAIRED DIFFERENCE
# ============================================================

n = len(differences)

standard_error = (
    sd_difference / np.sqrt(n)
)

critical_value = t.ppf(
    0.975,
    df=n - 1
)

ci_lower = (
    mean_difference
    - critical_value * standard_error
)

ci_upper = (
    mean_difference
    + critical_value * standard_error
)

print("\n" + "=" * 80)
print("95% CI FOR MEAN DIFFERENCE")
print("=" * 80)

print(
    f"Mean difference : {mean_difference:.4f}"
)

print(
    f"95% CI          : "
    f"[{ci_lower:.4f}, {ci_upper:.4f}]"
)


# ============================================================
# 15. COHEN'S dz
# ============================================================

cohens_dz = (
    mean_difference
    / sd_difference
)

print("\n" + "=" * 80)
print("EFFECT SIZE")
print("=" * 80)

print(
    f"Cohen's dz : {cohens_dz:.4f}"
)


# ============================================================
# 16. LONG FORMAT FOR PLOTTING
# ============================================================

plot_df = pd.concat([

    weighted_df[
        ["part", "index", "average"]
    ].assign(
        model="Weighted SFT"
    ),

    dpo_df[
        ["part", "index", "average"]
    ].assign(
        model="DPO"
    )

], ignore_index=True)


# ============================================================
# 17. WEIGHTED SFT DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

sns.histplot(
    weighted_df["average"],
    bins=20,
    kde=True
)

plt.xlabel(
    "Average Factuality Score"
)

plt.ylabel(
    "Number of Reports"
)

plt.title(
    "Weighted SFT — Average Factuality Score Distribution"
)

plt.tight_layout()
plt.show()


# ============================================================
# 18. DPO DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

sns.histplot(
    dpo_df["average"],
    bins=20,
    kde=True
)

plt.xlabel(
    "Average Factuality Score"
)

plt.ylabel(
    "Number of Reports"
)

plt.title(
    "DPO — Average Factuality Score Distribution"
)

plt.tight_layout()
plt.show()


# ============================================================
# 19. OVERLAID DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    weighted_df["average"],
    bins=20,
    kde=True,
    stat="density",
    alpha=0.35,
    label="Weighted SFT"
)

sns.histplot(
    dpo_df["average"],
    bins=20,
    kde=True,
    stat="density",
    alpha=0.35,
    label="DPO"
)

plt.xlabel(
    "Average Factuality Score"
)

plt.ylabel(
    "Density"
)

plt.title(
    "Average Factuality Score Distribution: "
    "Weighted SFT vs DPO"
)

plt.legend()

plt.tight_layout()
plt.show()


# ============================================================
# 20. BOX PLOT
# ============================================================

plt.figure(figsize=(7, 5))

sns.boxplot(
    data=plot_df,
    x="model",
    y="average"
)

plt.xlabel("Model")
plt.ylabel("Average Factuality Score")

plt.title(
    "Average Factuality Scores: Weighted SFT vs DPO"
)

plt.tight_layout()
plt.show()


# ============================================================
# 21. PAIRED DIFFERENCE DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

sns.histplot(
    differences,
    bins=20,
    kde=True
)

plt.axvline(
    0,
    linestyle="--"
)

plt.xlabel(
    "Difference in Average Factuality Score\n"
    "(DPO − Weighted SFT)"
)

plt.ylabel(
    "Number of Paired Reports"
)

plt.title(
    "Paired Differences: DPO vs Weighted SFT"
)

plt.tight_layout()
plt.show()


# ============================================================
# 22. FINAL RESULT
# ============================================================

print("\n" + "=" * 80)
print("FINAL RESULT")
print("=" * 80)

print(
    f"N paired reports = {len(paired)}"
)

print(
    f"Weighted SFT mean = "
    f"{weighted_df['average'].mean():.4f}"
)

print(
    f"DPO mean = "
    f"{dpo_df['average'].mean():.4f}"
)

print(
    f"DPO - Weighted SFT = "
    f"{mean_difference:.4f}"
)

print(
    f"95% CI = "
    f"[{ci_lower:.4f}, {ci_upper:.4f}]"
)

print(
    f"Paired t-test p = "
    f"{t_p:.6f}"
)

print(
    f"Wilcoxon p = "
    f"{w_p:.6f}"
)

print(
    f"Cohen's dz = "
    f"{cohens_dz:.4f}"
)

print("\nAnalysis complete.")
```

```python
# ============================================================
# Synthetic Consultation Diversity Analysis
# First 500 consultations
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from collections import Counter
from itertools import combinations

# NLP
import nltk
from nltk import word_tokenize, pos_tag
from nltk.util import ngrams

# Semantic embeddings
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity


# ------------------------------------------------------------
# 1. Select first 500 consultations
# ------------------------------------------------------------

consultations = (
    df["consultation"]
    .dropna()
    .astype(str)
    .head(500)
    .tolist()
)

print(f"Number of consultations analysed: {len(consultations)}")


# ------------------------------------------------------------
# 2. Download required NLTK resources if needed
# ------------------------------------------------------------

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger")
nltk.download("averaged_perceptron_tagger_eng")


# ------------------------------------------------------------
# 3. Tokenisation
# ------------------------------------------------------------

tokenized = [
    word_tokenize(text.lower())
    for text in consultations
]

# Keep alphabetic tokens for lexical analysis
word_tokens = [
    [word for word in tokens if word.isalpha()]
    for tokens in tokenized
]

all_tokens = [
    word
    for consultation in word_tokens
    for word in consultation
]

print(f"Total word tokens: {len(all_tokens)}")


# ============================================================
# METRIC 1: DISTINCT-2
# ============================================================

# Construct all bigrams across the 500 consultations
bigrams = list(ngrams(all_tokens, 2))

unique_bigrams = set(bigrams)

distinct_2 = len(unique_bigrams) / len(bigrams) if bigrams else 0

print("\n--- Lexical Diversity ---")
print(f"Distinct-2: {distinct_2:.4f}")


# ============================================================
# METRIC 2: RESPONSE-LENGTH VARIATION
# ============================================================

# Number of words in each consultation
lengths = np.array([
    len(tokens)
    for tokens in word_tokens
])

length_mean = lengths.mean()
length_std = lengths.std(ddof=1)
length_min = lengths.min()
length_median = np.median(lengths)
length_max = lengths.max()

print("\n--- Response Length ---")
print(f"Mean:   {length_mean:.2f}")
print(f"SD:     {length_std:.2f}")
print(f"Min:    {length_min}")
print(f"Median: {length_median:.2f}")
print(f"Max:    {length_max}")


# ============================================================
# METRIC 3: POS SHANNON ENTROPY
# ============================================================

all_pos_tags = []

for tokens in word_tokens:
    if tokens:
        tagged = pos_tag(tokens)
        all_pos_tags.extend([tag for _, tag in tagged])

pos_counts = Counter(all_pos_tags)

total_pos = sum(pos_counts.values())

pos_probabilities = np.array([
    count / total_pos
    for count in pos_counts.values()
])

# Shannon entropy
pos_entropy = -np.sum(
    pos_probabilities * np.log2(pos_probabilities)
)

print("\n--- Grammatical Diversity ---")
print(f"POS Shannon Entropy: {pos_entropy:.4f}")

print("\nPOS distribution:")
for tag, count in pos_counts.most_common():
    print(f"{tag:8s}: {count}")


# ============================================================
# METRIC 4: SEMANTIC DIVERSITY
# ============================================================

print("\n--- Semantic Diversity ---")
print("Loading embedding model...")

# Sentence-transformers model
model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

# Generate one embedding per consultation
embeddings = model.encode(
    consultations,
    show_progress_bar=True,
    normalize_embeddings=True
)

print("Embeddings generated.")


# ------------------------------------------------------------
# Calculate pairwise cosine similarities
# ------------------------------------------------------------

similarity_matrix = cosine_similarity(embeddings)

# Extract only the upper triangle
# This avoids comparing a consultation with itself
# and avoids counting each pair twice.

n = len(consultations)

pairwise_similarities = similarity_matrix[
    np.triu_indices(n, k=1)
]

semantic_similarity_mean = pairwise_similarities.mean()
semantic_similarity_std = pairwise_similarities.std(ddof=1)
semantic_similarity_min = pairwise_similarities.min()
semantic_similarity_median = np.median(pairwise_similarities)
semantic_similarity_max = pairwise_similarities.max()

print(f"Number of consultation pairs: {len(pairwise_similarities)}")
print(f"Mean cosine similarity:   {semantic_similarity_mean:.4f}")
print(f"SD:                        {semantic_similarity_std:.4f}")
print(f"Min:                       {semantic_similarity_min:.4f}")
print(f"Median:                    {semantic_similarity_median:.4f}")
print(f"Max:                       {semantic_similarity_max:.4f}")


# ============================================================
# SUMMARY TABLE
# ============================================================

results = pd.DataFrame({
    "Metric": [
        "Distinct-2",
        "Response length (words)",
        "POS Shannon entropy",
        "Pairwise cosine similarity"
    ],

    "Mean / Result": [
        distinct_2,
        length_mean,
        pos_entropy,
        semantic_similarity_mean
    ],

    "SD": [
        np.nan,
        length_std,
        np.nan,
        semantic_similarity_std
    ],

    "Min": [
        np.nan,
        length_min,
        np.nan,
        semantic_similarity_min
    ],

    "Median": [
        np.nan,
        length_median,
        np.nan,
        semantic_similarity_median
    ],

    "Max": [
        np.nan,
        length_max,
        np.nan,
        semantic_similarity_max
    ]
})

print("\n\n============================================================")
print("FINAL DIVERSITY RESULTS")
print("============================================================")

print(results.to_string(index=False))


# ============================================================
# FIGURE: DISTRIBUTION OF PAIRWISE SEMANTIC SIMILARITY
# ============================================================

plt.figure(figsize=(8, 5))

plt.hist(
    pairwise_similarities,
    bins=50
)

plt.xlabel("Pairwise Cosine Similarity")
plt.ylabel("Number of Consultation Pairs")
plt.title("Distribution of Pairwise Semantic Similarity")

plt.tight_layout()
plt.show()


# ============================================================
# OPTIONAL: RESPONSE LENGTH DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

plt.hist(
    lengths,
    bins=30
)

plt.xlabel("Consultation Length (words)")
plt.ylabel("Number of Consultations")
plt.title("Distribution of Consultation Lengths")

plt.tight_layout()
plt.show()
```

```python
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# 1. Select first 500 consultations
# ============================================================

consultations = (
    df["consultation"]
    .dropna()
    .astype(str)
    .head(500)
    .tolist()
)

print(f"Number of consultations: {len(consultations)}")


# ============================================================
# 2. Tokenisation
# ============================================================

def tokenize(text):
    """
    Convert consultation text into lowercase word tokens.
    """
    return re.findall(r"\b[a-zA-Z]+(?:['’][a-zA-Z]+)?\b", text.lower())


tokenized_consultations = [
    tokenize(text)
    for text in consultations
]


# ============================================================
# 3. MATTR
# ============================================================

def calculate_mattr(tokens, window_size=50):
    """
    Calculate Moving-Average Type-Token Ratio (MATTR).

    MATTR is the average TTR across moving windows.
    """

    if len(tokens) < window_size:
        # If consultation is shorter than the window,
        # calculate ordinary TTR
        return len(set(tokens)) / len(tokens) if tokens else np.nan

    ttr_values = []

    for i in range(len(tokens) - window_size + 1):

        window = tokens[i:i + window_size]

        unique_words = len(set(window))
        total_words = len(window)

        ttr = unique_words / total_words

        ttr_values.append(ttr)

    return np.mean(ttr_values)


# Calculate MATTR for every consultation

mattr_values = np.array([
    calculate_mattr(tokens, window_size=50)
    for tokens in tokenized_consultations
])


# ============================================================
# 4. MTLD
# ============================================================

def calculate_mtld(tokens, threshold=0.72):
    """
    Calculate Measure of Textual Lexical Diversity (MTLD).

    Higher MTLD = greater lexical diversity.
    """

    if len(tokens) == 0:
        return np.nan

    def process_sequence(sequence):

        factors = 0
        start = 0
        types = set()

        for i, word in enumerate(sequence):

            types.add(word)

            token_count = i - start + 1
            ttr = len(types) / token_count

            if ttr <= threshold:

                factors += 1

                start = i + 1
                types = set()

        # Handle remaining partial factor
        remaining_tokens = len(sequence) - start

        if remaining_tokens > 0:

            remaining_types = len(
                set(sequence[start:])
            )

            remaining_ttr = (
                remaining_types / remaining_tokens
            )

            if remaining_ttr != 1:

                factors += (
                    (1 - remaining_ttr)
                    / (1 - threshold)
                )

        return (
            len(sequence) / factors
            if factors > 0
            else len(sequence)
        )

    # Forward MTLD
    forward = process_sequence(tokens)

    # Reverse MTLD
    reverse = process_sequence(tokens[::-1])

    # Average forward and reverse estimates
    return np.mean([forward, reverse])


# Calculate MTLD for every consultation

mtld_values = np.array([
    calculate_mtld(tokens)
    for tokens in tokenized_consultations
])


# ============================================================
# 5. Summary statistics
# ============================================================

mattr_mean = np.nanmean(mattr_values)
mattr_sd = np.nanstd(mattr_values, ddof=1)
mattr_min = np.nanmin(mattr_values)
mattr_median = np.nanmedian(mattr_values)
mattr_max = np.nanmax(mattr_values)

mtld_mean = np.nanmean(mtld_values)
mtld_sd = np.nanstd(mtld_values, ddof=1)
mtld_min = np.nanmin(mtld_values)
mtld_median = np.nanmedian(mtld_values)
mtld_max = np.nanmax(mtld_values)


# ============================================================
# 6. Print results
# ============================================================

print("\n" + "=" * 60)
print("LEXICAL DIVERSITY RESULTS")
print("=" * 60)

print("\nMATTR")
print("-" * 60)
print(f"Mean    : {mattr_mean:.4f}")
print(f"SD      : {mattr_sd:.4f}")
print(f"Minimum : {mattr_min:.4f}")
print(f"Median  : {mattr_median:.4f}")
print(f"Maximum : {mattr_max:.4f}")

print("\nMTLD")
print("-" * 60)
print(f"Mean    : {mtld_mean:.2f}")
print(f"SD      : {mtld_sd:.2f}")
print(f"Minimum : {mtld_min:.2f}")
print(f"Median  : {mtld_median:.2f}")
print(f"Maximum : {mtld_max:.2f}")


# ============================================================
# 7. Results table
# ============================================================

lexical_results = pd.DataFrame({
    "Metric": [
        "MATTR",
        "MTLD"
    ],

    "Mean": [
        mattr_mean,
        mtld_mean
    ],

    "SD": [
        mattr_sd,
        mtld_sd
    ],

    "Min": [
        mattr_min,
        mtld_min
    ],

    "Median": [
        mattr_median,
        mtld_median
    ],

    "Max": [
        mattr_max,
        mtld_max
    ]
})

print("\n" + "=" * 60)
print("FINAL LEXICAL DIVERSITY TABLE")
print("=" * 60)

print(
    lexical_results.to_string(index=False)
)


# ============================================================
# 8. MATTR distribution
# ============================================================

plt.figure(figsize=(8, 5))

plt.hist(
    mattr_values[~np.isnan(mattr_values)],
    bins=30
)

plt.xlabel("MATTR")
plt.ylabel("Number of Consultations")
plt.title("Distribution of MATTR Across Synthetic Consultations")

plt.tight_layout()
plt.show()


# ============================================================
# 9. MTLD distribution
# ============================================================

plt.figure(figsize=(8, 5))

plt.hist(
    mtld_values[~np.isnan(mtld_values)],
    bins=30
)

plt.xlabel("MTLD")
plt.ylabel("Number of Consultations")
plt.title("Distribution of MTLD Across Synthetic Consultations")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd


# ============================================================
# ADD part AND index TO df1
# PRESERVES THE EXACT ORIGINAL ROW ORDER
# ============================================================

# Safety check
assert len(df1) == 300, f"Expected 300 rows, found {len(df1)}"


# Make a copy without changing row order
df1 = df1.copy()


# Generate sequential positions based on CURRENT row order
row_position = np.arange(len(df1))


# Part: every 50 consecutive rows
# 1-50 → 01
# 51-100 → 02
# ...
# 251-300 → 06
df1["part"] = (
    (row_position // 50 + 1)
    .astype(str)
)

df1["part"] = df1["part"].str.zfill(2)


# Index repeats from 1 to 50 for each part
df1["index"] = (
    row_position % 50 + 1
)


# ============================================================
# MOVE part AND index TO THE FRONT
# THIS DOES NOT CHANGE ROW ORDER
# ============================================================

columns = ["part", "index"] + [
    col for col in df1.columns
    if col not in ["part", "index"]
]

df1 = df1[columns]


# ============================================================
# VALIDATION
# ============================================================

print("=" * 70)
print("DF1 PART AND INDEX CREATION")
print("=" * 70)

print(f"\nTotal rows: {len(df1)}")

print("\nPart distribution:")
print(
    df1.groupby("part")["index"]
    .agg(["count", "min", "max"])
)

print("\nBoundary check:")
print(
    df1.iloc[
        [0, 49, 50, 99, 100, 149, 150, 199, 200, 249, 250, 299]
    ][["part", "index"]]
)
```

```python
import pandas as pd
import os


# ============================================================
# PATH
# ============================================================

BASE_PATH = (
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation"
    r"\Factuality_temp\Organising_file_1"
)


# ============================================================
# STORE MATCHED DATAFRAMES
# ============================================================

matched_parts = []


# ============================================================
# LOOP THROUGH PARTS 01 TO 06
# ============================================================

for part_num in range(1, 7):

    part = f"{part_num:02d}"

    file_path = os.path.join(
        BASE_PATH,
        f"factuality_scores_weighted_sft_qwen25_1.5b_part2_50_{part}.jsonl"
    )

    print("=" * 70)
    print(f"PROCESSING PART {part}")
    print("=" * 70)

    # --------------------------------------------------------
    # Load JSONL file
    # --------------------------------------------------------

    if not os.path.exists(file_path):
        print(f"File not found: {file_path}")
        continue

    part_df = pd.read_json(
        file_path,
        lines=True
    )

    print(f"Rows in JSONL: {len(part_df)}")
    print(f"Columns: {part_df.columns.tolist()}")


    # --------------------------------------------------------
    # Ensure index is integer
    # --------------------------------------------------------

    part_df["index"] = part_df["index"].astype(int)


    # --------------------------------------------------------
    # Get available indices from JSONL
    # --------------------------------------------------------

    available_indices = set(
        part_df["index"]
    )


    # --------------------------------------------------------
    # Filter df1
    # Keep rows from this part whose index exists in JSONL
    # --------------------------------------------------------

    matched = df1[
        (df1["part"] == part) &
        (df1["index"].isin(available_indices))
    ].copy()


    print(f"Matched rows from df1: {len(matched)}")


    # --------------------------------------------------------
    # Store
    # --------------------------------------------------------

    matched_parts.append(matched)


# ============================================================
# COMBINE ALL MATCHED CASES
# ============================================================

df1_108 = pd.concat(
    matched_parts,
    ignore_index=True
)


# ============================================================
# VALIDATION
# ============================================================

print("\n" + "=" * 80)
print("FINAL MATCHED DATASET")
print("=" * 80)

print(f"\nTotal matched rows: {len(df1_108)}")

print("\nPart distribution:")
print(
    df1_108.groupby("part")["index"]
    .count()
)

print("\nFirst few rows:")
print(
    df1_108[
        ["part", "index"]
    ].head(10)
)

print("\nLast few rows:")
print(
    df1_108[
        ["part", "index"]
    ].tail(10)
)
```

```python
import pandas as pd
import os


# ============================================================
# PATH
# ============================================================

BASE_PATH = (
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation"
    r"\Factuality_temp\Organising_file_1"
)


# ============================================================
# FIND INTERSECTION OF DF1 WITH ALL PART JSONL FILES
# ============================================================

matched_keys = []


for part_num in range(1, 7):

    part = f"{part_num:02d}"

    file_path = os.path.join(
        BASE_PATH,
        f"factuality_scores_weighted_sft_qwen25_1.5b_part2_50_{part}.jsonl"
    )

    part_df = pd.read_json(
        file_path,
        lines=True
    )

    part_df["index"] = part_df["index"].astype(int)

    # Keep only the keys that actually exist in the JSONL
    keys = pd.DataFrame({
        "part": part,
        "index": part_df["index"]
    })

    matched_keys.append(keys)


# ============================================================
# COMBINE ALL JSONL KEYS
# ============================================================

jsonl_keys = pd.concat(
    matched_keys,
    ignore_index=True
).drop_duplicates(
    subset=["part", "index"]
)


# ============================================================
# MATCH WITH DF1
# ============================================================

df1_108 = df1.merge(
    jsonl_keys,
    on=["part", "index"],
    how="inner",
    sort=False
)


# ============================================================
# IMPORTANT:
# RESTORE ORIGINAL DF1 ORDER
# ============================================================

df1_108 = df1_108.sort_index()


# ============================================================
# CHECK
# ============================================================

print("=" * 70)
print("INTERSECTION")
print("=" * 70)

print(f"\nOriginal df1: {len(df1)}")
print(f"JSONL unique keys: {len(jsonl_keys)}")
print(f"Final intersection: {len(df1_108)}")

print("\nPart distribution:")
print(
    df1_108.groupby("part")["index"]
    .agg(["count", "min", "max"])
)

print("\nFirst 20 cases:")
print(
    df1_108[
        ["part", "index"]
    ].head(20).to_string(index=False)
)
```
