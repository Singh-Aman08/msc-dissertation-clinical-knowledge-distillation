```python
import glob
import json
import pandas as pd
import os

# --------------------------------------------------
# Paths
# --------------------------------------------------

factuality_pattern = (
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation"
    r"\Factuality_temp\Organising_file_1"
    r"\factuality_scores_[baseline_qwen25_15b]_part2_*.jsonl"
)

coverage_pattern = (
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation"
    r"\Distillation"
    r"\coverage_scores_baseline_qwen25_15b_part2_*.jsonl"
)

factuality_files = sorted(glob.glob(factuality_pattern))
coverage_files = sorted(glob.glob(coverage_pattern))

print("Factuality files:", len(factuality_files))
print("Coverage files:", len(coverage_files))


# --------------------------------------------------
# Function to extract indices
# --------------------------------------------------

def get_indices(filename):

    indices = set()

    with open(filename, "r", encoding="utf-8") as f:

        for line in f:

            if not line.strip():
                continue

            data = json.loads(line)

            index = data.get("index")

            if index is not None:
                indices.add(index)

    return indices


# --------------------------------------------------
# Match 01↔01, 02↔02, ..., 06↔06
# --------------------------------------------------

results = []

for factuality_file, coverage_file in zip(
    factuality_files,
    coverage_files
):

    factuality_indices = get_indices(factuality_file)
    coverage_indices = get_indices(coverage_file)

    intersection = factuality_indices.intersection(
        coverage_indices
    )

    factuality_only = factuality_indices - coverage_indices
    coverage_only = coverage_indices - factuality_indices

    pair = os.path.basename(factuality_file).split("_")[-1].replace(
        ".jsonl", ""
    )

    results.append({
        "pair": pair,
        "factuality_count": len(factuality_indices),
        "coverage_count": len(coverage_indices),
        "intersection_count": len(intersection),
        "factuality_only": len(factuality_only),
        "coverage_only": len(coverage_only)
    })

    print(f"\n--- Part {pair} ---")
    print("Factuality indices :", len(factuality_indices))
    print("Coverage indices   :", len(coverage_indices))
    print("Intersection       :", len(intersection))
    print("Factuality only    :", len(factuality_only))
    print("Coverage only      :", len(coverage_only))

    print("Matched indices:")
    print(sorted(intersection))


# --------------------------------------------------
# Summary
# --------------------------------------------------

results_df = pd.DataFrame(results)

print("\n========================================")
print("INTERSECTION SUMMARY")
print("========================================")

print(results_df.to_string(index=False))

print("\nTotal matched across all six pairs:",
      results_df["intersection_count"].sum())
```

```python
import json
import os

# --------------------------------------------------
# Output folder = factuality folder
# --------------------------------------------------

output_folder = (
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation"
    r"\Factuality_temp\Organising_file_1"
)


# --------------------------------------------------
# Create intersection coverage files
# --------------------------------------------------

for factuality_file, coverage_file in zip(
    factuality_files,
    coverage_files
):

    # Get factuality indices
    factuality_indices = get_indices(factuality_file)

    # Part number
    part = os.path.basename(factuality_file).split("_")[-1].replace(
        ".jsonl", ""
    )

    # Output filename
    output_file = os.path.join(
        output_folder,
        f"coverage_scores_baseline_qwen25_15b_part2_{part}_intersection.jsonl"
    )

    matched_count = 0

    # Read coverage and keep only matching indices
    with open(coverage_file, "r", encoding="utf-8") as infile, \
         open(output_file, "w", encoding="utf-8") as outfile:

        for line in infile:

            if not line.strip():
                continue

            data = json.loads(line)

            index = data.get("index")

            if index in factuality_indices:
                outfile.write(
                    json.dumps(data, ensure_ascii=False) + "\n"
                )
                matched_count += 1

    print(
        f"Part {part}: {matched_count} records saved → "
        f"{output_file}"
    )

print("\nDone.")
```

```python
import re

def clean_report(text):
    if pd.isna(text):
        return text

    # Remove markdown headers
    text = re.sub(r'^#+\s*', '', text, flags=re.MULTILINE)

    # Remove bullet points
    text = re.sub(r'^\s*[-*]\s+', '', text, flags=re.MULTILINE)

    # Remove bold markdown
    text = re.sub(r'\*\*', '', text)

    # Remove numbered section markers: 1), 2), 10), etc.
    text = re.sub(r'\b\d+\)\s*', '', text)

    # Remove lettered subsection markers: a), b), etc.
    text = re.sub(r'\b[a-zA-Z]\)\s*', '', text)

    # Replace all newlines with spaces
    text = re.sub(r'\s*\n\s*', ' ', text)

    # Remove multiple spaces
    text = re.sub(r'\s+', ' ', text)

    return text.strip()


# Create cleaned reports column
dataset["Cleaned_Report"] = dataset["Report"].apply(clean_report)

# Check
print(dataset[["Report", "Cleaned_Report"]].head(2).to_string())
```

```python
import json
import os


# ============================================================
# DIRECTORIES
# ============================================================

organising_dir = r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation\Factuality_temp\Organising_file_1"

coverage_dir = r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation\Distillation"


# ============================================================
# FILE NAMES
# ============================================================

sft_factuality_files = [
    f"factuality_scores_[normal_sft_qwen25_1.5b]_part2_100_{i:02d}.jsonl"
    for i in range(1, 7)
]

sft_coverage_files = [
    f"coverage_scores_normal_sft_qwen25_1.5b_part2_100_{i:02d}.jsonl"
    for i in range(1, 7)
]


# ============================================================
# FUNCTION TO READ JSONL
# ============================================================

def read_jsonl(filepath):

    records = []

    with open(filepath, "r", encoding="utf-8") as f:

        for line in f:

            if line.strip():
                records.append(json.loads(line))

    return records


# ============================================================
# FUNCTION TO GET INDEX
# ============================================================

def get_index(record):

    return int(record["index"])


# ============================================================
# PROCESS EACH PART
# ============================================================

total_factuality = 0
total_coverage = 0
total_intersection = 0

print("=" * 70)
print("SFT FACTUALITY ∩ COVERAGE")
print("=" * 70)


for part in range(1, 7):

    factuality_path = os.path.join(
        organising_dir,
        sft_factuality_files[part - 1]
    )

    coverage_path = os.path.join(
        coverage_dir,
        sft_coverage_files[part - 1]
    )

    # --------------------------------------------------------
    # Read files
    # --------------------------------------------------------

    factuality_records = read_jsonl(factuality_path)
    coverage_records = read_jsonl(coverage_path)

    factuality_indices = {
        get_index(record)
        for record in factuality_records
    }

    coverage_indices = {
        get_index(record)
        for record in coverage_records
    }

    # --------------------------------------------------------
    # Intersection
    # --------------------------------------------------------

    common_indices = factuality_indices & coverage_indices

    total_factuality += len(factuality_indices)
    total_coverage += len(coverage_indices)
    total_intersection += len(common_indices)

    # --------------------------------------------------------
    # Keep ONLY coverage records in the intersection
    # --------------------------------------------------------

    intersected_coverage = [
        record
        for record in coverage_records
        if get_index(record) in common_indices
    ]

    # Keep original order
    intersected_coverage.sort(
        key=lambda x: get_index(x)
    )

    # --------------------------------------------------------
    # Output file
    # --------------------------------------------------------

    output_filename = (
        f"coverage_scores_[normal_sft_qwen25_1.5b]"
        f"_part2_100_{part:02d}_intersection.jsonl"
    )

    output_path = os.path.join(
        organising_dir,
        output_filename
    )

    # --------------------------------------------------------
    # WRITE NEW FILE
    # --------------------------------------------------------

    with open(output_path, "w", encoding="utf-8") as f:

        for record in intersected_coverage:

            f.write(
                json.dumps(record, ensure_ascii=False)
                + "\n"
            )

    # --------------------------------------------------------
    # Print results
    # --------------------------------------------------------

    print(f"\nPART {part:02d}")
    print("-" * 50)

    print(
        f"SFT factuality records : "
        f"{len(factuality_indices)}"
    )

    print(
        f"SFT coverage records   : "
        f"{len(coverage_indices)}"
    )

    print(
        f"Intersection           : "
        f"{len(common_indices)}"
    )

    print(
        f"Output file            : "
        f"{output_filename}"
    )


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("FINAL SUMMARY")
print("=" * 70)

print(f"Total SFT factuality records : {total_factuality}")
print(f"Total SFT coverage records   : {total_coverage}")
print(f"Total intersection           : {total_intersection}")

print("\nOriginal coverage files were NOT modified.")
print("New intersection files were created in Organising_file_1.")

if total_intersection == 108:
    print("\n✓ SUCCESS: Final SFT intersection = 108")
else:
    print(
        f"\n⚠️ Check result: expected 108, "
        f"but found {total_intersection}"
    )
```

```python
import re
import pandas as pd


def clean_report(text):
    if pd.isna(text):
        return text

    # Remove markdown headers
    text = re.sub(
        r'^#+\s*',
        '',
        text,
        flags=re.MULTILINE
    )

    # Remove bullet points
    text = re.sub(
        r'^\s*[-*]\s+',
        '',
        text,
        flags=re.MULTILINE
    )

    # Remove bold markdown
    text = re.sub(
        r'\*\*',
        '',
        text
    )

    # Remove numbered section markers: 1), 2), 10), etc.
    text = re.sub(
        r'\b\d+\)\s*',
        '',
        text
    )

    # Remove lettered subsection markers: a), b), etc.
    text = re.sub(
        r'\b[a-zA-Z]\)\s*',
        '',
        text
    )

    # Replace all newlines with spaces
    text = re.sub(
        r'\s*\n\s*',
        ' ',
        text
    )

    # Remove multiple spaces
    text = re.sub(
        r'\s+',
        ' ',
        text
    )

    return text.strip()


# ============================================================
# CREATE CLEANED REPORTS
# ============================================================

sft_dataset["Cleaned_Report"] = (
    sft_dataset["Report"].apply(clean_report)
)


# ============================================================
# CHECK
# ============================================================

print(
    sft_dataset[
        ["Report", "Cleaned_Report"]
    ].head(2).to_string()
)

print("\nShape:")
print(sft_dataset.shape)
```

```python
import numpy as np
import pandas as pd
from scipy.stats import wilcoxon
from statsmodels.stats.multitest import multipletests


# ============================================================
# 1. Prepare Baseline and SFT data
# ============================================================

baseline = dataset[
    [
        "part",
        "index",
        "Factuality Score",
        "coverage",
        "Conciseness Score"
    ]
].copy()

sft = sft_dataset[
    [
        "part",
        "index",
        "Factuality Score",
        "coverage",
        "Conciseness Score"
    ]
].copy()


# Ensure matching identifiers have the same type
baseline["part"] = baseline["part"].astype(str)
sft["part"] = sft["part"].astype(str)

baseline["index"] = baseline["index"].astype(int)
sft["index"] = sft["index"].astype(int)


# ============================================================
# 2. Match the 108 paired cases
# ============================================================

comparison = baseline.merge(
    sft,
    on=["part", "index"],
    suffixes=("_baseline", "_sft"),
    how="inner"
)

print("=" * 80)
print("PAIRED BASELINE vs STANDARD SFT COMPARISON")
print("=" * 80)

print(f"\nTotal matched cases: {len(comparison)}")


# ============================================================
# 3. Rank-biserial correlation for paired data
# ============================================================

def rank_biserial_paired(baseline_scores, sft_scores):

    baseline_scores = np.asarray(baseline_scores)
    sft_scores = np.asarray(sft_scores)

    # Difference is explicitly SFT - Baseline
    differences = sft_scores - baseline_scores

    # Remove zero differences
    nonzero = differences != 0
    differences = differences[nonzero]

    if len(differences) == 0:
        return 0.0

    # Rank absolute differences
    ranks = pd.Series(
        np.abs(differences)
    ).rank(
        method="average"
    ).to_numpy()

    positive_ranks = ranks[differences > 0].sum()
    negative_ranks = ranks[differences < 0].sum()

    # Signed rank-biserial correlation
    r_rb = (
        positive_ranks - negative_ranks
    ) / (
        positive_ranks + negative_ranks
    )

    return r_rb


# ============================================================
# 4. Define the three paired comparisons
# ============================================================

metrics = {
    "Factuality": (
        "Factuality Score_baseline",
        "Factuality Score_sft"
    ),

    "Coverage": (
        "coverage_baseline",
        "coverage_sft"
    ),

    "Conciseness": (
        "Conciseness Score_baseline",
        "Conciseness Score_sft"
    )
}


# ============================================================
# 5. Run Wilcoxon tests
# ============================================================

results = []

for dimension, (baseline_col, sft_col) in metrics.items():

    # --------------------------------------------------------
    # Keep only cases with BOTH scores available
    # --------------------------------------------------------

    temp = comparison[
        [baseline_col, sft_col]
    ].dropna()

    baseline_scores = temp[baseline_col].to_numpy()
    sft_scores = temp[sft_col].to_numpy()

    # --------------------------------------------------------
    # Paired difference: SFT - Baseline
    # --------------------------------------------------------

    differences = sft_scores - baseline_scores

    # --------------------------------------------------------
    # Wilcoxon signed-rank test
    # --------------------------------------------------------

    wilcoxon_result = wilcoxon(
        sft_scores,
        baseline_scores,
        alternative="two-sided",
        zero_method="wilcox",
        method="auto"
    )

    statistic = wilcoxon_result.statistic
    p_value = wilcoxon_result.pvalue

    # --------------------------------------------------------
    # Rank-biserial effect size
    # --------------------------------------------------------

    rank_biserial = rank_biserial_paired(
        baseline_scores,
        sft_scores
    )

    # --------------------------------------------------------
    # Store results
    # --------------------------------------------------------

    results.append({
        "Dimension": dimension,
        "N": len(temp),

        "Baseline Median": np.median(
            baseline_scores
        ),

        "SFT Median": np.median(
            sft_scores
        ),

        "Median Δ (SFT - Baseline)": np.median(
            differences
        ),

        "Wilcoxon W": statistic,

        "Raw p-value": p_value,

        "Rank-biserial r": rank_biserial
    })


# ============================================================
# 6. Create results dataframe
# ============================================================

results_df = pd.DataFrame(results)


# ============================================================
# 7. Holm multiple-comparison correction
# ============================================================

reject, p_adjusted, _, _ = multipletests(
    results_df["Raw p-value"],
    alpha=0.05,
    method="holm"
)

results_df["Holm-adjusted p-value"] = p_adjusted

results_df["Significant (α = 0.05)"] = reject


# ============================================================
# 8. Order columns
# ============================================================

results_df = results_df[
    [
        "Dimension",
        "N",
        "Baseline Median",
        "SFT Median",
        "Median Δ (SFT - Baseline)",
        "Wilcoxon W",
        "Raw p-value",
        "Holm-adjusted p-value",
        "Rank-biserial r",
        "Significant (α = 0.05)"
    ]
]


# ============================================================
# 9. Round numerical values
# ============================================================

results_df[
    [
        "Baseline Median",
        "SFT Median",
        "Median Δ (SFT - Baseline)",
        "Wilcoxon W",
        "Raw p-value",
        "Holm-adjusted p-value",
        "Rank-biserial r"
    ]
] = results_df[
    [
        "Baseline Median",
        "SFT Median",
        "Median Δ (SFT - Baseline)",
        "Wilcoxon W",
        "Raw p-value",
        "Holm-adjusted p-value",
        "Rank-biserial r"
    ]
].round(4)


# ============================================================
# 10. Print final dissertation-ready table
# ============================================================

print("\n" + "=" * 100)
print("WILCOXON SIGNED-RANK TEST WITH HOLM CORRECTION")
print("=" * 100)

print(
    results_df.to_string(index=False)
)


# ============================================================
# 11. Additional interpretation information
# ============================================================

print("\n" + "=" * 80)
print("INTERPRETATION")
print("=" * 80)

for _, row in results_df.iterrows():

    direction = (
        "SFT > Baseline"
        if row["Rank-biserial r"] > 0
        else
        "SFT < Baseline"
        if row["Rank-biserial r"] < 0
        else
        "No systematic difference"
    )

    significance = (
        "Statistically significant"
        if row["Significant (α = 0.05)"]
        else
        "Not statistically significant"
    )

    print(
        f"\n{row['Dimension']}: "
        f"{direction}; "
        f"{significance} "
        f"(Holm-adjusted p = "
        f"{row['Holm-adjusted p-value']:.4f}, "
        f"rank-biserial r = "
        f"{row['Rank-biserial r']:.4f})"
    )
```

```python
import re
import pandas as pd


# ============================================================
# CLEAN REPORT FUNCTION
# ============================================================

def clean_report(text):

    if pd.isna(text):
        return text

    # Remove markdown headers
    text = re.sub(
        r'^#+\s*',
        '',
        text,
        flags=re.MULTILINE
    )

    # Remove bullet points
    text = re.sub(
        r'^\s*[-*]\s+',
        '',
        text,
        flags=re.MULTILINE
    )

    # Remove bold markdown
    text = re.sub(
        r'\*\*',
        '',
        text
    )

    # Remove numbered section markers: 1), 2), 10), etc.
    text = re.sub(
        r'\b\d+\)\s*',
        '',
        text
    )

    # Remove lettered subsection markers: a), b), etc.
    text = re.sub(
        r'\b[a-zA-Z]\)\s*',
        '',
        text
    )

    # Replace all newlines with spaces
    text = re.sub(
        r'\s*\n\s*',
        ' ',
        text
    )

    # Remove multiple spaces
    text = re.sub(
        r'\s+',
        ' ',
        text
    )

    return text.strip()


# ============================================================
# CREATE CLEANED REPORTS FOR WEIGHTED SFT
# ============================================================

wght_dataframe["Cleaned_Report"] = (
    wght_dataframe["Report"].apply(clean_report)
)


# ============================================================
# CHECK
# ============================================================

print(
    wght_dataframe[
        ["Report", "Cleaned_Report"]
    ].head(2).to_string()
)

print("\nShape:")
print(wght_dataframe.shape)

print("\nMissing values:")
print(
    wght_dataframe["Cleaned_Report"].isna().sum()
)
```

```python
from transformers import AutoTokenizer
import pandas as pd


MODEL_ID = "Qwen/Qwen3-30B-A3B-Instruct-2507"


print("Loading tokenizer...", flush=True)

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_ID
)

print("Tokenizer loaded successfully\n", flush=True)


# ============================================================
# TOKEN LENGTH FUNCTION
# ============================================================

def get_token_length(text):

    if pd.isna(text):
        return 0

    tokens = tokenizer.encode(
        str(text),
        add_special_tokens=True
    )

    return len(tokens)


# ============================================================
# CALCULATE WEIGHTED SFT TOKEN LENGTHS
# ============================================================

print(
    "Calculating Weighted SFT token lengths...",
    flush=True
)

wght_dataframe["Cleaned_token_length"] = (
    wght_dataframe["Cleaned_Report"].apply(
        get_token_length
    )
)


print(
    "Weighted SFT token length calculation "
    "completed successfully.\n"
)


# ============================================================
# CHECK
# ============================================================

print(
    wght_dataframe[
        [
            "part",
            "index",
            "Cleaned_token_length"
        ]
    ].head(10)
)

print("\nShape:")
print(wght_dataframe.shape)

print("\nWeighted SFT token length statistics:")

print(
    wght_dataframe[
        "Cleaned_token_length"
    ].describe()
)
```

```python
import pandas as pd
import numpy as np

from scipy.stats import wilcoxon
from statsmodels.stats.multitest import multipletests


# ============================================================
# 1. SELECT RELEVANT COLUMNS
# ============================================================

sft = sft_dataset[
    [
        "part",
        "index",
        "Factuality Score",
        "coverage",
        "Conciseness Score"
    ]
].copy()

weighted = wght_dataframe[
    [
        "part",
        "index",
        "Factuality Score",
        "coverage",
        "Conciseness Score"
    ]
].copy()


# ============================================================
# 2. MATCH SFT AND WEIGHTED SFT
# ============================================================

comparison = pd.merge(
    sft,
    weighted,
    on=["part", "index"],
    how="inner",
    suffixes=("_SFT", "_Weighted_SFT")
)

print("=" * 70)
print("SFT vs WEIGHTED SFT")
print("=" * 70)

print("Number of matched cases:", len(comparison))


# ============================================================
# 3. RANK-BISERIAL CORRELATION
# ============================================================
#
# For paired Wilcoxon:
#
# r_rb = (W_positive - W_negative) / W_total
#
# Positive value  -> Weighted SFT tends to be higher
# Negative value  -> Standard SFT tends to be higher
#
# Zero differences are excluded, as they do not contribute
# to the Wilcoxon signed-rank statistic.
# ============================================================

def rank_biserial_paired(x, y):

    difference = np.asarray(y) - np.asarray(x)

    # Remove zero differences
    difference = difference[difference != 0]

    if len(difference) == 0:
        return 0.0

    abs_difference = np.abs(difference)

    # Rank absolute differences
    ranks = pd.Series(abs_difference).rank(
        method="average"
    ).to_numpy()

    positive_ranks = ranks[difference > 0].sum()
    negative_ranks = ranks[difference < 0].sum()

    total_ranks = positive_ranks + negative_ranks

    if total_ranks == 0:
        return 0.0

    return (positive_ranks - negative_ranks) / total_ranks


# ============================================================
# 4. METRICS TO COMPARE
# ============================================================

metrics = {
    "Factuality": (
        "Factuality Score_SFT",
        "Factuality Score_Weighted_SFT"
    ),
    "Coverage": (
        "coverage_SFT",
        "coverage_Weighted_SFT"
    ),
    "Conciseness": (
        "Conciseness Score_SFT",
        "Conciseness Score_Weighted_SFT"
    )
}


# ============================================================
# 5. RUN WILCOXON TESTS
# ============================================================

results = []
raw_p_values = []


for metric, (sft_col, weighted_col) in metrics.items():

    # --------------------------------------------------------
    # Keep only cases where BOTH models have valid scores
    # --------------------------------------------------------

    temp = comparison[
        [sft_col, weighted_col]
    ].dropna()

    sft_values = temp[sft_col].to_numpy()
    weighted_values = temp[weighted_col].to_numpy()

    n = len(temp)

    # --------------------------------------------------------
    # Paired difference
    # Weighted SFT - SFT
    # --------------------------------------------------------

    differences = weighted_values - sft_values

    median_difference = np.median(differences)

    # --------------------------------------------------------
    # Wilcoxon signed-rank test
    # --------------------------------------------------------

    if np.all(differences == 0):

        statistic = 0.0
        p_value = 1.0

    else:

        wilcoxon_result = wilcoxon(
            weighted_values,
            sft_values,
            alternative="two-sided",
            zero_method="wilcox",
            method="auto"
        )

        statistic = wilcoxon_result.statistic
        p_value = wilcoxon_result.pvalue

    # --------------------------------------------------------
    # Rank-biserial effect size
    # --------------------------------------------------------

    effect_size = rank_biserial_paired(
        sft_values,
        weighted_values
    )

    results.append({
        "Metric": metric,
        "N": n,
        "SFT Median": np.median(sft_values),
        "Weighted SFT Median": np.median(weighted_values),
        "Median Δ": median_difference,
        "Wilcoxon W": statistic,
        "Raw p": p_value,
        "Rank-biserial r": effect_size
    })

    raw_p_values.append(p_value)


# ============================================================
# 6. HOLM MULTIPLE-COMPARISON CORRECTION
# ============================================================

reject, adjusted_p_values, _, _ = multipletests(
    raw_p_values,
    alpha=0.05,
    method="holm"
)


# ============================================================
# 7. ADD HOLM RESULTS
# ============================================================

for i, result in enumerate(results):

    result["Holm-adjusted p"] = adjusted_p_values[i]

    result["Significant"] = (
        "Yes"
        if reject[i]
        else "No"
    )


# ============================================================
# 8. CREATE FINAL RESULTS TABLE
# ============================================================

results_df = pd.DataFrame(results)

# Reorder columns
results_df = results_df[
    [
        "Metric",
        "N",
        "SFT Median",
        "Weighted SFT Median",
        "Median Δ",
        "Wilcoxon W",
        "Raw p",
        "Holm-adjusted p",
        "Rank-biserial r",
        "Significant"
    ]
]


# ============================================================
# 9. ROUND RESULTS
# ============================================================

results_display = results_df.copy()

results_display["SFT Median"] = (
    results_display["SFT Median"].round(4)
)

results_display["Weighted SFT Median"] = (
    results_display["Weighted SFT Median"].round(4)
)

results_display["Median Δ"] = (
    results_display["Median Δ"].round(4)
)

results_display["Wilcoxon W"] = (
    results_display["Wilcoxon W"].round(4)
)

results_display["Raw p"] = (
    results_display["Raw p"].round(6)
)

results_display["Holm-adjusted p"] = (
    results_display["Holm-adjusted p"].round(6)
)

results_display["Rank-biserial r"] = (
    results_display["Rank-biserial r"].round(4)
)


# ============================================================
# 10. PRINT FINAL TABLE
# ============================================================

print("\n")
print("=" * 110)
print("WILCOXON SIGNED-RANK TEST: SFT vs WEIGHTED SFT")
print("=" * 110)

print(
    results_display.to_string(
        index=False
    )
)


# ============================================================
# 11. INTERPRETATION
# ============================================================

print("\n")
print("=" * 70)
print("INTERPRETATION")
print("=" * 70)

print(
    "Δ = Weighted SFT − SFT"
)

print(
    "Positive Δ / positive rank-biserial r → "
    "Weighted SFT tends to outperform SFT."
)

print(
    "Negative Δ / negative rank-biserial r → "
    "SFT tends to outperform Weighted SFT."
)

print(
    "Statistical significance is assessed using "
    "Holm-adjusted p < 0.05."
)
```

```python
import json
import pandas as pd
from pathlib import Path


# ============================================================
# CONFIGURATION
# ============================================================

DPO_FOLDER = Path(
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation\DPO\Factuality_Score\Factuality_Score_1"
)

WEIGHTED_FOLDER = Path(
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation\DPO\Factuality_Score\Factuality_Score_1"
)

PARTS = ["01", "02", "03", "04", "05", "06"]


# ============================================================
# LOAD JSONL
# ============================================================

def load_jsonl(file_path):
    
    records = []
    
    with open(file_path, "r", encoding="utf-8") as f:
        for line in f:
            if line.strip():
                records.append(json.loads(line))
    
    return pd.DataFrame(records)


# ============================================================
# STORE RESULTS
# ============================================================

results = []

all_dpo_ids = set()
all_weighted_ids = set()


# ============================================================
# COMPARE EACH PART
# ============================================================

for part in PARTS:
    
    print("\n" + "=" * 70)
    print(f"CHECKING PART {part}")
    print("=" * 70)
    
    
    # --------------------------------------------------------
    # FILE NAMES
    # --------------------------------------------------------
    
    dpo_file = (
        DPO_FOLDER
        / f"factuality_scores_patient_report_wgt_dpo_{part}.jsonl"
    )
    
    weighted_file = (
        WEIGHTED_FOLDER
        / f"factuality_scores_weighted_sft_qwen25_1.5b_part2_50_{part}.jsonl"
    )
    
    
    # --------------------------------------------------------
    # CHECK FILE EXISTENCE
    # --------------------------------------------------------
    
    if not dpo_file.exists():
        print(f"⚠ DPO file not found: {dpo_file}")
        continue
    
    if not weighted_file.exists():
        print(f"⚠ Weighted SFT file not found: {weighted_file}")
        continue
    
    
    # --------------------------------------------------------
    # LOAD DATA
    # --------------------------------------------------------
    
    dpo_df = load_jsonl(dpo_file)
    weighted_df = load_jsonl(weighted_file)
    
    
    # --------------------------------------------------------
    # EXTRACT INDICES
    # --------------------------------------------------------
    
    dpo_indices = set(dpo_df["index"])
    weighted_indices = set(weighted_df["index"])
    
    
    # --------------------------------------------------------
    # INTERSECTION
    # --------------------------------------------------------
    
    common_indices = dpo_indices.intersection(
        weighted_indices
    )
    
    only_in_dpo = dpo_indices - weighted_indices
    
    only_in_weighted = (
        weighted_indices - dpo_indices
    )
    
    
    # --------------------------------------------------------
    # PRINT PART RESULTS
    # --------------------------------------------------------
    
    print(f"\nDPO cases: {len(dpo_indices)}")
    print(f"Weighted SFT cases: {len(weighted_indices)}")
    print(f"Intersection: {len(common_indices)}")
    print(f"Only in DPO: {len(only_in_dpo)}")
    print(
        f"Only in Weighted SFT: "
        f"{len(only_in_weighted)}"
    )
    
    
    if dpo_indices == weighted_indices:
        print("✓ PERFECT MATCH")
    else:
        print("⚠ NOT A PERFECT MATCH")
        
        if only_in_dpo:
            print(
                "Indices only in DPO:",
                sorted(only_in_dpo)
            )
        
        if only_in_weighted:
            print(
                "Indices only in Weighted SFT:",
                sorted(only_in_weighted)
            )
    
    
    # --------------------------------------------------------
    # STORE RESULTS
    # --------------------------------------------------------
    
    results.append({
        "Part": part,
        "DPO Cases": len(dpo_indices),
        "Weighted SFT Cases": len(weighted_indices),
        "Intersection": len(common_indices),
        "Only in DPO": len(only_in_dpo),
        "Only in Weighted SFT": len(only_in_weighted),
        "Perfect Match": dpo_indices == weighted_indices
    })
    
    
    # --------------------------------------------------------
    # CREATE UNIQUE (PART, INDEX) IDENTIFIERS
    # --------------------------------------------------------
    
    all_dpo_ids.update(
        {(part, idx) for idx in dpo_indices}
    )
    
    all_weighted_ids.update(
        {(part, idx) for idx in weighted_indices}
    )


# ============================================================
# CREATE SUMMARY TABLE
# ============================================================

summary_df = pd.DataFrame(results)

print("\n\n" + "=" * 70)
print("SUMMARY: WEIGHTED SFT vs DPO FACTUALITY")
print("=" * 70)

print(summary_df.to_string(index=False))


# ============================================================
# OVERALL INTERSECTION
# ============================================================

overall_intersection = (
    all_dpo_ids.intersection(all_weighted_ids)
)

overall_only_dpo = (
    all_dpo_ids - all_weighted_ids
)

overall_only_weighted = (
    all_weighted_ids - all_dpo_ids
)


print("\n" + "=" * 70)
print("OVERALL INTERSECTION ACROSS ALL PARTS")
print("=" * 70)

print(f"\nTotal DPO cases: {len(all_dpo_ids)}")
print(
    f"Total Weighted SFT cases: "
    f"{len(all_weighted_ids)}"
)
print(
    f"Overall intersection: "
    f"{len(overall_intersection)}"
)
print(
    f"Cases only in DPO: "
    f"{len(overall_only_dpo)}"
)
print(
    f"Cases only in Weighted SFT: "
    f"{len(overall_only_weighted)}"
)


# ============================================================
# FINAL MATCH CHECK
# ============================================================

if all_dpo_ids == all_weighted_ids:
    
    print("\n✓ OVERALL PERFECT MATCH")
    print(
        "DPO and Weighted SFT contain exactly the same "
        "(part, index) evaluation cases."
    )

else:
    
    print("\n⚠ OVERALL NOT A PERFECT MATCH")
    
    if overall_only_dpo:
        print("\nCases only in DPO:")
        print(sorted(overall_only_dpo))
    
    if overall_only_weighted:
        print("\nCases only in Weighted SFT:")
        print(sorted(overall_only_weighted))
```

```python
import json
import pandas as pd
from pathlib import Path


# ============================================================
# CONFIGURATION
# ============================================================

FACTUALITY_FOLDER = Path(
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation\DPO\Factuality_Score\Factuality_Score_1"
)

COVERAGE_FOLDER = Path(
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation\DPO\Coverage_Score"
)

PARTS = ["01", "02", "03", "04", "05", "06"]


# ============================================================
# LOAD JSONL
# ============================================================

def load_jsonl(file_path):
    records = []

    with open(file_path, "r", encoding="utf-8") as f:
        for line in f:
            if line.strip():
                records.append(json.loads(line))

    return pd.DataFrame(records)


# ============================================================
# CHECK INTERSECTION FOR EACH PART
# ============================================================

all_intersections = []

print("=" * 70)
print("CHECKING FACTUALITY ↔ COVERAGE INTERSECTION")
print("=" * 70)

for part in PARTS:

    factuality_file = (
        FACTUALITY_FOLDER
        / f"factuality_scores_patient_report_wgt_dpo_{part}.jsonl"
    )

    coverage_file = (
        COVERAGE_FOLDER
        / f"coverage_scores_wgt_dpo_{part}.jsonl"
    )

    # --------------------------------------------------------
    # Check files exist
    # --------------------------------------------------------

    if not factuality_file.exists():
        print(f"\n⚠ Missing factuality file: {part}")
        continue

    if not coverage_file.exists():
        print(f"\n⚠ Missing coverage file: {part}")
        continue

    # --------------------------------------------------------
    # Load
    # --------------------------------------------------------

    factuality_df = load_jsonl(factuality_file)
    coverage_df = load_jsonl(coverage_file)

    # --------------------------------------------------------
    # IDs
    # --------------------------------------------------------

    factuality_ids = set(factuality_df["index"])
    coverage_ids = set(coverage_df["index"])

    common_ids = factuality_ids.intersection(coverage_ids)

    # --------------------------------------------------------
    # Print intersection information
    # --------------------------------------------------------

    print(f"\nPART {part}")
    print("-" * 50)

    print("Factuality cases :", len(factuality_ids))
    print("Coverage cases   :", len(coverage_ids))
    print("Intersection     :", len(common_ids))

    # Optional: show unmatched cases
    factuality_only = factuality_ids - coverage_ids
    coverage_only = coverage_ids - factuality_ids

    print("Factuality only   :", len(factuality_only))
    print("Coverage only     :", len(coverage_only))

    # --------------------------------------------------------
    # Keep only intersection from factuality
    # --------------------------------------------------------

    common_factuality = factuality_df[
        factuality_df["index"].isin(common_ids)
    ].copy()

    # Add part
    common_factuality["part"] = part

    # --------------------------------------------------------
    # Keep report + identifiers only
    # --------------------------------------------------------

    common_factuality = common_factuality[
        ["part", "index", "report"]
    ]

    all_intersections.append(common_factuality)


# ============================================================
# COMBINE ALL INTERSECTIONS
# ============================================================

dpo_intersection = pd.concat(
    all_intersections,
    ignore_index=True
)


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("FINAL INTERSECTION")
print("=" * 70)

print("Total common cases:", len(dpo_intersection))

print("\nCases by part:")
print(
    dpo_intersection
    .groupby("part")
    .size()
)


# ============================================================
# DISPLAY RESULT
# ============================================================

print("\nFirst 10 intersected cases:")

print(
    dpo_intersection[
        ["part", "index", "report"]
    ].head(10).to_string()
)
```

```python
import json
from pathlib import Path


# ============================================================
# CONFIGURATION
# ============================================================

FACTUALITY_FOLDER = Path(
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation\DPO\Factuality_Score\Factuality_Score_1"
)

COVERAGE_FOLDER = Path(
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation\DPO\Coverage_Score"
)

PARTS = ["01", "02", "03", "04", "05", "06"]


# ============================================================
# LOAD JSONL
# ============================================================

def load_jsonl(file_path):

    records = []

    with open(file_path, "r", encoding="utf-8") as f:
        for line in f:
            if line.strip():
                records.append(json.loads(line))

    return records


# ============================================================
# CREATE INTERSECTION COVERAGE FILES
# ============================================================

total_cases = 0

print("=" * 70)
print("CREATING INTERSECTION COVERAGE FILES")
print("=" * 70)


for part in PARTS:

    # --------------------------------------------------------
    # File paths
    # --------------------------------------------------------

    factuality_file = (
        FACTUALITY_FOLDER
        / f"factuality_scores_patient_report_wgt_dpo_{part}.jsonl"
    )

    coverage_file = (
        COVERAGE_FOLDER
        / f"coverage_scores_wgt_dpo_{part}.jsonl"
    )

    output_file = (
        COVERAGE_FOLDER
        / f"coverage_scores_wgt_dpo_intersection_{part}.jsonl"
    )


    # --------------------------------------------------------
    # Load files
    # --------------------------------------------------------

    factuality_records = load_jsonl(factuality_file)
    coverage_records = load_jsonl(coverage_file)


    # --------------------------------------------------------
    # Get factuality indices
    # --------------------------------------------------------

    factuality_ids = {
        record["index"]
        for record in factuality_records
    }


    # --------------------------------------------------------
    # Filter coverage to intersection
    # --------------------------------------------------------

    coverage_intersection = [
        record
        for record in coverage_records
        if record["index"] in factuality_ids
    ]


    # --------------------------------------------------------
    # Write NEW JSONL file
    # --------------------------------------------------------

    with open(output_file, "w", encoding="utf-8") as f:

        for record in coverage_intersection:
            f.write(
                json.dumps(
                    record,
                    ensure_ascii=False
                ) + "\n"
            )


    # --------------------------------------------------------
    # Print results
    # --------------------------------------------------------

    print(f"\nPART {part}")
    print("-" * 50)
    print("Original coverage cases :", len(coverage_records))
    print("Intersection cases      :", len(coverage_intersection))
    print("Output file             :", output_file.name)

    total_cases += len(coverage_intersection)


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("INTERSECTION FILES CREATED")
print("=" * 70)

print("Total intersection cases:", total_cases)
```

```python
import json
import os
import pandas as pd


# ============================================================
# DIRECTORIES
# ============================================================

FACTUALITY_DIR = (
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation"
    r"\DPO\Factuality_Score\Factuality_Score_1"
)

COVERAGE_DIR = (
    r"C:\Users\Aman Kumar Singh\Desktop\nlp_dissertation"
    r"\DPO\Coverage_Score"
)


# ============================================================
# STORAGE
# ============================================================

factuality_records = []
coverage_records = []


# ============================================================
# 1. READ DPO FACTUALITY FILES
# ============================================================

for part in range(1, 7):

    part_str = f"{part:02d}"

    filename = os.path.join(
        FACTUALITY_DIR,
        f"factuality_scores_patient_report_wgt_dpo_{part_str}.jsonl"
    )

    with open(filename, "r", encoding="utf-8") as f:

        for line in f:

            if not line.strip():
                continue

            data = json.loads(line)

            factuality_scores = data.get(
                "factuality_scores", {}
            )

            factuality_records.append({
                "part": part_str,
                "index": data.get("index"),
                "Report": data.get("report"),
                "Factuality Score": factuality_scores.get(
                    "average_factuality_score"
                )
            })


# ============================================================
# 2. READ DPO COVERAGE INTERSECTION FILES
# ============================================================

for part in range(1, 7):

    part_str = f"{part:02d}"

    filename = os.path.join(
        COVERAGE_DIR,
        f"coverage_scores_wgt_dpo_intersection_{part_str}.jsonl"
    )

    with open(filename, "r", encoding="utf-8") as f:

        for line in f:

            if not line.strip():
                continue

            data = json.loads(line)

            coverage_scores = data.get(
                "coverage_scores", {}
            )

            coverage_records.append({
                "part": part_str,
                "index": data.get("index"),
                "coverage": coverage_scores.get(
                    "patient_coverage_score"
                )
            })


# ============================================================
# 3. CONVERT TO DATAFRAMES
# ============================================================

factuality_df = pd.DataFrame(
    factuality_records
)

coverage_df = pd.DataFrame(
    coverage_records
)


# ============================================================
# 4. MERGE ON PART + INDEX
# ============================================================

dpo_dataset = factuality_df.merge(
    coverage_df,
    on=["part", "index"],
    how="inner"
)


# ============================================================
# 5. REMOVE DUPLICATES
# ============================================================

dpo_dataset = dpo_dataset.drop_duplicates(
    subset=["part", "index"]
).reset_index(drop=True)


# ============================================================
# 6. DISPLAY RESULTS
# ============================================================

print("=" * 70)
print("DPO DATAFRAME")
print("=" * 70)

print("\nFirst 10 rows:")

print(
    dpo_dataset.head(10).to_string()
)


print("\nShape:")

print(
    dpo_dataset.shape
)


print("\nMissing values:")

print(
    dpo_dataset.isna().sum()
)


print("\nRecords per part:")

print(
    dpo_dataset["part"]
    .value_counts()
    .sort_index()
)


print("\nDescriptive statistics:")

print(
    dpo_dataset[
        [
            "Factuality Score",
            "coverage"
        ]
    ].describe()
)
```

```python
import re
import pandas as pd


# ============================================================
# CLEAN REPORT FUNCTION
# ============================================================

def clean_report(text):

    if pd.isna(text):
        return text

    # Remove markdown headers
    text = re.sub(
        r'^#+\s*',
        '',
        text,
        flags=re.MULTILINE
    )

    # Remove bullet points
    text = re.sub(
        r'^\s*[-*]\s+',
        '',
        text,
        flags=re.MULTILINE
    )

    # Remove bold markdown
    text = re.sub(
        r'\*\*',
        '',
        text
    )

    # Remove numbered section markers: 1), 2), 10), etc.
    text = re.sub(
        r'\b\d+\)\s*',
        '',
        text
    )

    # Remove lettered subsection markers: a), b), etc.
    text = re.sub(
        r'\b[a-zA-Z]\)\s*',
        '',
        text
    )

    # Replace newlines with spaces
    text = re.sub(
        r'\s*\n\s*',
        ' ',
        text
    )

    # Remove multiple spaces
    text = re.sub(
        r'\s+',
        ' ',
        text
    )

    return text.strip()


# ============================================================
# CREATE CLEANED REPORTS FOR DPO
# ============================================================

dpo_dataset["Cleaned_Report"] = (
    dpo_dataset["Report"].apply(clean_report)
)


# ============================================================
# CHECK
# ============================================================

print(
    dpo_dataset[
        ["Report", "Cleaned_Report"]
    ].head(2).to_string()
)

print("\nShape:")
print(dpo_dataset.shape)

print("\nMissing values:")
print(
    dpo_dataset["Cleaned_Report"].isna().sum()
)
```

```python
from transformers import AutoTokenizer
import pandas as pd


MODEL_ID = "Qwen/Qwen3-30B-A3B-Instruct-2507"


print("Loading tokenizer...", flush=True)

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_ID
)

print("Tokenizer loaded successfully\n", flush=True)


# ============================================================
# TOKEN LENGTH FUNCTION
# ============================================================

def get_token_length(text):

    if pd.isna(text):
        return 0

    tokens = tokenizer.encode(
        str(text),
        add_special_tokens=True
    )

    return len(tokens)


# ============================================================
# CALCULATE DPO TOKEN LENGTHS
# ============================================================

print(
    "Calculating DPO token lengths...",
    flush=True
)

dpo_dataset["Cleaned_token_length"] = (
    dpo_dataset["Cleaned_Report"].apply(
        get_token_length
    )
)


print(
    "DPO token length calculation "
    "completed successfully.\n"
)


# ============================================================
# CHECK
# ============================================================

print(
    dpo_dataset[
        [
            "part",
            "index",
            "Cleaned_token_length"
        ]
    ].head(10)
)


print("\nShape:")
print(dpo_dataset.shape)


print("\nDPO token length statistics:")

print(
    dpo_dataset[
        "Cleaned_token_length"
    ].describe()
)
```

```python
from sklearn.preprocessing import MinMaxScaler


# ============================================================
# 1. MIN-MAX NORMALIZE TOKEN LENGTH
# ============================================================

L_norm = MinMaxScaler().fit_transform(
    dpo_dataset[["Cleaned_token_length"]]
).flatten()


# ============================================================
# 2. BREVITY FACTOR
# ============================================================

brevity_factor = 1 - L_norm

dpo_dataset["Brevity Factor"] = brevity_factor


# ============================================================
# 3. CONCISENESS SCORE
# ============================================================

dpo_dataset["Conciseness Score"] = (
    dpo_dataset["coverage"]
    * dpo_dataset["Factuality Score"]
    * dpo_dataset["Brevity Factor"]
) ** (1 / 3)


# ============================================================
# 4. CHECK RESULTS
# ============================================================

print("\nDPO conciseness calculation completed.\n")

print(
    dpo_dataset[
        [
            "part",
            "index",
            "coverage",
            "Factuality Score",
            "Cleaned_token_length",
            "Brevity Factor",
            "Conciseness Score"
        ]
    ].head(10)
)


# ============================================================
# 5. CONCISENESS STATISTICS
# ============================================================

print("\nDPO Conciseness statistics:")

print(
    dpo_dataset[
        "Conciseness Score"
    ].describe()
)
```

```python
import pandas as pd
import textstat


# ============================================================
# SAFE TEXT FUNCTION
# ============================================================

def safe_text(x):

    if pd.isna(x):
        return ""

    return str(x)


# ============================================================
# CALCULATE READABILITY METRICS
# ============================================================

dpo_dataset["FRE"] = dpo_dataset["Cleaned_Report"].apply(
    lambda x: textstat.flesch_reading_ease(
        safe_text(x)
    )
)

dpo_dataset["FKGL"] = dpo_dataset["Cleaned_Report"].apply(
    lambda x: textstat.flesch_kincaid_grade(
        safe_text(x)
    )
)

dpo_dataset["SMOG"] = dpo_dataset["Cleaned_Report"].apply(
    lambda x: textstat.smog_index(
        safe_text(x)
    )
)

dpo_dataset["Dale_Chall"] = dpo_dataset["Cleaned_Report"].apply(
    lambda x: textstat.dale_chall_readability_score(
        safe_text(x)
    )
)


# ============================================================
# CHECK RESULTS
# ============================================================

print("\nDPO readability metrics:")

print(
    dpo_dataset[
        [
            "part",
            "index",
            "FRE",
            "FKGL",
            "SMOG",
            "Dale_Chall"
        ]
    ].head()
)


print("\nDataset shape:")
print(dpo_dataset.shape)


print("\nDPO Readability statistics:")

print(
    dpo_dataset[
        [
            "FRE",
            "FKGL",
            "SMOG",
            "Dale_Chall"
        ]
    ].describe()
)
```

```python
import numpy as np
import pandas as pd


# ============================================================
# CONFIGURATION
# ============================================================

N_SIMULATIONS = 5000
RANDOM_SEED = 42

np.random.seed(RANDOM_SEED)


# ============================================================
# 1. CHECK AVAILABLE COLUMNS
# ============================================================

print("=" * 70)
print("AVAILABLE COLUMNS")
print("=" * 70)

print(wght_dataframe.columns.tolist())


# ============================================================
# 2. PREPARE WEIGHTED SFT DATA
# ============================================================

weighted = wght_dataframe[
    [
        "part",
        "index",
        "coverage"
    ]
].copy()

weighted = weighted.rename(
    columns={
        "coverage": "Weighted_SFT_Coverage"
    }
)


# ============================================================
# 3. PREPARE DPO DATA
# ============================================================

# IMPORTANT:
# Replace dpo_dataset with the dataframe containing
# the 68 real DPO cases if your dataframe has another name.

print("\nDPO dataframe columns:")
print(dpo_dataset.columns.tolist())


# ------------------------------------------------------------
# Find the actual DPO coverage column
# ------------------------------------------------------------

possible_coverage_columns = [
    "coverage",
    "coverage_score",
    "patient_coverage_score",
    "Coverage Score",
    "Coverage"
]

dpo_coverage_column = None

for col in possible_coverage_columns:
    if col in dpo_dataset.columns:
        dpo_coverage_column = col
        break


if dpo_coverage_column is None:

    raise KeyError(
        "Could not find the DPO coverage column.\n"
        "Available columns are:\n"
        + str(dpo_dataset.columns.tolist())
    )


print(
    "\nUsing DPO coverage column:",
    dpo_coverage_column
)


dpo = dpo_dataset[
    [
        "part",
        "index",
        dpo_coverage_column
    ]
].copy()

dpo = dpo.rename(
    columns={
        dpo_coverage_column: "DPO_Coverage"
    }
)


# ============================================================
# 4. MATCH WEIGHTED SFT ↔ DPO
# ============================================================

matched = pd.merge(
    weighted,
    dpo,
    on=["part", "index"],
    how="inner"
)


# ============================================================
# 5. CALCULATE OBSERVED DIFFERENCE
# ============================================================

matched["Observed_Difference"] = (
    matched["DPO_Coverage"]
    - matched["Weighted_SFT_Coverage"]
)


print("\n" + "=" * 70)
print("REAL WEIGHTED SFT ↔ DPO COVERAGE MATCH")
print("=" * 70)

print(
    "Weighted SFT cases :",
    len(weighted)
)

print(
    "Real DPO cases     :",
    len(dpo)
)

print(
    "Matched cases      :",
    len(matched)
)

print("\nObserved difference statistics:")

print(
    matched["Observed_Difference"].describe()
)


# ============================================================
# 6. IDENTIFY MISSING DPO CASES
# ============================================================

matched_ids = set(
    zip(
        matched["part"],
        matched["index"]
    )
)

weighted["has_DPO"] = weighted.apply(
    lambda row:
    (row["part"], row["index"]) in matched_ids,
    axis=1
)

missing_dpo = weighted[
    ~weighted["has_DPO"]
].copy()

print(
    "\nCases without real DPO results:",
    len(missing_dpo)
)


# ============================================================
# 7. CREATE COVERAGE STRATA
# ============================================================

# IMPORTANT:
# Use a COPY because numpy quantile values can be read-only.

bins = np.array(
    matched[
        "Weighted_SFT_Coverage"
    ].quantile(
        [0, 1/3, 2/3, 1]
    ).values,
    dtype=float
)

bins[0] = -np.inf
bins[-1] = np.inf


matched["Coverage_Stratum"] = pd.cut(
    matched["Weighted_SFT_Coverage"],
    bins=bins,
    labels=[
        "Low",
        "Medium",
        "High"
    ],
    include_lowest=True
)


missing_dpo["Coverage_Stratum"] = pd.cut(
    missing_dpo["Weighted_SFT_Coverage"],
    bins=bins,
    labels=[
        "Low",
        "Medium",
        "High"
    ],
    include_lowest=True
)


print("\n" + "=" * 70)
print("OBSERVED COVERAGE STRATA")
print("=" * 70)

print(
    matched[
        "Coverage_Stratum"
    ].value_counts()
    .sort_index()
)

print("\nMissing cases by stratum:")

print(
    missing_dpo[
        "Coverage_Stratum"
    ].value_counts()
    .sort_index()
)


# ============================================================
# 8. ESTIMATE DPO COVERAGE FOR MISSING CASES
# ============================================================

estimated_differences = []


for stratum in [
    "Low",
    "Medium",
    "High"
]:

    observed_stratum = matched[
        matched["Coverage_Stratum"] == stratum
    ]["Observed_Difference"].dropna().values

    missing_indices = missing_dpo.index[
        missing_dpo["Coverage_Stratum"] == stratum
    ]

    if len(missing_indices) == 0:
        continue

    sampled_differences = np.random.choice(
        observed_stratum,
        size=len(missing_indices),
        replace=True
    )

    estimated_differences.extend(
        zip(
            missing_indices,
            sampled_differences
        )
    )


# ============================================================
# 9. CREATE AUGMENTED 108-CASE DATASET
# ============================================================

augmented = weighted[
    [
        "part",
        "index",
        "Weighted_SFT_Coverage"
    ]
].copy()

augmented["DPO_Type"] = "Observed"

augmented["DPO_Coverage"] = np.nan

augmented["Difference"] = np.nan


# Insert real DPO values

for _, row in matched.iterrows():

    mask = (
        (augmented["part"] == row["part"])
        &
        (augmented["index"] == row["index"])
    )

    augmented.loc[
        mask,
        "DPO_Coverage"
    ] = row["DPO_Coverage"]

    augmented.loc[
        mask,
        "Difference"
    ] = row["Observed_Difference"]


# Insert estimated values

for idx, difference in estimated_differences:

    weighted_coverage = augmented.loc[
        idx,
        "Weighted_SFT_Coverage"
    ]

    estimated_dpo = (
        weighted_coverage
        + difference
    )

    # Keep coverage within [0,1]

    estimated_dpo = np.clip(
        estimated_dpo,
        0,
        1
    )

    augmented.loc[
        idx,
        "DPO_Coverage"
    ] = estimated_dpo

    augmented.loc[
        idx,
        "Difference"
    ] = (
        estimated_dpo
        - weighted_coverage
    )

    augmented.loc[
        idx,
        "DPO_Type"
    ] = "Estimated"


# ============================================================
# 10. CHECK AUGMENTED DATASET
# ============================================================

print("\n" + "=" * 70)
print("EXPLORATORY AUGMENTED DPO COVERAGE DATASET")
print("=" * 70)

print("\nShape:")
print(
    augmented.shape
)

print("\nObserved vs Estimated:")

print(
    augmented["DPO_Type"].value_counts()
)

print("\nCoverage statistics:")

print(
    augmented[
        [
            "Weighted_SFT_Coverage",
            "DPO_Coverage",
            "Difference"
        ]
    ].describe()
)


print("\nFirst 10 cases:")

print(
    augmented.head(10).to_string(
        index=False
    )
)


# ============================================================
# 11. 5000-IMPUTATION SENSITIVITY ANALYSIS
# ============================================================

simulation_results = []


for simulation in range(
    N_SIMULATIONS
):

    simulated_differences = (
        matched[
            "Observed_Difference"
        ].tolist()
    )

    # Generate missing values
    # separately within each stratum

    for stratum in [
        "Low",
        "Medium",
        "High"
    ]:

        observed = matched[
            matched["Coverage_Stratum"] == stratum
        ]["Observed_Difference"].dropna().values

        n_missing = (
            missing_dpo[
                "Coverage_Stratum"
            ] == stratum
        ).sum()

        if n_missing > 0:

            sampled = np.random.choice(
                observed,
                size=n_missing,
                replace=True
            )

            simulated_differences.extend(
                sampled.tolist()
            )


    simulated_differences = np.array(
        simulated_differences
    )

    simulated_dpo = (
        weighted[
            "Weighted_SFT_Coverage"
        ].values
        + simulated_differences
    )

    simulated_dpo = np.clip(
        simulated_dpo,
        0,
        1
    )


    simulation_results.append({

        "DPO_mean":
            np.mean(simulated_dpo),

        "Weighted_SFT_mean":
            np.mean(
                weighted[
                    "Weighted_SFT_Coverage"
                ]
            ),

        "Mean_difference":
            np.mean(
                simulated_dpo
                - weighted[
                    "Weighted_SFT_Coverage"
                ].values
            ),

        "DPO_median":
            np.median(simulated_dpo),

        "DPO_SD":
            np.std(
                simulated_dpo,
                ddof=1
            ),

        "Difference_SD":
            np.std(
                simulated_dpo
                - weighted[
                    "Weighted_SFT_Coverage"
                ].values,
                ddof=1
            )
    })


simulation_df = pd.DataFrame(
    simulation_results
)


# ============================================================
# 12. SENSITIVITY SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("5000-IMPUTATION COVERAGE SENSITIVITY ANALYSIS")
print("=" * 70)

print(
    simulation_df.describe(
        percentiles=[
            0.025,
            0.50,
            0.975
        ]
    )
)


# ============================================================
# 13. AUGMENTED MEAN DIFFERENCE
# ============================================================

mean_difference = (
    simulation_df[
        "Mean_difference"
    ].mean()
)

lower_ci = (
    simulation_df[
        "Mean_difference"
    ].quantile(0.025)
)

upper_ci = (
    simulation_df[
        "Mean_difference"
    ].quantile(0.975)
)


print("\n" + "=" * 70)
print("AUGMENTED COVERAGE MEAN DIFFERENCE")
print("=" * 70)

print(
    f"Mean estimated difference : "
    f"{mean_difference:.4f}"
)

print(
    f"95% simulation interval   : "
    f"[{lower_ci:.4f}, {upper_ci:.4f}]"
)


# ============================================================
# 14. SAVE
# ============================================================

augmented.to_csv(
    "dpo_coverage_augmented_108_exploratory.csv",
    index=False
)

simulation_df.to_csv(
    "dpo_coverage_5000_imputation_sensitivity.csv",
    index=False
)


print("\nSaved:")
print(
    "dpo_coverage_augmented_108_exploratory.csv"
)

print(
    "dpo_coverage_5000_imputation_sensitivity.csv"
)
```

```python
import numpy as np
import pandas as pd


# ============================================================
# CONFIGURATION
# ============================================================

RANDOM_SEED = 42

rng = np.random.default_rng(RANDOM_SEED)


# ============================================================
# 1. PREPARE REAL 68 DPO CASES
# ============================================================

real = matched[
    [
        "part",
        "index",
        "Weighted_SFT_Token_Length",
        "DPO_Token_Length",
        "Observed_Difference"
    ]
].copy()

real["DPO_Type"] = "Observed"


# ============================================================
# 2. PREPARE ALL 108 WEIGHTED SFT CASES
# ============================================================

all_cases = wght_dataframe[
    [
        "part",
        "index",
        "Cleaned_token_length"
    ]
].copy()

all_cases = all_cases.rename(
    columns={
        "Cleaned_token_length":
        "Weighted_SFT_Token_Length"
    }
)


# ============================================================
# 3. IDENTIFY THE 40 CASES WITHOUT REAL DPO
# ============================================================

real_keys = set(
    zip(
        real["part"],
        real["index"]
    )
)

missing = all_cases[
    ~all_cases.apply(
        lambda row:
        (row["part"], row["index"]) in real_keys,
        axis=1
    )
].copy()

print("=" * 70)
print("DPO TOKEN-LENGTH AUGMENTATION")
print("=" * 70)

print("Total Weighted SFT cases :", len(all_cases))
print("Real DPO cases           :", len(real))
print("Missing DPO cases       :", len(missing))


# ============================================================
# 4. CREATE LENGTH STRATA
# ============================================================
#
# We stratify the 68 real cases according to the length of
# their Weighted SFT reports.
#
# This prevents the bootstrap from ignoring the relationship
# between report length and DPO length.
# ============================================================

real["Length_Stratum"] = pd.qcut(
    real["Weighted_SFT_Token_Length"],
    q=3,
    labels=["Low", "Medium", "High"],
    duplicates="drop"
)

missing["Length_Stratum"] = pd.qcut(
    all_cases["Weighted_SFT_Token_Length"],
    q=3,
    labels=["Low", "Medium", "High"],
    duplicates="drop"
)


# ============================================================
# 5. SAMPLE OBSERVED DIFFERENCES WITHIN EACH STRATUM
# ============================================================

estimated_records = []

for stratum in ["Low", "Medium", "High"]:

    observed_stratum = real[
        real["Length_Stratum"] == stratum
    ]

    missing_stratum = missing[
        missing["Length_Stratum"] == stratum
    ]

    if len(missing_stratum) == 0:
        continue

    differences = (
        observed_stratum[
            "Observed_Difference"
        ]
        .dropna()
        .to_numpy()
    )

    # Bootstrap from the ACTUAL observed differences
    sampled_differences = rng.choice(
        differences,
        size=len(missing_stratum),
        replace=True
    )

    for (_, row), difference in zip(
        missing_stratum.iterrows(),
        sampled_differences
    ):

        estimated_dpo_length = (
            row["Weighted_SFT_Token_Length"]
            + difference
        )

        # Prevent impossible token lengths
        estimated_dpo_length = max(
            1,
            round(estimated_dpo_length)
        )

        estimated_records.append({
            "part": row["part"],
            "index": row["index"],
            "Weighted_SFT_Token_Length":
                row["Weighted_SFT_Token_Length"],
            "DPO_Token_Length":
                estimated_dpo_length,
            "Difference":
                estimated_dpo_length
                - row["Weighted_SFT_Token_Length"],
            "DPO_Type": "Estimated"
        })


# ============================================================
# 6. CONVERT ESTIMATED CASES TO DATAFRAME
# ============================================================

estimated = pd.DataFrame(
    estimated_records
)


# ============================================================
# 7. STANDARDISE REAL CASES
# ============================================================

observed = real[
    [
        "part",
        "index",
        "Weighted_SFT_Token_Length",
        "DPO_Token_Length"
    ]
].copy()

observed["Difference"] = (
    observed["DPO_Token_Length"]
    - observed["Weighted_SFT_Token_Length"]
)

observed["DPO_Type"] = "Observed"


# ============================================================
# 8. COMBINE 68 OBSERVED + 40 ESTIMATED
# ============================================================

dpo_token_108 = pd.concat(
    [
        observed,
        estimated
    ],
    ignore_index=True
)


# ============================================================
# 9. REMOVE DUPLICATES
# ============================================================

dpo_token_108 = (
    dpo_token_108
    .drop_duplicates(
        subset=["part", "index"]
    )
    .sort_values(
        ["part", "index"]
    )
    .reset_index(drop=True)
)


# ============================================================
# 10. CHECK DATASET
# ============================================================

print("\n" + "=" * 70)
print("EXPLORATORY 108-CASE DPO TOKEN-LENGTH DATASET")
print("=" * 70)

print("\nShape:")
print(dpo_token_108.shape)

print("\nObserved vs Estimated:")
print(
    dpo_token_108["DPO_Type"]
    .value_counts()
)


# ============================================================
# 11. SUMMARY STATISTICS
# ============================================================

print("\n" + "=" * 70)
print("TOKEN-LENGTH STATISTICS")
print("=" * 70)

print(
    dpo_token_108[
        [
            "Weighted_SFT_Token_Length",
            "DPO_Token_Length",
            "Difference"
        ]
    ].describe()
)


# ============================================================
# 12. OBSERVED VS ESTIMATED STATISTICS
# ============================================================

print("\n" + "=" * 70)
print("OBSERVED DPO DIFFERENCE")
print("=" * 70)

print(
    dpo_token_108[
        dpo_token_108["DPO_Type"] == "Observed"
    ]["Difference"].describe()
)

print("\nEstimated DPO difference:")

print(
    dpo_token_108[
        dpo_token_108["DPO_Type"] == "Estimated"
    ]["Difference"].describe()
)


# ============================================================
# 13. CHECK FIRST 10 CASES
# ============================================================

print("\n" + "=" * 70)
print("FIRST 10 CASES")
print("=" * 70)

print(
    dpo_token_108.head(10).to_string(
        index=False
    )
)


# ============================================================
# 14. SAVE
# ============================================================

OUTPUT_FILE = (
    "dpo_token_length_augmented_108_"
    "stratified_bootstrap.csv"
)

dpo_token_108.to_csv(
    OUTPUT_FILE,
    index=False
)

print("\nSaved:")
print(OUTPUT_FILE)
```

```python
import numpy as np
import pandas as pd

from scipy.stats import mannwhitneyu, ks_2samp


# ============================================================
# 1. SPLIT OBSERVED AND ESTIMATED CASES
# ============================================================

observed = dpo_108[
    dpo_108["DPO_Type"] == "Observed"
].copy()

estimated = dpo_108[
    dpo_108["DPO_Type"] == "Estimated"
].copy()


# ============================================================
# 2. METRICS
# ============================================================

metrics = [
    "DPO_Factuality",
    "DPO_Coverage",
    "DPO_Token_Length",
    "Brevity Factor",
    "Conciseness Score"
]


# ============================================================
# 3. COHEN'S D FUNCTION
# ============================================================

def cohens_d(x, y):

    x = np.asarray(x.dropna())
    y = np.asarray(y.dropna())

    nx = len(x)
    ny = len(y)

    pooled_sd = np.sqrt(
        (
            (nx - 1) * np.var(x, ddof=1)
            + (ny - 1) * np.var(y, ddof=1)
        )
        /
        (nx + ny - 2)
    )

    return (
        (np.mean(y) - np.mean(x))
        / pooled_sd
    )


# ============================================================
# 4. COMPARE OBSERVED vs ESTIMATED
# ============================================================

results = []

for metric in metrics:

    obs = observed[metric].dropna()
    est = estimated[metric].dropna()

    # Mann-Whitney U
    mw_stat, mw_p = mannwhitneyu(
        obs,
        est,
        alternative="two-sided"
    )

    # Kolmogorov-Smirnov
    ks_stat, ks_p = ks_2samp(
        obs,
        est
    )

    # Cohen's d
    d = cohens_d(
        obs,
        est
    )

    results.append({

        "Metric": metric,

        "Observed_Mean": obs.mean(),
        "Estimated_Mean": est.mean(),

        "Mean_Difference":
            est.mean() - obs.mean(),

        "Observed_SD": obs.std(),
        "Estimated_SD": est.std(),

        "Cohens_d": d,

        "MannWhitney_p": mw_p,

        "KS_Statistic": ks_stat,
        "KS_p": ks_p

    })


# ============================================================
# 5. RESULTS DATAFRAME
# ============================================================

similarity_results = pd.DataFrame(results)


# ============================================================
# 6. DISPLAY
# ============================================================

pd.set_option(
    "display.float_format",
    "{:.6f}".format
)

print("=" * 100)
print("DPO AUGMENTATION VALIDATION")
print("OBSERVED (68) vs ESTIMATED (40)")
print("=" * 100)

print(
    similarity_results.to_string(
        index=False
    )
)


# ============================================================
# 7. INTERPRETATION FLAGS
# ============================================================

similarity_results["Small_Effect"] = (
    similarity_results["Cohens_d"].abs()
    < 0.2
)

similarity_results["MW_Not_Significant"] = (
    similarity_results["MannWhitney_p"]
    > 0.05
)

similarity_results["KS_Not_Significant"] = (
    similarity_results["KS_p"]
    > 0.05
)


print("\n" + "=" * 100)
print("SIMILARITY FLAGS")
print("=" * 100)

print(
    similarity_results[
        [
            "Metric",
            "Cohens_d",
            "Small_Effect",
            "MannWhitney_p",
            "MW_Not_Significant",
            "KS_Statistic",
            "KS_p",
            "KS_Not_Significant"
        ]
    ].to_string(
        index=False
    )
)
```

```python
import numpy as np
import pandas as pd

from scipy.stats import ttest_rel, wilcoxon


# ============================================================
# CREATE MATCHED 108-CASE DATAFRAME
# ============================================================

comparison_108 = wght_dataframe[
    [
        "part",
        "index",
        "Factuality Score",
        "coverage",
        "Conciseness Score"
    ]
].merge(
    dpo_108[
        [
            "part",
            "index",
            "DPO_Factuality",
            "DPO_Coverage",
            "Conciseness Score"
        ]
    ].rename(
        columns={
            "Conciseness Score": "DPO_Conciseness"
        }
    ),
    on=[
        "part",
        "index"
    ],
    how="inner"
)


# Rename Weighted SFT columns
comparison_108 = comparison_108.rename(
    columns={
        "Factuality Score": "Weighted_SFT_Factuality",
        "coverage": "Weighted_SFT_Coverage",
        "Conciseness Score": "Weighted_SFT_Conciseness"
    }
)


# ============================================================
# METRICS
# ============================================================

metrics = {
    "Factuality": (
        "Weighted_SFT_Factuality",
        "DPO_Factuality"
    ),
    
    "Coverage": (
        "Weighted_SFT_Coverage",
        "DPO_Coverage"
    ),
    
    "Conciseness": (
        "Weighted_SFT_Conciseness",
        "DPO_Conciseness"
    )
}


# ============================================================
# PAIRED EFFECT SIZE
# ============================================================

def paired_cohens_d(diff):
    
    return (
        diff.mean()
        / diff.std(ddof=1)
    )


# ============================================================
# STATISTICAL COMPARISON
# ============================================================

results = []

for metric, columns in metrics.items():
    
    weighted_col, dpo_col = columns
    
    weighted = comparison_108[
        weighted_col
    ].dropna()
    
    dpo = comparison_108[
        dpo_col
    ].dropna()
    
    
    # Ensure pairing
    paired = comparison_108[
        [
            weighted_col,
            dpo_col
        ]
    ].dropna()
    
    weighted = paired[weighted_col]
    dpo = paired[dpo_col]
    
    
    # Difference: DPO - Weighted SFT
    difference = (
        dpo - weighted
    )
    
    
    # Paired t-test
    t_stat, t_p = ttest_rel(
        dpo,
        weighted
    )
    
    
    # Wilcoxon signed-rank test
    try:
        
        w_stat, w_p = wilcoxon(
            dpo,
            weighted
        )
        
    except ValueError:
        
        w_stat = np.nan
        w_p = np.nan
    
    
    # 95% Confidence Interval
    mean_diff = difference.mean()
    
    se = (
        difference.std(ddof=1)
        / np.sqrt(len(difference))
    )
    
    ci_lower = (
        mean_diff
        - 1.96 * se
    )
    
    ci_upper = (
        mean_diff
        + 1.96 * se
    )
    
    
    results.append({
        
        "Metric": metric,
        
        "N": len(paired),
        
        "Weighted SFT Mean": weighted.mean(),
        "DPO Mean": dpo.mean(),
        
        "Mean Difference (DPO-W)": mean_diff,
        
        "Weighted SFT SD": weighted.std(),
        "DPO SD": dpo.std(),
        
        "Paired Cohen's d": paired_cohens_d(
            difference
        ),
        
        "Paired t-test p": t_p,
        
        "Wilcoxon p": w_p,
        
        "95% CI Lower": ci_lower,
        "95% CI Upper": ci_upper
        
    })


# ============================================================
# RESULTS DATAFRAME
# ============================================================

stats_108 = pd.DataFrame(results)


# ============================================================
# DISPLAY
# ============================================================

pd.set_option(
    "display.float_format",
    "{:.6f}".format
)


print("=" * 100)
print("EXPLORATORY 108-CASE COMPARISON")
print("WEIGHTED SFT vs AUGMENTED DPO")
print("=" * 100)

print(
    stats_108.to_string(
        index=False
    )
)
```

```python
import numpy as np
import pandas as pd
from scipy.stats import wilcoxon
from statsmodels.stats.multitest import multipletests


# ============================================================
# 1. Prepare Weighted SFT — 108 cases
# ============================================================

weighted = wght_dataframe[
    [
        "part",
        "index",
        "Factuality Score",
        "coverage",
        "Conciseness Score"
    ]
].copy()


# ============================================================
# 2. Prepare DPO — ALL 108 cases
#    68 Observed + 40 Estimated
# ============================================================

dpo_full = dpo_108[
    [
        "part",
        "index",
        "DPO_Factuality",
        "DPO_Coverage",
        "Conciseness Score"
    ]
].copy()


# ============================================================
# 3. Ensure matching identifiers have same format
# ============================================================

weighted["part"] = weighted["part"].astype(str)
dpo_full["part"] = dpo_full["part"].astype(str)

weighted["index"] = weighted["index"].astype(int)
dpo_full["index"] = dpo_full["index"].astype(int)


# ============================================================
# 4. Rename columns BEFORE merging
#    This avoids the Conciseness Score KeyError
# ============================================================

weighted = weighted.rename(
    columns={
        "Factuality Score": "Weighted_Factuality",
        "coverage": "Weighted_Coverage",
        "Conciseness Score": "Weighted_Conciseness"
    }
)

dpo_full = dpo_full.rename(
    columns={
        "DPO_Factuality": "DPO_Factuality",
        "DPO_Coverage": "DPO_Coverage",
        "Conciseness Score": "DPO_Conciseness"
    }
)


# ============================================================
# 5. Match all 108 paired cases
# ============================================================

comparison_108 = weighted.merge(
    dpo_full,
    on=["part", "index"],
    how="inner"
)


print("=" * 90)
print("PAIRED WEIGHTED SFT vs AUGMENTED DPO COMPARISON")
print("=" * 90)

print(f"\nTotal matched cases: {len(comparison_108)}")

print("\nDPO composition:")
print(dpo_108["DPO_Type"].value_counts())


# ============================================================
# 6. Rank-biserial correlation for paired data
# ============================================================

def rank_biserial_paired(weighted_scores, dpo_scores):

    weighted_scores = np.asarray(weighted_scores)
    dpo_scores = np.asarray(dpo_scores)

    # Difference = DPO - Weighted SFT
    differences = dpo_scores - weighted_scores

    # Remove zero differences
    nonzero = differences != 0
    differences = differences[nonzero]

    if len(differences) == 0:
        return 0.0

    # Rank absolute differences
    ranks = pd.Series(
        np.abs(differences)
    ).rank(
        method="average"
    ).to_numpy()

    positive_ranks = ranks[differences > 0].sum()
    negative_ranks = ranks[differences < 0].sum()

    r_rb = (
        positive_ranks - negative_ranks
    ) / (
        positive_ranks + negative_ranks
    )

    return r_rb


# ============================================================
# 7. Define the three paired comparisons
# ============================================================

metrics = {

    "Factuality": (
        "Weighted_Factuality",
        "DPO_Factuality"
    ),

    "Coverage": (
        "Weighted_Coverage",
        "DPO_Coverage"
    ),

    "Conciseness": (
        "Weighted_Conciseness",
        "DPO_Conciseness"
    )
}


# ============================================================
# 8. Run Wilcoxon signed-rank tests
# ============================================================

results = []


for dimension, (weighted_col, dpo_col) in metrics.items():

    # --------------------------------------------------------
    # Keep only cases with both scores
    # --------------------------------------------------------

    temp = comparison_108[
        [weighted_col, dpo_col]
    ].dropna()

    weighted_scores = temp[weighted_col].to_numpy()
    dpo_scores = temp[dpo_col].to_numpy()

    # --------------------------------------------------------
    # Difference = DPO - Weighted SFT
    # --------------------------------------------------------

    differences = dpo_scores - weighted_scores

    # --------------------------------------------------------
    # Wilcoxon signed-rank test
    # --------------------------------------------------------

    wilcoxon_result = wilcoxon(
        dpo_scores,
        weighted_scores,
        alternative="two-sided",
        zero_method="wilcox",
        method="auto"
    )

    statistic = wilcoxon_result.statistic
    p_value = wilcoxon_result.pvalue

    # --------------------------------------------------------
    # Rank-biserial effect size
    # --------------------------------------------------------

    rank_biserial = rank_biserial_paired(
        weighted_scores,
        dpo_scores
    )

    # --------------------------------------------------------
    # Store results
    # --------------------------------------------------------

    results.append({

        "Dimension": dimension,

        "N": len(temp),

        "Weighted SFT Median": np.median(
            weighted_scores
        ),

        "DPO Median": np.median(
            dpo_scores
        ),

        "Median Δ (DPO - Weighted SFT)": np.median(
            differences
        ),

        "Wilcoxon W": statistic,

        "Raw p-value": p_value,

        "Rank-biserial r": rank_biserial
    })


# ============================================================
# 9. Create results dataframe
# ============================================================

results_108_df = pd.DataFrame(results)


# ============================================================
# 10. Holm multiple-comparison correction
# ============================================================

reject, p_adjusted, _, _ = multipletests(
    results_108_df["Raw p-value"],
    alpha=0.05,
    method="holm"
)

results_108_df["Holm-adjusted p-value"] = p_adjusted

results_108_df["Significant (α = 0.05)"] = reject


# ============================================================
# 11. Order columns
# ============================================================

results_108_df = results_108_df[
    [
        "Dimension",
        "N",
        "Weighted SFT Median",
        "DPO Median",
        "Median Δ (DPO - Weighted SFT)",
        "Wilcoxon W",
        "Raw p-value",
        "Holm-adjusted p-value",
        "Rank-biserial r",
        "Significant (α = 0.05)"
    ]
]


# ============================================================
# 12. Round numerical values
# ============================================================

numeric_columns = [
    "Weighted SFT Median",
    "DPO Median",
    "Median Δ (DPO - Weighted SFT)",
    "Wilcoxon W",
    "Raw p-value",
    "Holm-adjusted p-value",
    "Rank-biserial r"
]

results_108_df[numeric_columns] = (
    results_108_df[numeric_columns].round(4)
)


# ============================================================
# 13. Print final dissertation-ready table
# ============================================================

print("\n" + "=" * 110)
print("WILCOXON SIGNED-RANK TEST: WEIGHTED SFT vs AUGMENTED DPO")
print("=" * 110)

print(
    results_108_df.to_string(index=False)
)


# ============================================================
# 14. Interpretation
# ============================================================

print("\n" + "=" * 80)
print("INTERPRETATION")
print("=" * 80)

for _, row in results_108_df.iterrows():

    direction = (
        "DPO > Weighted SFT"
        if row["Rank-biserial r"] > 0
        else
        "DPO < Weighted SFT"
        if row["Rank-biserial r"] < 0
        else
        "No systematic difference"
    )

    significance = (
        "Statistically significant"
        if row["Significant (α = 0.05)"]
        else
        "Not statistically significant"
    )

    print(
        f"\n{row['Dimension']}: "
        f"{direction}; "
        f"{significance} "
        f"(Holm-adjusted p = "
        f"{row['Holm-adjusted p-value']:.4f}, "
        f"rank-biserial r = "
        f"{row['Rank-biserial r']:.4f})"
    )
```

```python
import numpy as np
import pandas as pd
from scipy.stats import wilcoxon
from statsmodels.stats.multitest import multipletests


# ============================================================
# 1. PREPARE WEIGHTED SFT — 108 CASES
# ============================================================

weighted = wght_dataframe[
    [
        "part",
        "index",
        "Factuality Score",
        "coverage",
        "Conciseness Score"
    ]
].copy()


# ============================================================
# 2. PREPARE AUGMENTED DPO — 108 CASES
# ============================================================

dpo = dpo_108[
    [
        "part",
        "index",
        "DPO_Factuality",
        "DPO_Coverage",
        "Conciseness Score"
    ]
].copy()


# ============================================================
# 3. STANDARDISE IDENTIFIERS
# ============================================================

weighted["part"] = weighted["part"].astype(str)
dpo["part"] = dpo["part"].astype(str)

weighted["index"] = weighted["index"].astype(int)
dpo["index"] = dpo["index"].astype(int)


# ============================================================
# 4. RENAME DPO COLUMNS
# ============================================================

dpo = dpo.rename(
    columns={
        "DPO_Factuality": "DPO_Factuality",
        "DPO_Coverage": "DPO_Coverage",
        "Conciseness Score": "DPO_Conciseness"
    }
)


# ============================================================
# 5. MATCH THE SAME 108 CASES
# ============================================================

comparison = weighted.merge(
    dpo,
    on=["part", "index"],
    how="inner"
)


print("=" * 100)
print("WEIGHTED SFT vs AUGMENTED DPO — 108 CASE COMPARISON")
print("=" * 100)

print(f"\nMatched cases: {len(comparison)}")

if len(comparison) != 108:
    print(
        "\nWARNING: Expected 108 matched cases, "
        f"but found {len(comparison)}."
    )


# ============================================================
# 6. DEFINE METRICS
# ============================================================

metrics = {

    "Factuality": (
        "Factuality Score",
        "DPO_Factuality"
    ),

    "Coverage": (
        "coverage",
        "DPO_Coverage"
    ),

    "Conciseness": (
        "Conciseness Score",
        "DPO_Conciseness"
    )
}


# ============================================================
# 7. DESCRIPTIVE STATISTICS
# ============================================================

descriptive_results = []

for dimension, (weighted_col, dpo_col) in metrics.items():

    weighted_scores = comparison[weighted_col].dropna()
    dpo_scores = comparison[dpo_col].dropna()

    descriptive_results.append({

        "Metric": dimension,

        "Weighted_SFT_N": len(weighted_scores),
        "DPO_N": len(dpo_scores),

        "Weighted_SFT_Mean": weighted_scores.mean(),
        "DPO_Mean": dpo_scores.mean(),

        "Weighted_SFT_SD": weighted_scores.std(),
        "DPO_SD": dpo_scores.std(),

        "Weighted_SFT_Median": weighted_scores.median(),
        "DPO_Median": dpo_scores.median(),

        "Weighted_SFT_Min": weighted_scores.min(),
        "DPO_Min": dpo_scores.min(),

        "Weighted_SFT_25%": weighted_scores.quantile(0.25),
        "DPO_25%": dpo_scores.quantile(0.25),

        "Weighted_SFT_75%": weighted_scores.quantile(0.75),
        "DPO_75%": dpo_scores.quantile(0.75),

        "Weighted_SFT_Max": weighted_scores.max(),
        "DPO_Max": dpo_scores.max()
    })


descriptive_df = pd.DataFrame(descriptive_results)


# ============================================================
# 8. IMPROVED / DETERIORATED / UNCHANGED
# ============================================================

change_results = []

for dimension, (weighted_col, dpo_col) in metrics.items():

    temp = comparison[
        [weighted_col, dpo_col]
    ].dropna()

    weighted_scores = temp[weighted_col]
    dpo_scores = temp[dpo_col]

    differences = dpo_scores - weighted_scores

    improved = (differences > 0).sum()
    deteriorated = (differences < 0).sum()
    unchanged = (differences == 0).sum()

    change_results.append({

        "Metric": dimension,

        "N": len(temp),

        "Improved": improved,

        "Deteriorated": deteriorated,

        "Unchanged": unchanged,

        "Improved_%": improved / len(temp) * 100,

        "Deteriorated_%": deteriorated / len(temp) * 100,

        "Unchanged_%": unchanged / len(temp) * 100
    })


change_df = pd.DataFrame(change_results)


# ============================================================
# 9. RANK-BISERIAL CORRELATION
# ============================================================

def rank_biserial_paired(
    weighted_scores,
    dpo_scores
):

    weighted_scores = np.asarray(weighted_scores)
    dpo_scores = np.asarray(dpo_scores)

    # DPO - Weighted SFT
    differences = dpo_scores - weighted_scores

    # Remove unchanged cases
    differences = differences[
        differences != 0
    ]

    if len(differences) == 0:
        return 0.0

    ranks = pd.Series(
        np.abs(differences)
    ).rank(
        method="average"
    ).to_numpy()

    positive_ranks = ranks[
        differences > 0
    ].sum()

    negative_ranks = ranks[
        differences < 0
    ].sum()

    return (
        positive_ranks - negative_ranks
    ) / (
        positive_ranks + negative_ranks
    )


# ============================================================
# 10. WILCOXON SIGNED-RANK TEST
# ============================================================

statistical_results = []

for dimension, (weighted_col, dpo_col) in metrics.items():

    temp = comparison[
        [weighted_col, dpo_col]
    ].dropna()

    weighted_scores = temp[weighted_col].to_numpy()
    dpo_scores = temp[dpo_col].to_numpy()

    differences = (
        dpo_scores - weighted_scores
    )

    # Wilcoxon test
    result = wilcoxon(
        dpo_scores,
        weighted_scores,
        alternative="two-sided",
        zero_method="wilcox",
        method="auto"
    )

    # Effect size
    rank_biserial = rank_biserial_paired(
        weighted_scores,
        dpo_scores
    )

    statistical_results.append({

        "Metric": dimension,

        "N": len(temp),

        "Weighted_SFT_Mean":
            np.mean(weighted_scores),

        "DPO_Mean":
            np.mean(dpo_scores),

        "Mean_Difference":
            np.mean(differences),

        "Weighted_SFT_Median":
            np.median(weighted_scores),

        "DPO_Median":
            np.median(dpo_scores),

        "Median_Difference":
            np.median(differences),

        "Wilcoxon_W":
            result.statistic,

        "Raw_p_value":
            result.pvalue,

        "Rank_biserial_r":
            rank_biserial
    })


stats_df = pd.DataFrame(statistical_results)


# ============================================================
# 11. HOLM MULTIPLE-COMPARISON CORRECTION
# ============================================================

reject, p_adjusted, _, _ = multipletests(
    stats_df["Raw_p_value"],
    alpha=0.05,
    method="holm"
)

stats_df["Holm_adjusted_p"] = p_adjusted

stats_df["Significant"] = reject


# ============================================================
# 12. ROUND RESULTS
# ============================================================

descriptive_df = descriptive_df.round(4)

change_df = change_df.round(2)

stats_df = stats_df.round(4)


# ============================================================
# 13. PRINT DESCRIPTIVE STATISTICS
# ============================================================

print("\n")
print("=" * 110)
print("DESCRIPTIVE STATISTICS — WEIGHTED SFT (108) vs AUGMENTED DPO (108)")
print("=" * 110)

print(
    descriptive_df.to_string(
        index=False
    )
)


# ============================================================
# 14. PRINT IMPROVED / DETERIORATED / UNCHANGED
# ============================================================

print("\n")
print("=" * 100)
print("IMPROVED / DETERIORATED / UNCHANGED — DPO RELATIVE TO WEIGHTED SFT")
print("=" * 100)

print(
    change_df.to_string(
        index=False
    )
)


# ============================================================
# 15. PRINT STATISTICAL TESTS
# ============================================================

print("\n")
print("=" * 120)
print("WILCOXON SIGNED-RANK TEST — WEIGHTED SFT vs AUGMENTED DPO")
print("=" * 120)

print(
    stats_df[
        [
            "Metric",
            "N",
            "Weighted_SFT_Mean",
            "DPO_Mean",
            "Mean_Difference",
            "Weighted_SFT_Median",
            "DPO_Median",
            "Median_Difference",
            "Wilcoxon_W",
            "Raw_p_value",
            "Holm_adjusted_p",
            "Rank_biserial_r",
            "Significant"
        ]
    ].to_string(
        index=False
    )
)


# ============================================================
# 16. INTERPRETATION
# ============================================================

print("\n")
print("=" * 100)
print("INTERPRETATION")
print("=" * 100)

for _, row in stats_df.iterrows():

    if row["Rank_biserial_r"] > 0:
        direction = "DPO > Weighted SFT"

    elif row["Rank_biserial_r"] < 0:
        direction = "DPO < Weighted SFT"

    else:
        direction = "No systematic difference"

    if row["Significant"]:
        significance = "Statistically significant"

    else:
        significance = "Not statistically significant"

    print(
        f"\n{row['Metric']}: "
        f"{direction}; "
        f"{significance}; "
        f"Holm-adjusted p = "
        f"{row['Holm_adjusted_p']:.4f}; "
        f"rank-biserial r = "
        f"{row['Rank_biserial_r']:.4f}"
    )
```

```python
import pandas as pd
import re


# ============================================================
# 1. NORMALISE REPORT FORMATTING
# ============================================================

def normalise_report(text):
    """
    Lightly normalise reports for teacher-student similarity analysis.

    Removes:
        - Numbered section markers: 1), 2), 10), etc.
        - Alphabetical subsection markers: a), b)

    Preserves:
        - Clinical content
        - Section headings
        - Report structure
        - Recommendations
    """

    # Handle missing values
    if pd.isna(text):
        return ""

    text = str(text)

    # --------------------------------------------------------
    # Remove numbered section markers
    # Examples: 1), 2), 10), 13)
    # --------------------------------------------------------
    text = re.sub(r'\b\d+\)\s*', '', text)

    # --------------------------------------------------------
    # Remove alphabetical subsection markers
    # Examples: a), b)
    # --------------------------------------------------------
    text = re.sub(r'\b[a-zA-Z]\)\s*', '', text)

    # --------------------------------------------------------
    # Normalise whitespace
    # --------------------------------------------------------
    text = re.sub(r'\s+', ' ', text).strip()

    return text


# ============================================================
# 2. CREATE NORMALISED TEACHER REPORTS
# ============================================================

df_teacher = df_teacer.copy()

df_teacer["Similarity_Report"] = (
    df_teacer["cleaned_report"]
    .apply(normalise_report)
)


# ============================================================
# 3. CREATE NORMALISED WEIGHTED SFT REPORTS
# ============================================================

wght_dataframe = wght_dataframe.copy()

wght_dataframe["Similarity_Report"] = (
    wght_dataframe["Cleaned_Report"]
    .apply(normalise_report)
)


# ============================================================
# 4. INSPECT EXAMPLE
# ============================================================

example_index = 4

print("=" * 100)
print("NORMALISED TEACHER REPORT")
print("=" * 100)

print(
    df_teacer["Similarity_Report"].iloc[example_index]
)


print("\n" + "=" * 100)
print("NORMALISED WEIGHTED SFT REPORT")
print("=" * 100)

print(
    wght_dataframe["Similarity_Report"].iloc[example_index]
)


# ============================================================
# 5. CHECK DATASET SIZE AND MISSING REPORTS
# ============================================================

print("\n" + "=" * 100)
print("DATASET CHECK")
print("=" * 100)

print(f"Teacher reports: {len(df_teacer)}")
print(f"Weighted SFT reports: {len(wght_dataframe)}")

print(
    f"\nMissing teacher reports: "
    f"{df_teacer['Similarity_Report'].eq('').sum()}"
)

print(
    f"Missing Weighted SFT reports: "
    f"{wght_dataframe['Similarity_Report'].eq('').sum()}"
)
```

```python
# ============================================================
# TEACHER ↔ WEIGHTED SFT KNOWLEDGE TRANSFER ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
from rouge_score import rouge_scorer
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
from bert_score import score as bert_score
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity


# ============================================================
# 1. PREPARE TEACHER DATA
# ============================================================

teacer = df_teacer[
    [
        "part",
        "index",
        "Similarity_Report"
    ]
].copy()

teacer = teacer.rename(
    columns={
        "Similarity_Report": "Teacher_Report"
    }
)


# ============================================================
# 2. PREPARE WEIGHTED SFT DATA
# ============================================================

weighted = wght_dataframe[
    [
        "part",
        "index",
        "Similarity_Report",
        "Factuality Score",
        "coverage",
        "Conciseness Score"
    ]
].copy()

weighted = weighted.rename(
    columns={
        "Similarity_Report": "Weighted_SFT_Report"
    }
)


# ============================================================
# 3. STANDARDISE IDENTIFIER TYPES
# ============================================================

teacer["part"] = teacer["part"].astype(str)
weighted["part"] = weighted["part"].astype(str)

teacer["index"] = teacer["index"].astype(int)
weighted["index"] = weighted["index"].astype(int)


# ============================================================
# 4. MATCH TEACHER ↔ WEIGHTED SFT
# ============================================================

comparison_transfer = pd.merge(
    teacer,
    weighted,
    on=["part", "index"],
    how="inner"
)


print("=" * 90)
print("TEACHER ↔ WEIGHTED SFT KNOWLEDGE TRANSFER ANALYSIS")
print("=" * 90)

print(f"\nTeacher reports:      {len(teacer)}")
print(f"Weighted SFT reports: {len(weighted)}")
print(f"Matched reports:      {len(comparison_transfer)}")


# ============================================================
# 5. REMOVE EMPTY REPORT PAIRS
# ============================================================

comparison_transfer = comparison_transfer[
    comparison_transfer["Teacher_Report"].notna()
    & comparison_transfer["Weighted_SFT_Report"].notna()
    & (comparison_transfer["Teacher_Report"].str.strip() != "")
    & (comparison_transfer["Weighted_SFT_Report"].str.strip() != "")
].copy()


print(f"Valid report pairs:   {len(comparison_transfer)}")


# ============================================================
# 6. ROUGE SCORES
# ============================================================

scorer = rouge_scorer.RougeScorer(
    ["rouge1", "rouge2", "rougeL"],
    use_stemmer=True
)

comparison_transfer["ROUGE-1"] = comparison_transfer.apply(
    lambda row: scorer.score(
        row["Teacher_Report"],
        row["Weighted_SFT_Report"]
    )["rouge1"].fmeasure,
    axis=1
)

comparison_transfer["ROUGE-2"] = comparison_transfer.apply(
    lambda row: scorer.score(
        row["Teacher_Report"],
        row["Weighted_SFT_Report"]
    )["rouge2"].fmeasure,
    axis=1
)

comparison_transfer["ROUGE-L"] = comparison_transfer.apply(
    lambda row: scorer.score(
        row["Teacher_Report"],
        row["Weighted_SFT_Report"]
    )["rougeL"].fmeasure,
    axis=1
)


# ============================================================
# 7. BLEU SCORE
# ============================================================

smooth = SmoothingFunction().method1

comparison_transfer["BLEU"] = comparison_transfer.apply(
    lambda row: sentence_bleu(
        [row["Teacher_Report"].split()],
        row["Weighted_SFT_Report"].split(),
        weights=(0.25, 0.25, 0.25, 0.25),
        smoothing_function=smooth
    ),
    axis=1
)


# ============================================================
# 8. BERTSCORE
# ============================================================

print("\nCalculating BERTScore...")

P, R, F1 = bert_score(
    cands=comparison_transfer["Weighted_SFT_Report"].tolist(),
    refs=comparison_transfer["Teacher_Report"].tolist(),
    lang="en",
    verbose=True
)

comparison_transfer["BERTScore_Precision"] = P.cpu().numpy()
comparison_transfer["BERTScore_Recall"] = R.cpu().numpy()
comparison_transfer["BERTScore_F1"] = F1.cpu().numpy()


# ============================================================
# 9. EMBEDDING COSINE SIMILARITY
# ============================================================

print("\nLoading embedding model...")

embedding_model = SentenceTransformer(
    "all-mpnet-base-v2"
)

print("Calculating embedding similarity...")

teacher_embeddings = embedding_model.encode(
    comparison_transfer["Teacher_Report"].tolist(),
    convert_to_numpy=True,
    show_progress_bar=True
)

student_embeddings = embedding_model.encode(
    comparison_transfer["Weighted_SFT_Report"].tolist(),
    convert_to_numpy=True,
    show_progress_bar=True
)


comparison_transfer["Embedding_Similarity"] = [
    cosine_similarity(
        teacher_embedding.reshape(1, -1),
        student_embedding.reshape(1, -1)
    )[0][0]
    
    for teacher_embedding, student_embedding
    in zip(
        teacher_embeddings,
        student_embeddings
    )
]


# ============================================================
# 10. DESCRIPTIVE STATISTICS
# ============================================================

metrics = [
    "ROUGE-1",
    "ROUGE-2",
    "ROUGE-L",
    "BLEU",
    "BERTScore_F1",
    "Embedding_Similarity"
]

summary = comparison_transfer[metrics].describe().T[
    ["count", "mean", "std", "min", "25%", "50%", "75%", "max"]
]

summary = summary.rename(
    columns={
        "count": "N",
        "mean": "Mean",
        "std": "SD",
        "min": "Min",
        "25%": "Q1",
        "50%": "Median",
        "75%": "Q3",
        "max": "Max"
    }
)

print("\n" + "=" * 90)
print("TEACHER ↔ WEIGHTED SFT SIMILARITY STATISTICS")
print("=" * 90)

print(summary.round(4))


# ============================================================
# 11. CHECK INDIVIDUAL EXAMPLE
# ============================================================

example = comparison_transfer.iloc[0]

print("\n" + "=" * 90)
print("EXAMPLE COMPARISON")
print("=" * 90)

print("\nTEACHER REPORT:\n")
print(example["Teacher_Report"])

print("\n" + "-" * 90)

print("\nWEIGHTED SFT REPORT:\n")
print(example["Weighted_SFT_Report"])

print("\n" + "-" * 90)

print("\nSIMILARITY SCORES:\n")

for metric in metrics:
    print(
        f"{metric}: "
        f"{example[metric]:.4f}"
    )


# ============================================================
# 12. SAVE RESULTS (OPTIONAL BUT RECOMMENDED)
# ============================================================

comparison_transfer.to_csv(
    "teacher_weighted_sft_similarity.csv",
    index=False
)

print("\nSaved: teacher_weighted_sft_similarity.csv")
```

```python
# ============================================================
# TEACHER ↔ WEIGHTED SFT KNOWLEDGE TRANSFER ANALYSIS
# ============================================================

import pandas as pd
import numpy as np

from rouge_score import rouge_scorer
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
from bert_score import score as bert_score
from sentence_transformers import SentenceTransformer


# ============================================================
# 1. PREPARE TEACHER DATA
# ============================================================

teacher = df_teacer[
    [
        "part",
        "index",
        "Similarity_Report"
    ]
].copy()

teacher = teacher.rename(
    columns={
        "Similarity_Report": "Teacher_Report"
    }
)


# ============================================================
# 2. PREPARE WEIGHTED SFT DATA
# ============================================================

weighted = wght_dataframe[
    [
        "part",
        "index",
        "Similarity_Report",
        "Factuality Score",
        "coverage",
        "Conciseness Score"
    ]
].copy()

weighted = weighted.rename(
    columns={
        "Similarity_Report": "Weighted_SFT_Report"
    }
)


# ============================================================
# 3. STANDARDISE IDENTIFIERS
# ============================================================

# Teacher part: 1, 2, 3...
teacher["part"] = teacher["part"].astype(int)

# Weighted SFT part: "01", "02", "03"... → 1, 2, 3...
weighted["part"] = weighted["part"].astype(int)

# Ensure index is integer in both datasets
teacher["index"] = teacher["index"].astype(int)
weighted["index"] = weighted["index"].astype(int)


# ============================================================
# 4. MATCH TEACHER ↔ WEIGHTED SFT
# ============================================================

comparison_transfer = pd.merge(
    teacher,
    weighted,
    on=["part", "index"],
    how="inner"
)


print("=" * 90)
print("TEACHER ↔ WEIGHTED SFT KNOWLEDGE TRANSFER ANALYSIS")
print("=" * 90)

print(f"\nTeacher reports:       {len(teacher)}")
print(f"Weighted SFT reports:  {len(weighted)}")
print(f"Matched reports:       {len(comparison_transfer)}")


# ============================================================
# 5. VALIDATE MATCHING
# ============================================================

if len(comparison_transfer) == 0:
    raise ValueError(
        "No reports matched. Check the 'part' and 'index' identifiers."
    )

if len(comparison_transfer) != len(teacher):
    print(
        "\nWARNING: Not all teacher reports were matched."
    )
else:
    print(
        "\n✓ All teacher reports successfully matched."
    )


# ============================================================
# 6. REMOVE EMPTY REPORT PAIRS
# ============================================================

comparison_transfer = comparison_transfer[
    comparison_transfer["Teacher_Report"].notna()
    & comparison_transfer["Weighted_SFT_Report"].notna()
].copy()


comparison_transfer["Teacher_Report"] = (
    comparison_transfer["Teacher_Report"]
    .astype(str)
    .str.strip()
)

comparison_transfer["Weighted_SFT_Report"] = (
    comparison_transfer["Weighted_SFT_Report"]
    .astype(str)
    .str.strip()
)


comparison_transfer = comparison_transfer[
    (comparison_transfer["Teacher_Report"] != "")
    & (comparison_transfer["Weighted_SFT_Report"] != "")
].copy()


print(f"Valid report pairs:    {len(comparison_transfer)}")


# ============================================================
# 7. ROUGE SCORES
# ============================================================

print("\nCalculating ROUGE scores...")

scorer = rouge_scorer.RougeScorer(
    ["rouge1", "rouge2", "rougeL"],
    use_stemmer=True
)


def calculate_rouge(row):

    scores = scorer.score(
        row["Teacher_Report"],
        row["Weighted_SFT_Report"]
    )

    return pd.Series({
        "ROUGE-1": scores["rouge1"].fmeasure,
        "ROUGE-2": scores["rouge2"].fmeasure,
        "ROUGE-L": scores["rougeL"].fmeasure
    })


rouge_results = comparison_transfer.apply(
    calculate_rouge,
    axis=1
)


comparison_transfer = pd.concat(
    [
        comparison_transfer,
        rouge_results
    ],
    axis=1
)


# ============================================================
# 8. BLEU SCORE
# ============================================================

print("Calculating BLEU scores...")

smooth = SmoothingFunction().method1


comparison_transfer["BLEU"] = comparison_transfer.apply(
    lambda row: sentence_bleu(
        [row["Teacher_Report"].split()],
        row["Weighted_SFT_Report"].split(),
        weights=(0.25, 0.25, 0.25, 0.25),
        smoothing_function=smooth
    ),
    axis=1
)


# ============================================================
# 9. BERTSCORE
# ============================================================

print("\nCalculating BERTScore...")

P, R, F1 = bert_score(
    cands=comparison_transfer[
        "Weighted_SFT_Report"
    ].tolist(),

    refs=comparison_transfer[
        "Teacher_Report"
    ].tolist(),

    lang="en",
    verbose=True
)


comparison_transfer["BERTScore_Precision"] = (
    P.cpu().numpy()
)

comparison_transfer["BERTScore_Recall"] = (
    R.cpu().numpy()
)

comparison_transfer["BERTScore_F1"] = (
    F1.cpu().numpy()
)


# ============================================================
# 10. EMBEDDING COSINE SIMILARITY
# ============================================================

print("\nLoading embedding model...")

embedding_model = SentenceTransformer(
    "all-mpnet-base-v2"
)


print("Calculating embedding similarity...")


teacher_embeddings = embedding_model.encode(
    comparison_transfer[
        "Teacher_Report"
    ].tolist(),

    convert_to_numpy=True,
    show_progress_bar=True,
    normalize_embeddings=True
)


student_embeddings = embedding_model.encode(
    comparison_transfer[
        "Weighted_SFT_Report"
    ].tolist(),

    convert_to_numpy=True,
    show_progress_bar=True,
    normalize_embeddings=True
)


# Because embeddings are normalised,
# cosine similarity = dot product

comparison_transfer[
    "Embedding_Similarity"
] = np.sum(
    teacher_embeddings * student_embeddings,
    axis=1
)


# ============================================================
# 11. DESCRIPTIVE STATISTICS
# ============================================================

metrics = [
    "ROUGE-1",
    "ROUGE-2",
    "ROUGE-L",
    "BLEU",
    "BERTScore_F1",
    "Embedding_Similarity"
]


summary = comparison_transfer[
    metrics
].describe().T[
    [
        "count",
        "mean",
        "std",
        "min",
        "25%",
        "50%",
        "75%",
        "max"
    ]
]


summary = summary.rename(
    columns={
        "count": "N",
        "mean": "Mean",
        "std": "SD",
        "min": "Min",
        "25%": "Q1",
        "50%": "Median",
        "75%": "Q3",
        "max": "Max"
    }
)


print("\n" + "=" * 90)
print("TEACHER ↔ WEIGHTED SFT SIMILARITY STATISTICS")
print("=" * 90)

print(
    summary.round(4)
)


# ============================================================
# 12. PRINT CLINICAL QUALITY STATISTICS
# ============================================================

quality_metrics = [
    "Factuality Score",
    "coverage",
    "Conciseness Score"
]


quality_summary = comparison_transfer[
    quality_metrics
].describe().T[
    [
        "count",
        "mean",
        "std",
        "min",
        "25%",
        "50%",
        "75%",
        "max"
    ]
]


quality_summary = quality_summary.rename(
    columns={
        "count": "N",
        "mean": "Mean",
        "std": "SD",
        "min": "Min",
        "25%": "Q1",
        "50%": "Median",
        "75%": "Q3",
        "max": "Max"
    }
)


print("\n" + "=" * 90)
print("WEIGHTED SFT CLINICAL QUALITY STATISTICS")
print("=" * 90)

print(
    quality_summary.round(4)
)


# ============================================================
# 13. EXAMPLE COMPARISON
# ============================================================

example = comparison_transfer.iloc[0]


print("\n" + "=" * 90)
print("EXAMPLE TEACHER ↔ WEIGHTED SFT COMPARISON")
print("=" * 90)


print("\nTEACHER REPORT:\n")
print(
    example["Teacher_Report"]
)


print("\n" + "-" * 90)


print("\nWEIGHTED SFT REPORT:\n")
print(
    example["Weighted_SFT_Report"]
)


print("\n" + "-" * 90)


print("\nSIMILARITY SCORES:\n")


for metric in metrics:

    print(
        f"{metric}: "
        f"{example[metric]:.4f}"
    )


# ============================================================
# 14. SAVE COMPLETE PER-CASE RESULTS
# ============================================================

comparison_transfer.to_csv(
    "teacher_weighted_sft_similarity.csv",
    index=False
)


# ============================================================
# 15. SAVE SUMMARY TABLE
# ============================================================

summary.to_csv(
    "teacher_weighted_sft_similarity_summary.csv"
)


print("\n" + "=" * 90)
print("ANALYSIS COMPLETE")
print("=" * 90)

print(
    "\nSaved:"
)

print(
    "1. teacher_weighted_sft_similarity.csv"
)

print(
    "2. teacher_weighted_sft_similarity_summary.csv"
)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# TEACHER ↔ WEIGHTED SFT
# CLINICAL-QUALITY DISTRIBUTION RETENTION
# ============================================================

# ============================================================
# 1. PREPARE DATA
# ============================================================

teacher = df_teacer[
    [
        "part",
        "index",
        "factuality_score",
        "coverage_score",
        "conciseness_score"
    ]
].copy()

weighted = wght_dataframe[
    [
        "part",
        "index",
        "Factuality Score",
        "coverage",
        "Conciseness Score"
    ]
].copy()


# ============================================================
# 2. STANDARDISE IDENTIFIERS
# ============================================================

teacher["part"] = (
    teacher["part"]
    .astype(str)
    .str.zfill(2)
)

weighted["part"] = (
    weighted["part"]
    .astype(str)
    .str.zfill(2)
)

teacher["index"] = teacher["index"].astype(int)
weighted["index"] = weighted["index"].astype(int)


# ============================================================
# 3. MATCH TEACHER AND WEIGHTED SFT CASES
# ============================================================

comparison_q2 = teacher.merge(
    weighted,
    on=["part", "index"],
    how="inner"
)


print("=" * 90)
print("TEACHER ↔ WEIGHTED SFT")
print("CLINICAL-QUALITY DISTRIBUTION RETENTION")
print("=" * 90)

print(f"\nTeacher reports:       {len(teacher)}")
print(f"Weighted SFT reports:  {len(weighted)}")
print(f"Matched reports:       {len(comparison_q2)}")


# ============================================================
# 4. EXTRACT MATCHED DISTRIBUTIONS
# ============================================================

teacher_factuality = (
    comparison_q2["factuality_score"]
    .dropna()
)

weighted_factuality = (
    comparison_q2["Factuality Score"]
    .dropna()
)

teacher_coverage = (
    comparison_q2["coverage_score"]
    .dropna()
)

weighted_coverage = (
    comparison_q2["coverage"]
    .dropna()
)

teacher_conciseness = (
    comparison_q2["conciseness_score"]
    .dropna()
)

weighted_conciseness = (
    comparison_q2["Conciseness Score"]
    .dropna()
)


# ============================================================
# 5. DESCRIPTIVE STATISTICS
# ============================================================

def descriptive_stats(series):
    return {
        "N": len(series),
        "Mean": series.mean(),
        "SD": series.std(),
        "Min": series.min(),
        "Q1": series.quantile(0.25),
        "Median": series.median(),
        "Q3": series.quantile(0.75),
        "Max": series.max()
    }


statistics = pd.DataFrame(
    {
        "Teacher_Factuality": descriptive_stats(
            teacher_factuality
        ),
        "Weighted_SFT_Factuality": descriptive_stats(
            weighted_factuality
        ),
        "Teacher_Coverage": descriptive_stats(
            teacher_coverage
        ),
        "Weighted_SFT_Coverage": descriptive_stats(
            weighted_coverage
        ),
        "Teacher_Conciseness": descriptive_stats(
            teacher_conciseness
        ),
        "Weighted_SFT_Conciseness": descriptive_stats(
            weighted_conciseness
        )
    }
).T


print("\n" + "=" * 90)
print("DESCRIPTIVE STATISTICS")
print("=" * 90)

print(
    statistics.round(4)
)


# ============================================================
# 6. CALCULATE DISTRIBUTION RETENTION DIFFERENCES
# ============================================================

retention_summary = pd.DataFrame(
    {
        "Dimension": [
            "Factuality",
            "Coverage",
            "Conciseness"
        ],
        "Teacher_Mean": [
            teacher_factuality.mean(),
            teacher_coverage.mean(),
            teacher_conciseness.mean()
        ],
        "Weighted_SFT_Mean": [
            weighted_factuality.mean(),
            weighted_coverage.mean(),
            weighted_conciseness.mean()
        ],
        "Mean_Difference": [
            weighted_factuality.mean()
            - teacher_factuality.mean(),

            weighted_coverage.mean()
            - teacher_coverage.mean(),

            weighted_conciseness.mean()
            - teacher_conciseness.mean()
        ],
        "Teacher_SD": [
            teacher_factuality.std(),
            teacher_coverage.std(),
            teacher_conciseness.std()
        ],
        "Weighted_SFT_SD": [
            weighted_factuality.std(),
            weighted_coverage.std(),
            weighted_conciseness.std()
        ]
    }
)


# Percentage of teacher mean retained
retention_summary["Mean_Retention_%"] = (
    retention_summary["Weighted_SFT_Mean"]
    / retention_summary["Teacher_Mean"]
    * 100
)


print("\n" + "=" * 90)
print("TEACHER QUALITY DISTRIBUTION RETENTION")
print("=" * 90)

print(
    retention_summary.round(4)
)


# ============================================================
# 7. CREATE 1 × 3 DISTRIBUTION FIGURE
# ============================================================

fig, axes = plt.subplots(
    1,
    3,
    figsize=(15, 5)
)


# ============================================================
# FACTUALITY
# ============================================================

sns.histplot(
    weighted_factuality,
    bins=20,
    kde=True,
    color="#4A4A4A",
    alpha=0.20,
    edgecolor="white",
    linewidth=0.8,
    line_kws={
        "color": "black",
        "linewidth": 2.2
    },
    label="Weighted SFT",
    ax=axes[0]
)

sns.histplot(
    teacher_factuality,
    bins=20,
    kde=True,
    color="#E6B800",
    alpha=0.20,
    edgecolor="white",
    linewidth=0.8,
    line_kws={
        "color": "#E6B800",
        "linewidth": 2.2
    },
    label="Teacher",
    ax=axes[0]
)

axes[0].set_xlabel(
    "Factuality Score",
    fontname="Times New Roman",
    fontsize=12
)

axes[0].set_ylabel(
    "Number of Reports",
    fontname="Times New Roman",
    fontsize=12
)

axes[0].set_title(
    "Factuality",
    fontname="Times New Roman",
    fontsize=14
)

axes[0].set_xlim(0, 1)


# ============================================================
# COVERAGE
# ============================================================

sns.histplot(
    weighted_coverage,
    bins=20,
    kde=True,
    color="#4A4A4A",
    alpha=0.20,
    edgecolor="white",
    linewidth=0.8,
    line_kws={
        "color": "black",
        "linewidth": 2.2
    },
    label="Weighted SFT",
    ax=axes[1]
)

sns.histplot(
    teacher_coverage,
    bins=20,
    kde=True,
    color="#E6B800",
    alpha=0.20,
    edgecolor="white",
    linewidth=0.8,
    line_kws={
        "color": "#E6B800",
        "linewidth": 2.2
    },
    label="Teacher",
    ax=axes[1]
)

axes[1].set_xlabel(
    "Coverage Score",
    fontname="Times New Roman",
    fontsize=12
)

axes[1].set_ylabel(
    "Number of Reports",
    fontname="Times New Roman",
    fontsize=12
)

axes[1].set_title(
    "Coverage",
    fontname="Times New Roman",
    fontsize=14
)

axes[1].set_xlim(0, 1)


# ============================================================
# CONCISENESS
# ============================================================

sns.histplot(
    weighted_conciseness,
    bins=20,
    kde=True,
    color="#4A4A4A",
    alpha=0.20,
    edgecolor="white",
    linewidth=0.8,
    line_kws={
        "color": "black",
        "linewidth": 2.2
    },
    label="Weighted SFT",
    ax=axes[2]
)

sns.histplot(
    teacher_conciseness,
    bins=20,
    kde=True,
    color="#E6B800",
    alpha=0.20,
    edgecolor="white",
    linewidth=0.8,
    line_kws={
        "color": "#E6B800",
        "linewidth": 2.2
    },
    label="Teacher",
    ax=axes[2]
)

axes[2].set_xlabel(
    "Conciseness Score",
    fontname="Times New Roman",
    fontsize=12
)

axes[2].set_ylabel(
    "Number of Reports",
    fontname="Times New Roman",
    fontsize=12
)

axes[2].set_title(
    "Conciseness",
    fontname="Times New Roman",
    fontsize=14
)

axes[2].set_xlim(0, 1)


# ============================================================
# 8. FORMATTING
# ============================================================

for ax in axes:

    ax.grid(False)

    for spine in ax.spines.values():
        spine.set_color("#D3D3D3")
        spine.set_linewidth(1.0)

    ax.tick_params(
        axis="both",
        which="both",
        length=0
    )

    for label in ax.get_xticklabels():
        label.set_fontname("Times New Roman")
        label.set_fontsize(11)

    for label in ax.get_yticklabels():
        label.set_fontname("Times New Roman")
        label.set_fontsize(11)

    ax.legend(
        prop={
            "family": "Times New Roman",
            "size": 11
        },
        frameon=False
    )


plt.tight_layout()

plt.savefig(
    "teacher_weighted_quality_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 9. SAVE STATISTICS
# ============================================================

statistics.to_csv(
    "teacher_weighted_quality_statistics.csv"
)

retention_summary.to_csv(
    "teacher_weighted_quality_retention.csv",
    index=False
)

print("\nSaved:")
print("1. teacher_weighted_quality_distribution.png")
print("2. teacher_weighted_quality_statistics.csv")
print("3. teacher_weighted_quality_retention.csv")
```

```python
import pandas as pd
from scipy.stats import spearmanr

# ============================================================
# Q3 — TEACHER SIMILARITY vs CLINICAL QUALITY
# ============================================================

similarity_metrics = [
    "ROUGE-1",
    "ROUGE-2",
    "ROUGE-L",
    "BLEU",
    "BERTScore_F1",
    "Embedding_Similarity"
]

quality_metrics = [
    "Factuality Score",
    "coverage",
    "Conciseness Score"
]

results = []

for similarity in similarity_metrics:

    for quality in quality_metrics:

        data = comparison_transfer[
            [similarity, quality]
        ].dropna()

        rho, p = spearmanr(
            data[similarity],
            data[quality]
        )

        results.append({
            "Similarity Metric": similarity,
            "Clinical Quality": quality,
            "N": len(data),
            "Spearman_rho": rho,
            "p_value": p
        })

correlation_results = pd.DataFrame(results)

# ============================================================
# HOLM CORRECTION
# ============================================================

from statsmodels.stats.multitest import multipletests

correlation_results["Holm_adjusted_p"] = multipletests(
    correlation_results["p_value"],
    method="holm"
)[1]

correlation_results["Significant"] = (
    correlation_results["Holm_adjusted_p"] < 0.05
)

# ============================================================
# PRINT RESULTS
# ============================================================

print("=" * 100)
print("TEACHER SIMILARITY vs CLINICAL QUALITY")
print("=" * 100)

print(
    correlation_results.round(4).to_string(index=False)
)

# ============================================================
# SAVE RESULTS
# ============================================================

correlation_results.to_csv(
    "teacher_similarity_clinical_quality_correlations.csv",
    index=False
)

print("\nSaved:")
print("teacher_similarity_clinical_quality_correlations.csv")
```
